# EDA

Exploratory Data Analysis (EDA) is the process of analyzing and summarizing a dataset to understand its key characteristics before applying any machine learning model or statistical technique. It involves:
- ✔️ Summarizing Data – Getting an overview of distributions, missing values, and relationships between variables.
- ✔️ Data Cleaning – Handling missing values, duplicates, and outliers.
- ✔️ Visualizing Data – Using plots and charts to uncover patterns and trends.
- ✔️ Feature Engineering – Creating new variables or transforming existing ones for better insights.

**Key Questions Answered in EDA:**
- What is the distribution of the data?
- Are there missing values or outliers?
- How are variables related to each other?
- Does the data need transformation or feature engineering?
