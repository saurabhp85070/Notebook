# Mean
The mean, also known as the arithmetic average, is a measure of central tendency that represents the "typical" value of a dataset.

**🔹 What It Tells Us:**
- The mean gives an overall sense of the dataset's central value.
- Sensitive to outliers → A single extreme value can significantly impact the mean.
- Used when data is normally distributed or has a symmetric distribution, is not ideal for skewed distributions..

**Example:**
Income data: If the average salary in a company is ₹50,000, but a few executives earn ₹5,00,000, the mean will be inflated and not represent most employees.

## When and Where Mean is Useful in Data Science
1. Exploratory Data Analysis (EDA)
    - Mean helps in summarizing the dataset.
    - It provides an initial understanding of the distribution of numerical data.

2. Comparing Different Groups
    - If you have multiple groups (e.g., average income by city, average test scores by school), mean is a useful metric to compare them.

3. Feature Engineering
    - Mean is often used to impute missing values (e.g., replacing missing values in a dataset with the column mean).

4. Anomaly Detection
    - A value far from the mean may indicate an outlier (e.g., fraud detection in banking).

5. Performance Metrics in Machine Learning
    - The mean squared error (MSE) in regression is calculated based on the mean of squared differences between predicted and actual values.



# Median
The median is another measure of central tendency that represents the middle value of a dataset when arranged in ascending order. Unlike the mean, the median is not affected by outliers, making it more robust for skewed distributions.
                                                                                                                                                                                                                
**What It Tells Us:**
- Not affected by outliers → A better measure of center for skewed data.
- Used when data has extreme values or is not normally distributed.

🔹 Example:
House prices: If most houses are ₹50 lakh, but a few are ₹10 crore, the median price gives a more realistic estimate of the typical house price.

```python
import numpy as np
import statistics
import pandas as pd

np.median(data) 
statistics.median(data)
df['col'].median() 
```

## When and Where the Median is Useful in Data Science
1. Skewed Data Distributions
    - The median is preferred when the data is skewed because it is not affected by extreme values.

    - Example: In income data, a few billionaires can inflate the mean, but the median remains representative of the majority.

2. Outlier-Resistant Analysis
    - If a dataset contains outliers, the median provides a more stable central value.

    - Example: In house prices, a few luxury mansions won’t distort the median price as much as the mean.

3. Comparing Groups Fairly
    - Median helps compare groups where outliers could be misleading.

    - Example: If we compare salaries of engineers in two different companies, the median salary will be a better metric than the mean.

4. Data Imputation (Handling Missing Values)
    - Median is often used to fill missing values instead of mean, especially in skewed datasets.

5. Machine Learning Models
    - The median is used in decision trees and robust regression models that handle skewed data better.

# Variance
Variance measures how spread out the data points are from the mean (average).

- High variance → Data points are widely spread (more variability).

- Low variance → Data points are closely packed around the mean (less variability).

## What Does Variance Tell Us About the Data?
- 1️⃣ If variance is close to zero, data points are very close to the mean (little variation).
- 2️⃣ If variance is large, data points are spread out (high variation in data).
- 3️⃣ Variance helps compare two datasets to understand which one has more variability.

📌 Example:

- Dataset A (low variance): [50, 51, 50, 49, 50]

- Dataset B (high variance): [10, 90, 30, 70, 50]

    - Both datasets have the same mean (50), but Dataset B is more spread out.

## Where Is Variance Useful in Data Science?

### 1️⃣ Feature Selection in Machine Learning
- Features with very low variance (almost constant values) are not useful in predictions.

- Variance helps remove redundant features to improve model performance.

📌 Example:
If a dataset has a feature like "Country" with almost all values as "India," it has very low variance and is not useful for prediction.

✅ Use Case: Helps in feature selection and dimensionality reduction.

### 2️⃣ Identifying Data Stability
- In financial markets, low variance in stock prices means stability, while high variance means high risk.

- Variance helps measure risk in investment portfolios.

📌 Example:

- Stock A: Daily prices fluctuate between $99 and $101 (low variance, stable).

- Stock B: Prices fluctuate between $50 and $150 (high variance, risky).

✅ Use Case: Helps in risk assessment and volatility analysis.

### 3️⃣ Performance Analysis in A/B Testing
- In an A/B test, if one variant has higher variance, the results might not be consistent.

- A lower variance means the results are more reliable.

📌 Example:

- Variant A: Conversion rates = [5%, 6%, 5.5%, 6%, 5.2%] → Low variance, reliable results.

- Variant B: Conversion rates = [2%, 10%, 3%, 12%, 1%] → High variance, inconsistent results.

✅ Use Case: Helps in reliable decision-making in experiments.

### 4️⃣ Quality Control in Manufacturing
- If variance in product weight is too high, some products may be defective.

- Low variance ensures consistency in product quality.

📌 Example:

- Product A: Weight varies between 99g to 101g (low variance, consistent).

- Product B: Weight varies between 90g to 110g (high variance, inconsistent).

✅ Use Case: Helps in manufacturing process optimization.

```python
import numpy as np

data = [10, 12, 23, 23, 16, 23, 21, 16]
variance = np.var(data)  # By default, N (population variance)
print("Variance:", variance)
```

```python
import statistics

data = [10, 12, 23, 23, 16, 23, 21, 16]
variance = statistics.variance(data)  # Uses (N-1) for sample variance
print("Sample Variance:", variance)
```

Key Difference:

- np.var(data) → Uses N (population variance).

- statistics.variance(data) → Uses N-1 (sample variance, Bessel’s correction).

```python
import matplotlib.pyplot as plt
import seaborn as sns

# Generate two datasets
low_variance = np.random.normal(50, 5, 1000)  # Mean=50, Std=5
high_variance = np.random.normal(50, 20, 1000) # Mean=50, Std=20

# Plot distributions
plt.figure(figsize=(10, 5))
sns.kdeplot(low_variance, label="Low Variance", fill=True)
sns.kdeplot(high_variance, label="High Variance", fill=True)
plt.legend()
plt.title("Comparison of Low and High Variance Distributions")
plt.show()

# Narrow peak → Low variance (data is concentrated).
# Wider spread → High variance (data is more dispersed).
```

# Standard deviation 
Standard deviation (σ for population, s for sample) measures how much data deviates from the mean. It is simply the square root of variance

## What Does Standard Deviation Tell Us?
- 🔹 Low standard deviation (σ ≈ 0) → Data points are close to the mean (low variability).
- 🔹 High standard deviation (σ ≫ 0) → Data points are spread out from the mean (high variability).
- 🔹 Zero standard deviation (σ = 0) → All values in the dataset are identical.

📌 Example:

- Dataset A (Low Std Dev): [50, 51, 50, 49, 50] → σ small, data is stable.

- Dataset B (High Std Dev): [10, 90, 30, 70, 50] → σ large, data is highly variable.



## When & Where Is Standard Deviation Useful in Data Science?

### 1️⃣ Understanding Data Distribution
Standard deviation helps determine if data is normally distributed.

In a normal distribution:

- 68% of data falls within 1σ of the mean.

- 95% of data falls within 2σ of the mean.

- 99.7% of data falls within 3σ of the mean.

📌 Use Case:
Used in statistical modeling & hypothesis testing to check how data is distributed.

### 2️⃣ Risk Measurement in Finance
- A stock with a high standard deviation means high volatility (risky investment).

- A stock with a low standard deviation means stable returns (low risk).

📌 Example:

- Stock A (σ = 2%) → Stable, predictable returns.

- Stock B (σ = 15%) → High fluctuations, risky investment.

✅ Use Case: Portfolio risk assessment.

### 3️⃣ Quality Control in Manufacturing
- If the weight of a product has high standard deviation, it indicates inconsistent quality.

- A low standard deviation means the manufacturing process is precise.

📌 Example:

- Low σ: Product weight = [99g, 100g, 101g] → Consistent quality.

- High σ: Product weight = [90g, 110g, 85g] → Poor quality control.

✅ Use Case: Ensures product quality consistency.

### 4️⃣ Model Evaluation in Machine Learning
- Standard deviation helps assess the stability of model accuracy.

- If cross-validation accuracies have low standard deviation, the model is consistent.

- If they have high standard deviation, the model performs inconsistently.

📌 Example:

- Low σ (good model): Accuracies = [85%, 86%, 84%, 85%]

- High σ (unstable model): Accuracies = [60%, 90%, 50%, 95%]

✅ Use Case: Helps in model selection & hyperparameter tuning.

```python
import numpy as np

data = [10, 12, 23, 23, 16, 23, 21, 16]
std_dev = np.std(data)  # Population standard deviation
print("Standard Deviation:", std_dev)

# If std_dev ≈ 0, data is tightly packed around the mean.
```

```python
import statistics

data = [10, 12, 23, 23, 16, 23, 21, 16]
std_dev = statistics.stdev(data)  # Sample standard deviation (N-1)
print("Sample Standard Deviation:", std_dev)
```

ifference:

- np.std(data) → Population standard deviation (N)

- statistics.stdev(data) → Sample standard deviation (N-1)



```python
import matplotlib.pyplot as plt
import seaborn as sns

# Generate two datasets
low_std_dev = np.random.normal(50, 5, 1000)  # Mean=50, Std=5
high_std_dev = np.random.normal(50, 20, 1000) # Mean=50, Std=20

# Plot distributions
plt.figure(figsize=(10, 5))
sns.kdeplot(low_std_dev, label="Low Standard Deviation", fill=True)
sns.kdeplot(high_std_dev, label="High Standard Deviation", fill=True)
plt.legend()
plt.title("Comparison of Low and High Standard Deviation")
plt.show()

# Narrow peak → Low std dev (less spread).
# Wider spread → High std dev (more spread).
```

# Skewness
Skewness measures the asymmetry of a distribution around its mean. It tells us whether the data is symmetrically distributed or leans more toward one side.

## Types of Skewness

### - 1️⃣ Symmetric (Zero Skewness)
- Mean ≈ Median ≈ Mode

- The data is normally distributed (bell-shaped).

Example: Heights of adults, IQ scores.

📌 Use Case:
✅ Valid for parametric statistical tests like t-tests & regression models.

### 2️⃣ Positive Skew (Right-Skewed)
- Mean > Median > Mode

- The right tail is longer (more extreme values on the right).

Example: Income distribution, house prices (most people earn less, few earn a lot).

📌 Use Case:
✅ Used in financial risk analysis (stock market returns are often right-skewed).

### 3️⃣ Negative Skew (Left-Skewed)
- Mean < Median < Mode

- The left tail is longer (more extreme values on the left).

Example: Age of death, exam scores (if the test is too easy, most students score high).

📌 Use Case:
✅ Used in quality control (detecting anomalies in manufacturing defect rates).



## Why Is Skewness Important in Data Science?

### 1️⃣ Determining the Right Statistical Test

- Normally distributed data → Use parametric tests (t-tests, ANOVA, regression).

- Skewed data → Use non-parametric tests (Mann-Whitney U, Kruskal-Wallis).

### 2️⃣ Improving Machine Learning Models

- Many ML models (like linear regression) assume normality.

- Log transformation or Box-Cox transformation can fix right-skewed data.

### 3️⃣ Risk Assessment in Finance

- Stock returns are often right-skewed (few stocks give huge profits).

- Insurance claims tend to be right-skewed (few large claims, many small ones).

### 4️⃣ Detecting Anomalies

- Left-skewed income data might indicate fraud (unusual high expenses).

- Right-skewed response times in websites → Some users face high latency.



```python
import numpy as np
from scipy.stats import skew

data = [10, 12, 23, 23, 16, 23, 21, 16, 25, 30]
skewness = skew(data)
print("Skewness:", skewness)


# skewness > 0 → Right-skewed
# skewness < 0 → Left-skewed
# skewness ≈ 0 → Symmetric
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

# Generate data
right_skewed = np.random.exponential(scale=2, size=1000)
left_skewed = -np.random.exponential(scale=2, size=1000)
symmetric = np.random.normal(0, 1, 1000)

# Plot distributions
plt.figure(figsize=(12, 5))

plt.subplot(1, 3, 1)
sns.histplot(left_skewed, kde=True, bins=30)
plt.title("Left-Skewed Distribution")

plt.subplot(1, 3, 2)
sns.histplot(symmetric, kde=True, bins=30)
plt.title("Symmetric Distribution")

plt.subplot(1, 3, 3)
sns.histplot(right_skewed, kde=True, bins=30)
plt.title("Right-Skewed Distribution")

plt.tight_layout()
plt.show()


# Left-skewed: More data is concentrated on the right.
# Right-skewed: More data is concentrated on the left.
# Symmetric: Balanced on both sides.
```

# Kurtosis
Kurtosis measures the "tailedness" of a distribution—how heavy or light the tails are compared to a normal distribution. It tells us whether a dataset has more or fewer extreme values than a normal distribution.

## Types of Kurtosis
### 1️⃣ Mesokurtic (Kurtosis ≈ 3)
- Normal distribution (bell-shaped).

- Moderate tails, no extreme outliers.

Example: IQ scores, heights of adults.

📌 Use Case:
✅ Used as a reference for comparing other distributions.

### 2️⃣ Leptokurtic (Kurtosis > 3)
- Heavy-tailed distribution (more extreme values).

- More outliers than a normal distribution.

Example: Stock market crashes, financial returns.

📌 Use Case:
✅ Risk analysis (detecting extreme financial losses or profits).
✅ Fraud detection (unusual patterns in transactions).

### 3️⃣ Platykurtic (Kurtosis < 3)
- Light-tailed distribution (fewer extreme values).

- Data is more evenly spread out.

Example: Uniform distribution (test scores in an easy exam).

📌 Use Case:
✅ Used in manufacturing to ensure consistent product quality.

## Why Is Kurtosis Important in Data Science?

### 1️⃣ Outlier Detection

- High kurtosis → More extreme values → Possible anomalies or fraud.

- Low kurtosis → Data is evenly distributed, less concern for outliers.

### 2️⃣ Financial Risk Management

- Leptokurtic distributions → Market crashes, stock volatility.

- Used in Value-at-Risk (VaR) models.

### 3️⃣ Image Processing

- Detecting sharp edges in images using kurtosis in feature extraction.

### 4️⃣ Machine Learning

- High kurtosis in features might indicate outliers that affect model performance.



```python
import numpy as np
from scipy.stats import kurtosis

data = [10, 12, 23, 23, 16, 23, 21, 16, 25, 30]
kurt = kurtosis(data)
print("Kurtosis:", kurt)


# kurtosis ≈ 0 → Mesokurtic
# kurtosis > 0 → Leptokurtic (heavy tails)
# kurtosis < 0 → Platykurtic (light tails)
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

# Generate data
normal_data = np.random.normal(0, 1, 1000)  # Mesokurtic
heavy_tailed = np.random.laplace(0, 1, 1000)  # Leptokurtic
light_tailed = np.random.uniform(-2, 2, 1000)  # Platykurtic

# Plot distributions
plt.figure(figsize=(12, 5))

plt.subplot(1, 3, 1)
sns.histplot(normal_data, kde=True, bins=30)
plt.title("Mesokurtic (Normal Distribution)")

plt.subplot(1, 3, 2)
sns.histplot(heavy_tailed, kde=True, bins=30)
plt.title("Leptokurtic (Heavy Tails)")

plt.subplot(1, 3, 3)
sns.histplot(light_tailed, kde=True, bins=30)
plt.title("Platykurtic (Light Tails)")

plt.tight_layout()
plt.show()


# Mesokurtic → Moderate tails (normal distribution).
# Leptokurtic → High peaks and long tails (more extreme values).
# Platykurtic → Flat, fewer extreme values (more uniform spread).
```

# Data distribution based on mode of data
When analyzing a dataset, we often look for the mode (most frequently occurring value). Depending on how many modes exist, we classify the distribution as unimodal, bimodal, or multimodal.

## 1️⃣ Unimodal Distribution
A unimodal distribution has one peak (one mode), meaning there is a single most frequent value in the dataset.

### Real-World Examples:

- Heights of people in a country (most people are of average height).
- Exam scores where most students score around the average.
- Daily temperatures in a city (e.g., most days have a common temperature range).

```python
import matplotlib.pyplot as plt
import seaborn as sns

# Sample Unimodal Data
data = [3, 6, 8, 9, 9, 9, 12, 14, 18]

# Plot Histogram
sns.histplot(data, bins=5, kde=True)
plt.title("Unimodal Distribution")
plt.show()
```

## 2️⃣ Bimodal Distribution
A bimodal distribution has two peaks, meaning there are two distinct values that occur most frequently.

### Real-World Examples:

- Exam scores where two groups of students perform differently (e.g., high vs. low performers).

- Customer ages in a shopping mall (e.g., kids and senior citizens visit frequently).

- Salaries in a company with junior and senior employees.

```python
import numpy as np

# Sample Bimodal Data
data1 = np.random.normal(30, 5, 1000)  # First peak (around 30)
data2 = np.random.normal(70, 5, 1000)  # Second peak (around 70)
bimodal_data = np.concatenate([data1, data2])

# Plot Histogram
sns.histplot(bimodal_data, bins=30, kde=True)
plt.title("Bimodal Distribution")
plt.show()
```

## 3️⃣ Multimodal Distribution
A multimodal distribution has three or more peaks, meaning there are multiple frequently occurring values.

### Real-World Examples:

- Customer preferences for different product categories.

- Test scores where students belong to multiple performance groups.

- Travel patterns where different age groups prefer different modes of transport.

```python
# Sample Multimodal Data
data1 = np.random.normal(20, 3, 500)  # Peak 1
data2 = np.random.normal(50, 3, 500)  # Peak 2
data3 = np.random.normal(80, 3, 500)  # Peak 3
multimodal_data = np.concatenate([data1, data2, data3])

# Plot Histogram
sns.histplot(multimodal_data, bins=30, kde=True)
plt.title("Multimodal Distribution")
plt.show()
```

## Why Is It Important to Identify Unimodal, Bimodal, or Multimodal Distributions?
Understanding whether a dataset is unimodal, bimodal, or multimodal helps in various areas of data analysis, machine learning, and decision-making. Here’s how and where this information is useful:

### 1️⃣ Exploratory Data Analysis (EDA)
- Detecting patterns in data: Knowing the distribution helps understand the natural structure of the data.

- Identifying multiple groups: If the data is bimodal or multimodal, it suggests that multiple subpopulations exist within the dataset.

📌 Example:
If you analyze customer ages at a shopping mall and find a bimodal distribution, it might indicate that two main age groups frequently visit, such as teenagers and senior citizens.

- ✅ Use Case: Helps in customer segmentation for targeted marketing.

### 2️⃣ Choosing the Right Statistical or Machine Learning Model
- Many statistical tests (e.g., t-tests, ANOVA) assume a normal (unimodal) distribution.

- If the data is bimodal or multimodal, applying these tests without transformation may lead to incorrect conclusions.

📌 Example:

If salaries in a company have a bimodal distribution (junior vs. senior employees), using a simple linear regression model might not work well. Instead, you may need to separate the two groups or use a more complex model.

✅ Use Case: Helps in choosing the correct modeling approach.

### 3️⃣ Clustering & Segmentation in Machine Learning
- If the data is bimodal or multimodal, it indicates that there are multiple clusters.

- Clustering techniques (like K-Means, Gaussian Mixture Models) become useful to separate different groups.

📌 Example:
If a company wants to classify customer behavior, but the data is bimodal, it suggests that there are two different types of customers. Instead of applying one model to all customers, clustering can group them for better analysis.

✅ Use Case: Helps in customer segmentation, recommendation systems, fraud detection.

### 4️⃣ Detecting Anomalies and Outliers
- If the data is supposed to be unimodal but appears bimodal or multimodal, it might indicate errors, anomalies, or multiple data sources.

- This is useful in fraud detection, medical diagnoses, or sensor data analysis.

📌 Example:

- A unimodal distribution is expected in a manufacturing process (e.g., bottle-filling measurements).

- If the data shows a bimodal distribution, it may suggest a machine fault that is producing two different levels of output.

✅ Use Case: Helps in anomaly detection and quality control.

### 5️⃣ Data Transformation for Normalization
- Many machine learning algorithms assume normality (e.g., linear regression, logistic regression).

- If the data is bimodal or multimodal, applying log transformation or kernel density estimation (KDE) can make it more suitable for modeling.

📌 Example:
If housing prices show a multimodal distribution due to different property types (apartments, villas, townhouses), a log transformation might be required before using a model.

✅ Use Case: Helps in feature engineering and preprocessing.
