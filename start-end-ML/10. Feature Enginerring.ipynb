Feature Engineering is the process of transforming raw data into meaningful features that improve the performance of machine learning models. It involves:
- ✅ Creating new features
- ✅ Modifying existing features
- ✅ Handling missing values, outliers, and categorical data
- ✅ Selecting the most relevant features

Example
Imagine you're building a model to predict house prices. Your dataset includes:

- 🏠 Size in sq ft

- 🏠 Number of bedrooms

- 🏠 Location

Feature engineering can help by:
- ✔️ Creating a price per square foot feature
- ✔️ Converting categorical location into numerical form
- ✔️ Generating interaction features like "bedrooms per square foot"

**📌 Why is Feature Engineering Important?**
- 🔹 Boosts Model Performance – Well-engineered features can lead to higher accuracy.
- 🔹 Reduces Model Complexity – Helps simplify the model while maintaining good predictions.
- 🔹 Improves Interpretability – Well-crafted features make models easier to understand.

🚀 Real-World Example:
A credit card fraud detection system may use:
- ✔️ Transaction Amount
- ✔️ Transaction Location
- ✔️ Time of Transaction

Using Feature Engineering, you can create:
- ✅ Time Difference from Previous Transaction (to detect suspiciously frequent transactions)
- ✅ Distance Between Transactions (to catch geographically impossible transactions)



**📌 Steps in Feature Engineering**
- 1️⃣ Understand the Data – Identify feature types (numerical, categorical, time-series, text, images, etc.)
- 2️⃣ Handle Missing Data – Use imputation techniques like mean, median, or predictive models
- 3️⃣ Remove Outliers – Detect using Z-score, IQR, or visualization
- 4️⃣ Transform Features – Normalize, scale, or encode categorical data
- 5️⃣ Create New Features – Generate interaction features, polynomial features, and domain-specific features
- 6️⃣ Reduce Dimensionality – Use PCA, t-SNE, or feature selection techniques

# Feature Understanding & Exploration
Before applying feature engineering techniques, it’s crucial to understand the dataset. This step helps in identifying issues and opportunities for improving data quality.

## 1. Types of Data and Features
Understanding the kind of data you're working with is essential. Data can broadly be classified as:

1. Numerical Data:

    - Continuous: Real numbers (e.g., height, weight, temperature).- 
    - Discrete: Countable numbers (e.g., number of visits, number of products purchased).

2. Categorical Data:

    - Nominal: Categories without inherent order (e.g., color, gender).
    - Ordinal: Categories with an order (e.g., rankings, education level: High School < Bachelor’s < Master’s).

3. Text Data: Unstructured data like reviews, comments, or documents.

4. Time Series Data: Data points indexed in time order (e.g., stock prices, weather data).

5. Image/Data from Other Modalities: Specialized data like images or signals that require different preprocessing techniques.

## 2. Data Quality and Integrity
Before engineering features, you must assess the quality of your data:

- Missing Values:

    - Identify if there are any missing values in the dataset.

    - Understand their patterns (random vs. systematic missingness).

- Outliers:

    - Look for values that deviate significantly from others.

    - Use statistical methods (Z-score, IQR) or visualizations (boxplots) to detect them.

- Data Consistency:

    - Check for duplicate records or inconsistent data formats.

    - Validate if the ranges and distributions make sense.

#### Pearson Correlation (r)

- Pearson correlation measures the linear relationship between two continuous numerical variables.

- It assumes that the data is normally distributed.

- The correlation coefficient (r) ranges from -1 to +1

**When to Use Pearson Correlation?**
- ✅ When both variables are continuous (e.g., height vs. weight).
- ✅ When the relationship between the variables is linear.
- ✅ When both variables are normally distributed.
- ✅ When there are no significant outliers.

```python
import pandas as pd
import numpy as np
from scipy.stats import pearsonr

# Sample data
data = {'X': [10, 20, 30, 40, 50], 'Y': [12, 24, 36, 48, 60]}
df = pd.DataFrame(data)

# Compute Pearson correlation
pearson_corr, p_value = pearsonr(df['X'], df['Y'])
print(f"Pearson Correlation: {pearson_corr:.3f}, P-value: {p_value:.3f}")
```

#### Spearman Rank Correlation (ρ)

- Spearman correlation measures the monotonic relationship between two variables.

- It does not assume a linear relationship but instead focuses on the rank of the values.

- The coefficient (ρ) ranges from -1 to +1, similar to Pearson.

- It works well for ordinal data (ordered categories) or non-normally distributed data.

**When to Use Spearman Correlation?**
- ✅ When variables have a monotonic relationship (not necessarily linear).
- ✅ When one or both variables are ordinal (e.g., education level vs. job satisfaction).
- ✅ When the data contains outliers, as ranks reduce their influence.
- ✅ When the data is not normally distributed.

```python
from scipy.stats import spearmanr

# Compute Spearman correlation
spearman_corr, p_value = spearmanr(df['X'], df['Y'])
print(f"Spearman Correlation: {spearman_corr:.3f}, P-value: {p_value:.3f}")
```

#### Kendall Rank Correlation (τ)

- Kendall correlation is similar to Spearman but is based on concordant and discordant pairs rather than rank differences.

- It measures the strength of association between two ordinal or continuous variables.

- The coefficient (τ) ranges from -1 to +1.

**When to Use Kendall Correlation?**
- ✅ When working with small datasets.
- ✅ When you need a more robust alternative to Spearman.
- ✅ When the data contains many tied ranks (e.g., customer ratings: 5 stars, 4 stars, etc.).
- ✅ When analyzing ordinal data.

```python
from scipy.stats import kendalltau

# Compute Kendall correlation
kendall_corr, p_value = kendalltau(df['X'], df['Y'])
print(f"Kendall Correlation: {kendall_corr:.3f}, P-value: {p_value:.3f}")
```

```python
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from scipy.stats import spearmanr, pearsonr

# Set random seed for reproducibility
np.random.seed(42)

# Generating example datasets
x = np.linspace(1, 10, 100)

# 1. Linear Increasing (Strong Positive Linear & Monotonic)
y1 = x + np.random.normal(scale=0.5, size=len(x))

# 2. Non-Linear Monotonic (Spearman should be high, Pearson lower)
y2 = np.log(x) + np.random.normal(scale=0.2, size=len(x))

# 3. Non-Monotonic Relationship (Pearson & Spearman should be low)
y3 = np.sin(x) + np.random.normal(scale=0.1, size=len(x))

# Compute correlations
pearson1, _ = pearsonr(x, y1)
spearman1, _ = spearmanr(x, y1)

pearson2, _ = pearsonr(x, y2)
spearman2, _ = spearmanr(x, y2)

pearson3, _ = pearsonr(x, y3)
spearman3, _ = spearmanr(x, y3)

# Plot
fig, axes = plt.subplots(1, 3, figsize=(18, 5))

# 1. Linear Increasing (Pearson & Spearman High)
sns.scatterplot(x=x, y=y1, ax=axes[0])
axes[0].set_title(f"Linear Monotonic (Pearson: {pearson1:.2f}, Spearman: {spearman1:.2f})")

# 2. Logarithmic Monotonic (Spearman High, Pearson Lower)
sns.scatterplot(x=x, y=y2, ax=axes[1])
axes[1].set_title(f"Non-Linear Monotonic (Pearson: {pearson2:.2f}, Spearman: {spearman2:.2f})")

# 3. Non-Monotonic (Pearson & Spearman Low)
sns.scatterplot(x=x, y=y3, ax=axes[2])
axes[2].set_title(f"Non-Monotonic (Pearson: {pearson3:.2f}, Spearman: {spearman3:.2f})")

plt.show()
```
