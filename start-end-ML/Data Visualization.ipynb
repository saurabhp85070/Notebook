# Univariate Analysis
Univariate analysis is the simplest form of data analysis where we examine and visualize one variable at a time. The primary goal is to understand the distribution, central tendency, dispersion, and outliers of a single variable.

It is crucial in data science because it helps us:

- Identify patterns and trends in data.

- Detect anomalies or outliers.

- Decide the appropriate transformations or modeling techniques.

**Types of Univariate Analysis**
There are two main types of data:

1. Numerical (Continuous & Discrete) – e.g., age, salary, height.
2. Categorical (Nominal & Ordinal) – e.g., gender, color, education level.

## 1. Univariate Analysis for Numerical Data
For numerical variables, we focus on:

- Measures of Central Tendency (Mean, Median, Mode)

- Measures of Dispersion (Variance, Standard Deviation, Range, IQR)

- Distribution Shape (Skewness, Kurtosis)

### Histogram

- Represents the frequency distribution of numerical data.

- Helps in understanding the spread and skewness of the data.

```python
import matplotlib.pyplot as plt
import seaborn as sns

data = [25, 26, 28, 30, 32, 35, 38, 40, 42, 45, 50, 55, 60]
sns.histplot(data, bins=8, kde=True)  # KDE shows density estimation
plt.title('Histogram of Age')
plt.xlabel('Age')
plt.ylabel('Frequency')
plt.show()
```

### Box Plot (Box-and-Whisker Plot)

- Summarizes the minimum, Q1, median, Q3, and maximum values.

- Identifies outliers using IQR.

```python
sns.boxplot(data=data)
plt.title('Box Plot of Age')
plt.show()
```

### Density Plot (KDE Plot)

- A smoothed version of a histogram.

- Helps in visualizing the probability density function.



```python
sns.kdeplot(data, fill=True)
plt.title('Density Plot')
plt.show()
```

### Violin Plot

- Combines a box plot and KDE.

- Useful for comparing distributions.

```python
sns.violinplot(data=data)
plt.title('Violin Plot')
plt.show()
```

## 2. Univariate Analysis for Categorical Data
For categorical variables, we focus on:

- Mode (most frequent category)

- Count/Frequency of each category

### Bar Chart

- Represents the count of each category.

```python
categories = ['Red', 'Blue', 'Green', 'Red', 'Green', 'Blue', 'Red']
sns.countplot(x=categories)
plt.title('Category Frequency')
plt.show()
```

### Pie Chart

- Shows the proportion of each category.

```python
import pandas as pd
labels = ['Red', 'Blue', 'Green']
values = [3, 2, 2]
plt.pie(values, labels=labels, autopct='%1.1f%%', colors=['red', 'blue', 'green'])
plt.title('Category Distribution')
plt.show()
```

# Bivariate Analysis 
Bivariate analysis examines the relationship between two variables. This is crucial in data science because it helps us identify correlations, dependencies, and trends between different factors in a dataset.

**Types of Bivariate Analysis**
Bivariate analysis is categorized based on the types of variables:



**Why is Bivariate Analysis Important?**
- Helps understand cause-and-effect relationships.

- Determines the strength and direction of association between two variables.

- Useful for predictive modeling and feature selection in machine learning.

- Aids in hypothesis testing (e.g., do higher temperatures increase sales?).

## 1️⃣ Bivariate Analysis for Two Numerical Variables

- This type of analysis explores the relationship between two continuous variables.
- Typically measured using correlation and regression analysis.



Example 1: Analyzing Relationship Between House Price & Size

- 📌 Dataset: house_prices.csv
- 📌 Variables: Price (Numerical) & Size (sq. ft.) (Numerical)

```python
# Scatter Plot (Visualizing Relationships)

# Load dataset
df = pd.read_csv("house_prices.csv")

# Scatter plot
sns.scatterplot(x=df['Size'], y=df['Price'])
plt.title("House Price vs Size")
plt.xlabel("Size (sq. ft.)")
plt.ylabel("Price ($)")
plt.show()

# If the points show an upward trend, larger houses tend to be more expensive.
# If there's no clear pattern, price may depend on other factors (e.g., location).
```

```python
# Correlation Analysis (Strength of Relationship)

corr = df[['Size', 'Price']].corr()
print(corr)

# Correlation > 0.7 → Strong positive correlation.
# Correlation < -0.7 → Strong negative correlation.
# Correlation ≈ 0 → No significant relationship.
```

**Example**

```python
import numpy as np

# Generate dummy data
np.random.seed(42)
size = np.random.randint(500, 5000, 100)  # House sizes (sq. ft.)
price = size * np.random.uniform(200, 400) + np.random.randint(5000, 50000, 100)  # Price formula

# Create DataFrame
df_house = pd.DataFrame({'Size': size, 'Price': price})

# Scatter Plot
plt.figure(figsize=(8, 5))
sns.scatterplot(x=df_house['Size'], y=df_house['Price'])
plt.title("House Price vs Size")
plt.xlabel("Size (sq. ft.)")
plt.ylabel("Price ($)")
plt.show()
```

```python
# Correlation
df_house.corr()
```

## 3️⃣ Bivariate Analysis for Two Categorical Variables

- This type of analysis examines the relationship between two categorical variables.
- Typically measured using contingency tables and chi-square tests.



📌 Example:
Relationship Between Education Level and Employment Type
- Education Level (High School, Bachelor's, Master's)

- Employment Type (Unemployed, Part-time, Full-time)



Example 3: Customer Satisfaction Based on Payment Method

- 📌 Dataset: customer_feedback.csv
- 📌 Variables: Satisfaction Level (Categorical: Low, Medium, High) & Payment Method (Categorical: Credit, Debit, PayPal)

```python
# Contingency Table (Cross-tabulation)

pd.crosstab(df['Payment Method'], df['Satisfaction Level'])

# If one method (e.g., "Credit Card") has a higher count of "High Satisfaction," it might be preferred by customers.
```

```python
# Stacked Bar Chart

pd.crosstab(df['Payment Method'], df['Satisfaction Level']).plot(kind='bar', stacked=True)
plt.title("Customer Satisfaction by Payment Method")
plt.xlabel("Payment Method")
plt.ylabel("Count")
plt.show()

# Helps visualize which payment method has the highest satisfaction levels.
```

```python
# Chi-Square Test (Checking for Dependency)

from scipy.stats import chi2_contingency

table = pd.crosstab(df['Payment Method'], df['Satisfaction Level'])
chi2, p, dof, expected = chi2_contingency(table)

print(f"Chi-Square Statistic: {chi2}, P-value: {p}")

# If p-value < 0.05, there is a significant relationship between payment method and satisfaction.
# If p-value > 0.05, they are independent.
```

**Examples**

```python
# Generate dummy data
np.random.seed(42)
payment_methods = ["Credit Card", "Debit Card", "PayPal"]
satisfaction_levels = ["Low", "Medium", "High"]

# Simulated categorical data
data = {
    "Payment Method": np.random.choice(payment_methods, 100),
    "Satisfaction Level": np.random.choice(satisfaction_levels, 100, p=[0.2, 0.4, 0.4])
}

df_feedback = pd.DataFrame(data)

# Contingency Table
contingency_table = pd.crosstab(df_feedback['Payment Method'], df_feedback['Satisfaction Level'])
print(contingency_table)

```

```python
# Stacked Bar Chart
contingency_table.plot(kind='bar', stacked=True, figsize=(8, 5), colormap="viridis")
plt.title("Customer Satisfaction by Payment Method")
plt.xlabel("Payment Method")
plt.ylabel("Count")
plt.show()

```

```python
from scipy.stats import chi2_contingency

# Chi-Square Test
chi2, p, dof, expected = chi2_contingency(contingency_table)
print(f"Chi-Square Statistic: {chi2}, P-value: {p}")
```

# Multivariate Analysis
Multivariate analysis refers to statistical techniques used to analyze datasets with more than two variables simultaneously. It helps identify patterns, relationships, and dependencies among multiple variables, allowing for deeper insights into the data.

**Why is Multivariate Analysis Important?**
- Real-world data is multidimensional (e.g., customer behavior, medical diagnostics, and stock market trends involve multiple factors).

- It helps in dimensionality reduction and feature selection for machine learning models.

- Provides a comprehensive view of relationships between variables.

- Identifies hidden structures and dependencies in data.

**Types of Multivariate Analysis**
Multivariate analysis can be divided into two main types:

1. Dependence Methods – When one or more dependent variables depend on multiple independent variables.

2. Interdependence Methods – When no explicit dependent variable exists, and the goal is to explore relationships among all variables.



## Key Techniques in Multivariate Analysis

### 1️⃣ Multiple Linear Regression (MLR)
📌 Definition: A technique used to predict the value of one dependent variable using multiple independent variables.

📌 Example Use Case:
Predicting house prices based on features like size, number of bedrooms, and location.

### 2️⃣ Principal Component Analysis (PCA)
📌 Definition:
PCA is a dimensionality reduction technique that transforms high-dimensional data into a lower-dimensional space while preserving as much variance as possible.

📌 How It Works:
1. Standardize the dataset (mean = 0, variance = 1).

2. Compute the covariance matrix.

3. Find the eigenvalues and eigenvectors.

4. Select the top principal components that explain the most variance.

📌 Example Use Case:
Reducing 100 features in an image recognition dataset to the top 10 principal components for easier processing.

### 3️⃣ Cluster Analysis (Clustering)
📌 Definition:
A technique used to group similar observations together based on shared characteristics.

📌 Common Clustering Methods:
- K-Means Clustering (divides data into k groups)

- Hierarchical Clustering (builds a tree of clusters)

- DBSCAN (identifies clusters based on density)

📌 Example Use Case:
Segmenting customers into different groups based on their purchasing behavior.

### 4️⃣ Factor Analysis
📌 Definition: 
A technique used to identify hidden underlying factors that explain the relationships among observed variables.

📌 Example Use Case:
Analyzing student performance based on different subjects to determine if there are latent skills (e.g., problem-solving, analytical thinking).

### 5️⃣ Canonical Correlation Analysis (CCA)
📌 Definition:
A method used to analyze the relationship between two sets of variables.

📌 Example Use Case:
Studying the relationship between marketing strategies (social media ads, TV ads, online campaigns) and sales performance (revenue, customer retention, conversion rate).

## Multivariate Visualization Techniques
Since multivariate analysis involves multiple variables, visualization is key to understanding relationships.



### 1. Scatter Plot with Hue (3 Numerical Variables)

Example: House Price vs Size, Colored by Number of Bedrooms

```python
# Generate dummy data
np.random.seed(42)
size = np.random.randint(500, 5000, 100)  # House sizes (sq. ft.)
bedrooms = np.random.randint(1, 6, 100)  # Number of bedrooms
price = size * np.random.uniform(200, 400) + (bedrooms * 10000) + np.random.randint(5000, 50000, 100)

# Create DataFrame
df_house = pd.DataFrame({'Size': size, 'Price': price, 'Bedrooms': bedrooms})

# Scatter Plot with Hue
plt.figure(figsize=(8, 5))
sns.scatterplot(x=df_house['Size'], y=df_house['Price'], hue=df_house['Bedrooms'], palette="viridis", size=df_house['Bedrooms'])
plt.title("House Price vs Size (Colored by Bedrooms)")
plt.xlabel("Size (sq. ft.)")
plt.ylabel("Price ($)")
plt.show()

# Price generally increases with size.
# More bedrooms tend to push the price higher.
```

### 2. Pairplot (Multiple Numerical Variables)
Example: Relationships Between Size, Price, and Number of Bedrooms

```python
sns.pairplot(df_house, hue="Bedrooms", palette="coolwarm")
plt.show()

# Pairwise relationships across all numerical features.
# Correlations can be visually observed.
```

### 3. Heatmap (Correlation Between Multiple Variables)

```python
plt.figure(figsize=(6, 4))
sns.heatmap(df_house.corr(), annot=True, cmap="coolwarm", linewidths=0.5)
plt.title("Feature Correlation Heatmap")
plt.show()

# Strong correlation between Size & Price.
# Bedrooms also impact price but less strongly.
```

###  4. Grouped Bar Chart (Categorical & Numerical Data)
Example: Average Salary by Job Role and Experience Level

```python
# Dummy data
np.random.seed(42)
roles = ["Data Scientist", "Software Engineer", "Manager", "Analyst"]
experience = ["Junior", "Mid", "Senior"]
data = {
    "Job Role": np.random.choice(roles, 100),
    "Experience": np.random.choice(experience, 100),
    "Salary": np.random.randint(50000, 180000, 100),
}

df_salary = pd.DataFrame(data)

# Bar plot
plt.figure(figsize=(8, 5))
sns.barplot(x="Job Role", y="Salary", hue="Experience", data=df_salary, errorbar=None)
plt.xticks(rotation=45)
plt.title("Average Salary by Job Role and Experience")
plt.show()

# Senior roles tend to have higher salaries.
# Managers earn more across all experience levels.
```

###  4. Bubble chart
Example: Analyzing Sales Performance

Let's consider a dataset where we analyze Sales Performance based on:

- Revenue ($) → X-axis

- Profit ($) → Y-axis

- Number of Units Sold → Bubble Size

- Region → Bubble Color

```python
# Sample dataset with 10 records
np.random.seed(42)  # For reproducibility
data = pd.DataFrame({
    "Revenue ($)": np.random.randint(10000, 50000, 10),  # Sales revenue
    "Profit ($)": np.random.randint(2000, 10000, 10),  # Profit earned
    "Units Sold": np.random.randint(50, 500, 10),  # Number of units sold
    "Region": np.random.choice(["North", "South", "East", "West"], 10)  # Region
})

# Define color palette for regions
palette = {"North": "blue", "South": "green", "East": "red", "West": "orange"}

# Plot Bubble Chart
plt.figure(figsize=(10, 6))
sns.scatterplot(
    x=data["Revenue ($)"],
    y=data["Profit ($)"],
    size=data["Units Sold"],
    hue=data["Region"],
    sizes=(50, 1000),  # Min and max bubble sizes
    palette=palette,
    edgecolor="black",
    alpha=0.7  # Transparency
)

# Add labels and title
plt.xlabel("Revenue ($)")
plt.ylabel("Profit ($)")
plt.title("Multivariate Analysis: Sales Performance")
plt.legend(title="Region")
plt.grid(True)
plt.show()


# Higher Revenue & Higher Profit: Larger bubbles in the upper-right indicate top-performing regions.
# Lower Revenue & Lower Profit: Smaller bubbles in the lower-left indicate underperforming regions.
# Region-wise Comparison: Color coding helps compare regions effectively.
# Impact of Units Sold: Bigger bubbles represent regions where more units were sold.
```

## Challenges in Multivariate Analysis
1. Multicollinearity – When independent variables are highly correlated.

    - Solution: Use Variance Inflation Factor (VIF) to detect it.

2. Curse of Dimensionality – When too many features reduce model performance.

    - Solution: Use PCA to reduce dimensions.

3. Overfitting – When the model learns noise instead of patterns.

    - Solution: Use regularization (Lasso, Ridge).
