# List comprehension

List comprehension in Python is an easy and compact syntax for creating a list from a string or another list. It is a very concise way to create a new list by performing an operation on each item in the existing list. List comprehension is considerably faster than processing a list using the for loop.

**Syntax:** `[expression for element in iterable if condition]`

```python
even_nums = [x for x in range(21) if x%2 == 0]
print(even_nums)
```

```python
# List comprehension works with string lists also

names = ['Steve', 'Bill', 'Ram', 'Mohan', 'Abdul']
names2 = [s for s in names if 'a' in s]
print(names2)
```

```python
# list comprehension to build a list of squares of the numbers between 1 and 10

squares = [x*x for x in range(11)] 
print(squares) 
```

```python
# List Comprehension using Nested Loops

nums1 = [1, 2, 3]
nums2 = [4, 5, 6]
nums=[(x,y) for x in nums1 for y in nums2]
print(nums)
```

```python
# List Comprehension with Multiple if Conditions

nums = [x for x in range(21) if x%2==0 if x%3==0] 
print(nums)
```

```python
# List Comprehension with if-else Condition

odd_even_list = ["Even" if i%2==0 else "Odd" for i in range(5)]
print(odd_even_list)
odd_even_list = [str(i) + '=Even' if i%2==0 else str(i) + "=Odd" for i in range(5)]
print(odd_even_list)
```

```python
a=[i for i in range(20) if i%2==0]
a
```

```python
# Flatten List using List Comprehension

matrix=[[1, 2, 3], [4, 5, 6], [7, 8, 9]]
flatList=[num for row in matrix for num in row]
print(flatList)
```

**Note:**

List comprehensions in Python do not support the usage of `elif` statements. List comprehensions are primarily designed for simple if and else conditions. If you need more complex logic involving `elif` statements, it's recommended to use regular for loops and conditional statements instead.

# Dictionary Comprehension

It allows you to create dictionaries using a similar concise syntax as list comprehension.
Here's the basic syntax for dictionary comprehension:

**Syntax:** `new_dict = {key_expression: value_expression for item in iterable if condition}`

```python
squares_dict = {x: x ** 2 for x in range(1, 6)}
print(squares_dict)
```

```python
# Creating a dictionary from a list

names = ['Alice', 'Bob', 'Charlie']
name_lengths = {name: len(name) for name in names}
print(name_lengths)
```

```python
# Filtering elements based on a condition

numbers = [1, 2, 3, 4, 5, 6]
even_squares = {num: num**2 for num in numbers if num % 2 == 0}
print(even_squares)
```

```python
# Creating a dictionary from two lists

keys = ['a', 'b', 'c']
values = [1, 2, 3]
dictionary = {key: value for key, value in zip(keys, values)}
print(dictionary)
```

```python
# Modifying values based on a condition

numbers = {'a': 1, 'b': 2, 'c': 3, 'd': 4}
numbers_squared = {key: value**2 if value % 2 == 0 else value for key, value in numbers.items()}
print(numbers_squared)
```

# Name mangling

1. **Private attributes:** In Python, attributes with a double underscore prefix (`__attribute`) are subject to name mangling, which makes them harder (but not impossible) to access from outside the class. However, they're not truly "private" in the sense that other languages enforce privacy - they can still be accessed with modified names.

2. **Protected attributes:** In Python, attributes with a single underscore prefix (`_attribute`) are considered "protected" by convention. This is just a naming convention that signals to other developers that the attribute is intended for internal use, but there's no enforcement mechanism preventing external access.

## `__attribute`

- Using a double underscore prefix (`__attr`) triggers name mangling.

- Python changes `__attr` internally to `_ClassName__attr` to make it harder to access from outside the class.

```python
class MyClass:
    def __init__(self):
        self.__private = "secret"

obj = MyClass()
print(obj.__private)          # ❌ Error

# print(obj._MyClass__private)    # ✅ Access via name mangling
```

```python
class MyClass:
    def __init__(self):
        self.__private = "secret"

obj = MyClass()
obj.__private = "not secret"

print(obj._MyClass__private)    # ✅ Access via name mangling
```

This does not raise an error because you're creating a new attribute `__private` on the `obj` instance — it's not the same as the private attribute defined in the class.

**Whyy?** Because of name mangling, Python internally renames `self.__private` to `self._MyClass__private`. 

But when you do this `obj.__private = "not secret"`, You're assigning a new attribute literally called `__private` (not mangled) to the instance `obj`. So now your object has two different attributes:

- `obj._MyClass__private` → from the constructor
- `obj.__private` → created externally, not mangled 

You can check this:

```python
print(obj.__dict__)
```

## `_attribute`

A single underscore prefix (`_attr`) indicates that the attribute is protected:

- Intended to be used only within the class or its subclasses.
- But it's not enforced by Python — just a convention.

```python
class Parent:
    def __init__(self):
        self._protected = "for subclass"

class Child(Parent):
    def show(self):
        print(self._protected)  # ✅ Access allowed

obj = Child()
obj.show()
print(obj._protected)  # ⚠️ Not recommended outside class, but allowed
```

```python
class BankAccount:
    def __init__(self, owner, balance):
        self.owner = owner            # Public attribute - accessible anywhere
        self._balance = balance       # Protected attribute - accessible but conventionally "internal use only"
        self.__account_number = "123456789"  # Private attribute - name mangling applied

    def deposit(self, amount):
        self._balance += amount
        return self._balance
    
    def get_account_details(self):
        return f"Owner: {self.owner}, Balance: {self._balance}, Account: {self.__account_number}"


# Creating an instance
account = BankAccount("Alice", 1000)

# Accessing different attributes
print(account.owner)  # Public attribute - works fine
print(account._balance)  # Protected attribute - works but not recommended (convention only)

# This will raise an AttributeError - name has been mangled
try:
    print(account.__account_number)  
except AttributeError as e:
    print(f"Error: {e}")

# But we can still access the private attribute if we know how name mangling works
print(account._BankAccount__account_number)  # This works - accessing the mangled name
```

# `@property` Decorator

Python programming provides us with a built-in `@property` decorator which makes usage of getter and setters much easier in Object-Oriented Programming.

Before going into details on what`@property` decorator is, let us first build an intuition on why it would be needed in the first place.

**Class Without Getters and Setters**

Let us assume that we decide to make a class that stores the temperature in degrees Celsius. And, it would also implement a method to convert the temperature into degrees Fahrenheit.

One way of doing this is as follows:

```python
class Celsius:
    def __init__(self, temperature = 0):
        self.temperature = temperature

    def to_fahrenheit(self):
        return (self.temperature * 1.8) + 32
```

We can make objects out of this class and manipulate the `temperature` attribute as we wish:

```python
# Basic method of setting and getting attributes in Python
class Celsius:
    def __init__(self, temperature=0):
        self.temperature = temperature

    def to_fahrenheit(self):
        return (self.temperature * 1.8) + 32


# Create a new object
human = Celsius()

# Set the temperature
human.temperature = 37

# Get the temperature attribute
print(human.temperature)

# Get the to_fahrenheit method
print(human.to_fahrenheit())
```

Here, the extra decimal places when converting into Fahrenheit is due to the Floating Point Arithmetic Error.

So, whenever we assign or retrieve any object attribute like `temperature` as shown above, Python searches it in the object's built-in `__dict__` dictionary attribute as

```python
print(human.__dict__) 
# Output: {'temperature': 37}
```

Therefore, `human.temperature` internally becomes `human.__dict__['temperature']`.

**Using Getters and Setters**


Suppose we want to extend the usability of the Celsius class defined above. We know that the temperature of any object cannot reach below -273.15 degrees Celsius.

Let's update our code to implement this value constraint.

An obvious solution to the above restriction will be to hide the attribute temperature (make it private) and define new getter and setter methods to manipulate it.

This can be done as follows:

```python
# Making Getters and Setter methods
class Celsius:
    def __init__(self, temperature=0):
        self.set_temperature(temperature)

    def to_fahrenheit(self):
        return (self.get_temperature() * 1.8) + 32

    # getter method
    def get_temperature(self):
        return self._temperature

    # setter method
    def set_temperature(self, value):
        if value < -273.15:
            raise ValueError("Temperature below -273.15 is not possible.")
        self._temperature = value
```

As we can see, the above method introduces two new `get_temperature()` and `set_temperature()` methods.

Furthermore, temperature was replaced with `_temperature`. An underscore `_` at the beginning is used to denote **private variables** in Python.

Now, let's use this implementation:

```python
# Making Getters and Setter methods
class Celsius:
    def __init__(self, temperature=0):
        self.set_temperature(temperature)

    def to_fahrenheit(self):
        return (self.get_temperature() * 1.8) + 32

    # getter method
    def get_temperature(self):
        return self._temperature

    # setter method
    def set_temperature(self, value):
        if value < -273.15:
            raise ValueError("Temperature below -273.15 is not possible.")
        self._temperature = value


# Create a new object, set_temperature() internally called by __init__
human = Celsius(37)

# Get the temperature attribute via a getter
print(human.get_temperature())

# Get the to_fahrenheit method, get_temperature() called by the method itself
print(human.to_fahrenheit())

# new constraint implementation
human.set_temperature(-300)

# Get the to_fahreheit method
print(human.to_fahrenheit())
```

This update successfully implemented the new restriction. We are no longer allowed to set the `temperature` below -273.15 degrees Celsius.

**Note:** The private variables don't actually exist in Python. There are simply norms to be followed. The language itself doesn't apply any restrictions.

However, the bigger problem with the above update is that all the programs that implemented our previous class have to modify their code from `obj.temperature` to `obj.get_temperature()` and all expressions like` obj.temperature = val` to `obj.set_temperature(val)`.

This refactoring can cause problems while dealing with hundreds of thousands of lines of codes.

All in all, our new update was not backwards compatible. This is where `@property` comes to rescue.

**The property Class**

A pythonic way to deal with the above problem is to use the property class. Here is how we can update our code:

```python
# using property class
class Celsius:
    def __init__(self, temperature=0):
        self.temperature = temperature

    def to_fahrenheit(self):
        return (self.temperature * 1.8) + 32

    # getter
    def get_temperature(self):
        print("Getting value...")
        return self._temperature

    # setter
    def set_temperature(self, value):
        print("Setting value...")
        if value < -273.15:
            raise ValueError("Temperature below -273.15 is not possible")
        self._temperature = value

    # creating a property object
    temperature = property(get_temperature, set_temperature)
```

We added the `print()` function inside `get_temperature()` and `set_temperature()` to clearly observe that they are being executed.

The last line of the code makes a property object `temperature`. Simply put, property attaches some code (`get_temperature` and `set_temperature`) to the member attribute accesses (`temperature`).

Let's use this update code:

```python
# using property class
class Celsius:
    def __init__(self, temperature=0):
        self.temperature = temperature

    def to_fahrenheit(self):
        return (self.temperature * 1.8) + 32

    # getter
    def get_temperature(self):
        print("Getting value...")
        return self._temperature

    # setter
    def set_temperature(self, value):
        print("Setting value...")
        if value < -273.15:
            raise ValueError("Temperature below -273.15 is not possible")
        self._temperature = value

    # creating a property object
    temperature = property(get_temperature, set_temperature)


human = Celsius(37)

print(human.temperature)

print(human.to_fahrenheit())

human.temperature = -300
```

As we can see, any code that retrieves the value of `temperature` will automatically call `get_temperature()` instead of a dictionary (`__dict__`) look-up.

Similarly, any code that assigns a value to `temperature` will automatically call `set_temperature()`.

We can even see above that `set_temperature()` was called even when we created an object.



```python
human = Celsius(37) # prints Setting value...
```

**Can you guess why?**

The reason is that when an object is created, the `__init__()` method gets called. This method has the line `self.temperature = temperature`. This expression automatically calls `set_temperature()`.

Similarly, any access like `c.temperature` automatically calls `get_temperature()`. This is what property does.

By using property, we can see that no modification is required in the implementation of the value constraint. Thus, our implementation is backward compatible.

**Note:** The actual temperature value is stored in the private _temperature variable. The temperature attribute is a property object which provides an interface to this private variable.

**The `@property` Decorator**

In Python, `property()` is a built-in function that creates and returns a property object. The syntax of this function is:

`property(fget=None, fset=None, fdel=None, doc=None)`

Here,

* `fget` is function to get value of the attribute
* `fset` is function to set value of the attribute
* `fdel` is function to delete the attribute
* `doc` is a string (like a comment)

As seen from the implementation, these function arguments are optional.

A property object has three methods, `getter()`, `setter()`, and `deleter()` to specify `fget`, `fset` and `fdel` at a later point. This means, the line:

`temperature = property(get_temperature,set_temperature)`

can be broken down as:

    # make empty property
    temperature = property()

    # assign fget
    temperature = temperature.getter(get_temperature)

    # assign fset
    temperature = temperature.setter(set_temperature)

These two pieces of code are equivalent.

Programmers familiar with Python Decorators can recognize that the above construct can be implemented as decorators.

We can even not define the names `get_temperature` and `set_temperature` as they are unnecessary and pollute the class namespace.

For this, we reuse the temperature name while defining our `getter` and `setter` functions. Let's look at how to implement this as a decorator:

```python
# Using @property decorator
class Celsius:
    def __init__(self, temperature=0):
        self.temperature = temperature

    def to_fahrenheit(self):
        return (self.temperature * 1.8) + 32

    @property
    def temperature(self):
        print("Getting value...")
        return self._temperature

    @temperature.setter
    def temperature(self, value):
        print("Setting value...")
        if value < -273.15:
            raise ValueError("Temperature below -273 is not possible")
        self._temperature = value


# create an object
human = Celsius(37)

print(human.temperature)

print(human.to_fahrenheit())

coldest_thing = Celsius(-300)
```

The above implementation is simple and efficient. It is the recommended way to use `property`.

```python
from dataclasses import dataclass

@dataclass
class User:
    name: str
    age: int
```

This is equivalent to:

```python
class User:
    def __init__(self, name: str, age: int) -> None:
        self.name = name
        self.age = age

    def __repr__(self) -> str:
        return f"User(name={self.name}, age={self.age})"

    def __eq__(self, other) -> bool:
        return self.name == other.name and self.age == other.age
```

```python
u1 = User("Alice", 30)
print(u1)  # User(name='Alice', age=30)

u2 = User("Alice", 30)
print(u1 == u2)  # True (compares values, not memory addresses)
```

# Walrus Operator `:=`

It allows you to assign a value to a variable as part of an expression. The `official name is the assignment expression`, but everyone calls it the Walrus operator because it looks like a walrus face := 🦭

**Syntax:** `variable := expression`

It both:

1. Assigns the result of `expression` to `variable`, and
2. Returns that value (so you can use it in the same line)


- `:=` (walrus)	Assign + use value inline

```python
# Without Walrus operator:

value = input("Enter something: ")
if len(value) > 5:
    print(f"You entered: {value}")
```

```python
# With Walrus operator:

if (value := input("Enter something: ")).strip() != "":
    print(f"You entered: {value}")
```

```python
data = input("Say something: ")
while data != "quit":
    print(f"You said: {data}")
    data = input("Say something: ")
```

```python
while (data := input("Say something: ")) != "quit":
    print(f"You said: {data}")
```

# Match case

The match statement in Python is a feature introduced in PEP 634 (Python Enhancement Proposal) and added in Python 3.10. It provides a way to perform pattern matching on the values of an expression. Pattern matching is a mechanism for checking a value against a set of patterns and executing code based on the matched pattern. It simplifies conditional branching and is a more expressive and concise alternative to the traditional if, elif, and else statements.

Here's an example of how the match statement works:

```python
match expression:
    case pattern1:
        # Code to execute when expression matches pattern1
    case pattern2:
        # Code to execute when expression matches pattern2
    case pattern3:
        # Code to execute when expression matches pattern3
    case _:
        # Code to execute when no pattern matches (similar to "else" in an "if" statement)
```

```python
def get_day_message(day: str) -> str:
    match day:
        case "Monday":
            return "Start of the week!"
        case "Friday":
            return "Almost weekend!"
        case "Saturday" | "Sunday":
            return "Weekend vibes 🌞"
        case _:
            return "Just a regular day."
```

```python
# Matching Multiple Options

match status_code:
    case 200 | 201:
        print("Success!")
    case 400 | 404:
        print("Client error")
    case 500:
        print("Server error")
```

Key points to understand about the match statement:

1. **Expression:** The match statement starts with an expression that you want to match against various patterns. The expression's value is compared with each pattern sequentially.

2. **Patterns:** The case keyword is followed by a pattern. Patterns can be simple values, literals, variable names, or more complex structures like sequences, dictionaries, or class instances.

3. **Execution:** The code block associated with the first matching pattern is executed. If no patterns match, the code block under case _: is executed, similar to an "else" block in an "if" statement.

4. **Pattern Guards:** You can add additional conditions using the if keyword to further filter the patterns. For example:

```python
case (x, y) if x > 0:
    # Code to execute when x is a positive number
```

5. **Deconstruction:** Patterns can destructure sequences or objects. For example, you can extract elements from a list or attributes from an object in a pattern.

```python
case [x, y]:
    # Code to execute when the expression is a list with two elements
```

6. **Value Capture:** You can capture the values from the matched expression using variable names in the patterns.

```python
case Point(x, y):
    # Code to execute when expression matches a Point object with x and y attributes
```

7. **Wildcard `_`:** The `_` is a special pattern that matches anything but discards the value. It's similar to the "default" case in a traditional switch or if statement.

8. **Nested Patterns:** You can nest match statements to perform more complex pattern matching operations.

Here's a basic example to illustrate the usage of the match statement:

```python
from dataclasses import dataclass

@dataclass
class Point:
    x: int
    y: int

def print_quadrant(point: Point):
    match point:
        case Point(x=0, y=0):
            print("At the origin")
        case Point(x>0, y>0):
            print("In the first quadrant")
        case Point(x<0, y>0):
            print("In the second quadrant")
        case Point(x<0, y<0):
            print("In the third quadrant")
        case Point(x>0, y<0):
            print("In the fourth quadrant")
        case _:
            print("On an axis")

# Usage
p = Point(3, -2)
print_quadrant(p)
```

The match statement provides a more readable and structured way to handle complex conditional logic, making Python code more expressive and less error-prone.

```python
# Matching Data Structures (Tuple Example)

point = (3, 4)

match point:
    case (0, 0):
        print("Origin")
    case (x, 0):
        print(f"X-axis at {x}")
    case (0, y):
        print(f"Y-axis at {y}")
    case (x, y):
        print(f"Somewhere else at ({x}, {y})")
```

```python
# Matching Classes (with Data Classes!)

from dataclasses import dataclass

@dataclass
class User:
    name: str
    age: int

def greet(person):
    match person:
        case User(name="Alice", age=30):
            print("Hi Alice!")
        case User(name=name, age=age) if age > 18:
            print(f"Hello {name}, you're an adult!")
        case User(name=name, age=age):
            print(f"Hey {name}, you're underage.")
```

```python
# Matching Lists

items = ["a", "b", "c"]

match items:
    case []:
        print("Empty list")
    case [x]:
        print(f"One item: {x}")
    case [x, y]:
        print(f"Two items: {x}, {y}")
    case [x, *rest]:
        print(f"First: {x}, Rest: {rest}")
```

# Context Managers with `with`

```python
with open("file.txt") as f:
    contents = f.read()
```

**How It Works Internally**

A context manager must implement:

- `__enter__(self)` → runs at the start (with block)

- `__exit__(self, exc_type, exc_val, exc_tb)` → runs when exiting the block, even if an exception occurred

```python
class MyContext:
    def __enter__(self):
        print("Entering!")
        return "Hello"

    def __exit__(self, exc_type, exc_val, exc_tb):
        print("Exiting!")

with MyContext() as msg:
    print(msg)  # Hello
```

**🛠️ Custom Context Manager (with `@contextmanager`)**

```python
from contextlib import contextmanager

@contextmanager
def my_context():
    print("Entering")
    yield
    print("Exiting")

with my_context():
    print("Inside")
```

```python
from contextlib import contextmanager

@contextmanager
def custom_open(filename):
    f = open(filename, 'r')
    try:
        yield f
    finally:
        f.close()

with custom_open("data.txt") as f:
    print(f.read())
```

# Monkey Patching

Monkey patching in Python refers to dynamically modifying or extending the behavior of classes or modules at runtime, usually by replacing or adding methods or attributes.

```python
class Person:
    def greet(self):
        print("Hello!")

# Original behavior
p = Person()
p.greet()  # Output: Hello!

# Monkey patching the greet method
def new_greet(self):
    print("Hi there!")

Person.greet = new_greet  # Monkey patching

# New behavior
p.greet()  # Output: Hi there!
```

**It is not recommended in production code unless absolutely necessary, as it can:**

- Lead to unexpected behavior
- Break encapsulation
- Make debugging difficult

# async and await

- `async` and `await` are used to write asynchronous (non-blocking) code.

In traditional Python (synchronous), each line waits for the previous one to finish.

```python
import time

def main():
    print("Start")
    time.sleep(2)
    print("End")

main()

# Everything runs one after another. The program is blocked during the sleep.
```

- `async` and `await` let your code do other things while waiting, improving performance.

With `async` and `await`, we can write code that doesn’t block, which is useful for:
- Web scraping
- API calls
- File I/O
- Background tasks

```python
import asyncio

async def say_hello():
    print("Hello")
    await asyncio.sleep(2)
    print("World")

asyncio.run(say_hello()) # Runs the main async function and starts the event loop.
# This starts the event loop and runs the say_hello() coroutine.


# async def makes the function asynchronous.
# await tells Python: “Wait here, but allow other tasks to run.”
```

**We are seeing this error because you're trying to run `asyncio.run(say_hello())` inside a running event loop — which commonly happens in `Jupyter Notebooks`, `IPython`, or `Google Colab`.**

**Step 1:** `asyncio.run(say_hello())`
- This starts the event loop and runs the say_hello() coroutine.

**Step 2:** Inside `say_hello()`
- First, it runs: `print("Hello")` → This prints Hello immediately.

**Step 3:** Then, it hits `await asyncio.sleep(2)`
- Now this is the key part.
    - 🔹 `asyncio.sleep(2)` is an asynchronous, non-blocking sleep.
    - 🔹 But you have only one task running — `say_hello()`.

    So:
    - When `await asyncio.sleep(2)` is encountered, `say_hello()` pauses.
    - The event loop checks if there's anything else to do (like other tasks).
    - ✅ But in your code, there is nothing else to do, no other tasks.
    
    So the event loop just waits for 2 seconds to let the sleep finish, then continues.

**Step 4:** After 2 seconds, it resumes and prints "World"

```python
import asyncio

async def say_hello():
    print("Hello")
    await asyncio.sleep(2)
    print("World")

# asyncio.run(say_hello()) # ❌ causes RuntimeError in notebooks

await say_hello()  # ✅ Use await directly if inside a Jupyter notebook
```

**Why This Works**

- `asyncio.run()` is meant for standalone scripts (like `.py` files).
- But in notebooks, an event loop is already running, so calling `asyncio.run()` again causes `RuntimeError`.

## Running Multiple Async Tasks

```python
import asyncio

async def task(name):
    print(f"Task {name} started")
    await asyncio.sleep(2)
    print(f"Task {name} done")

async def main():
    await asyncio.gather( runs all tasks concurrentl
        task("A"),
        task("B"),
        task("C"),
    )

# asyncio.run(main()) # ❌ causes RuntimeError in notebooks

await main()
```

- All three tasks started at the same time.
- All tasks hit `await asyncio.sleep(2)` — a non-blocking delay.
- The event loop handled both pauses efficiently, allowing them to complete concurrently.



- `asyncio.gather()` runs all tasks concurrently (at the same time).
- No blocking — efficient!

**Working:**

1. `asyncio.run(main())` starts the program.
    - Starts the event loop.
    - Schedules `main()` coroutine to run.

2. Inside `main()`:
    - Calls `asyncio.gather(...)` with three coroutine objects: task("A"), task("B"), task("C").

3. `asyncio.gather(...)`:
    - Schedules all three tasks to run concurrently.
    - It returns a single coroutine that waits for all the provided tasks to complete.

4. Now all 3 tasks start in parallel:
    - Each task runs this line: `print(f"Task {name} started")`

5. Each task hits this line: `await asyncio.sleep(2)`
    - This pauses the task.
    - The event loop sees that the task is "sleeping" and switches to another.
    - But since all 3 are sleeping for 2 seconds, the loop just waits 2 seconds in the background — not blocking.

    **⏱️ This is where true concurrency happens. The event loop waits once for 2 seconds, not 2s × 3 = 6 seconds.**

6. ✅ After ~2 seconds:
    - All 3 tasks wake up.
    - Each task continues:
    - Each runs: `print(f"Task {name} done")`


```python
import asyncio

async def say_hello():
    print("Hello")
    await asyncio.sleep(1)  # waits 1 second
    print("World")

async def say_hi():
    print("Hi")
    await asyncio.sleep(2)
    print("There")

async def main():
    await asyncio.gather(say_hello(), say_hi()) # runs all tasks concurrently

# asyncio.run(main()) # ❌ causes RuntimeError in notebooks

await main()  # ✅ Use await directly if inside a Jupyter notebook
```

**Note:**
- `async def` defines a coroutine (a function that can be paused and resumed).
- `await` is used to wait without blocking.
- `asyncio.gather()` runs multiple coroutines concurrently (at the same time).
- `asyncio.run()` starts the async program.

**If your task is CPU-intensive (e.g., heavy calculations), use `multi-threading` or `multi-processing` instead of `async`.**

## When to use async and when to use await

They go hand in hand, but they’re used in different places with different purposes:

`async` — Used to define an asynchronous function.

```python
async def function_name():
    ...
```

**You're telling Python:** “This function might have parts that should pause and let other work happen in the meantime.”

`await` — Used inside an `async def` function to pause and wait for another async task to finish.

```python
await async_function()
```

**You're telling Python:** “Pause this function here and wait for this task to complete, but don’t block other work.”

`await` doesn't force other tasks to run — it just gives permission to the event loop to schedule other tasks if they exist.

```python
# ✅ Valid: Async For

class AsyncCounter:
    def __init__(self, limit):
        self.limit = limit
        self.counter = 0

    def __aiter__(self):
        return self

    async def __anext__(self):
        if self.counter >= self.limit:
            raise StopAsyncIteration
        self.counter += 1
        return self.counter

async def main():
    async for i in AsyncCounter(3):
        print(i)

# Output:
# 1
# 2
# 3
```

## How to make something awaitable
If you're building a class and want to make it awaitable. You can make a class awaitable by implementing the `__await__()` method.

```python
class Waiter:
    def __init__(self, seconds):
        self.seconds = seconds

    def __await__(self):
        # This is what gets called when you do `await Waiter(2)`
        return asyncio.sleep(self.seconds).__await__()

async def main():
    print("Waiting...")
    await Waiter(2)  # Custom awaitable object
    print("Done!")

# asyncio.run(main()) # ❌ causes RuntimeError in notebooks

await main()
```

Here, `Waiter` is a class you can `await`, because it defines `__await__()`.

```python
class CustomAwaitable:
    def __await__(self):
        print("Awaiting custom object...")
        yield
        return "Done!"

async def main():
    result = await CustomAwaitable()
    print(result)

# asyncio.run(main()) # ❌ causes RuntimeError in notebooks

await main()
```
