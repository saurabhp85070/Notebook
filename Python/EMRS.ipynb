```python
def electricity_bill(units):
    bill = 0
    
    if units <= 100:
        bill = units * 5
    elif units <= 200:
        bill = 100 * 5 + (units - 100) * 7
    elif units <= 300:
        bill = 100 * 5 + 100 * 7 + (units - 200) * 10
    else:
        bill = 100 * 5 + 100 * 7 + 100 * 10 + (units - 300) * 15

    return bill

# Example usage
units = int(input("Enter units consumed: "))
total_bill = electricity_bill(units)
print(f"Electricity Bill for {units} units: ₹{total_bill}")
```

```python
def pyramid(n):
    for i in range(1, n+1):
        
        # spaces
        for s in range(n-i):
            print(" ", end="")
        
        # stars
        for j in range(2*i-1):
            print("*", end="")
        
        print()

pyramid(5)
```

```python
def reverse_pyramid(n):
    for i in range(n, 0, -1):
        
        # spaces
        for s in range(n - i):
            print(" ", end="")
        
        # stars
        for j in range(2*i - 1):
            print("*", end="")
        
        print()

reverse_pyramid(5)
```

```python
# NCERT

def diamond(n):

    # upper pyramid
    for i in range(1, n+1):
        
        # spaces
        for s in range(n-i):
            print(" ", end="")
        
        # stars
        for j in range(2*i-1):
            print("*", end="")
        
        print()

    # lower inverted pyramid
    for i in range(n-1, 0, -1):
        
        # spaces
        for s in range(n-i):
            print(" ", end="")
        
        # stars
        for j in range(2*i-1):
            print("*", end="")
        
        print()

diamond(5)
```

```python
# NCERT

def hollow_diamond(n):

    # upper part
    for i in range(1, n+1):
        for s in range(n-i):
            print(" ", end="")

        for j in range(1, 2*i):
            if j == 1 or j == 2*i-1:
                print("*", end="")
            else:
                print(" ", end="")
        print()

    # lower part
    for i in range(n-1, 0, -1):
        for s in range(n-i):
            print(" ", end="")

        for j in range(1, 2*i):
            if j == 1 or j == 2*i-1:
                print("*", end="")
            else:
                print(" ", end="")
        print()

hollow_diamond(3)
```

```python
# Program to write and read employee records in a binary file

import pickle
print("WORKING WITH BINARY FILES")
bfile=open("empfile.dat","ab")
recno=1
print ("Enter Records of Employees")
print()
#taking data from user and dumping in the file as list object
while True:
    print("RECORD No.", recno)
    eno=int(input("\tEmployee number : "))
    ename=input("\tEmployee Name : ")
    ebasic=int(input("\tBasic Salary : "))
    allow=int(input("\tAllowances : "))
    totsal=ebasic+allow
    print("\tTOTAL SALARY : ", totsal)
    edata=[eno,ename,ebasic,allow,totsal]
    pickle.dump(edata,bfile)
    ans=input("Do you wish to enter more records (y/n)? ")
    recno=recno+1
    if ans.lower()=='n':
        print("Record entry OVER ")
        print()
        break
# retrieving the size of file
print("Size of binary file (in bytes):",bfile.tell())
bfile.close()
# Reading the employee records from the file using load() module
print("Now reading the employee records from the file")
print()
readrec=1
try:
    with open("empfile.dat","rb") as bfile:
        while True:
            edata=pickle.load(bfile)
            print("Record Number : ",readrec)
            print(edata)
            readrec=readrec+1
except EOFError:
    pass
bfile.close()  
```

# SQL query

```python
# To display the total Quantity for each Product, excluding Products with total Quantity less than 5.

SELECT Product, SUM(Quantity) 
FROM Orders 
GROUP BY Product 
HAVING SUM(Quantity) >= 5;
```

```python
# Display the sum of Price of all the orders for which the quantity is null.

SELECT SUM(Price) AS Total_Price 
FROM Orders 
WHERE Quantity IS NULL;
```

```python
# To display all the records from the Work table in alphabetical order of W_Name.

SELECT *
FROM Work
ORDER BY W_Name ASC;
```

```python
# To display the names of contractors where W_Amt is more than 15000.

SELECT C_Name
FROM Contractor
INNER JOIN Work
ON Contractor.C_ID = Work.C_ID
WHERE W_Amt > 15000;
```
