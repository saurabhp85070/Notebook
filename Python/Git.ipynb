- `git init`: Starts a new Git repository in your project folder.
- `git clone https://github.com/user/repo.git`: Makes a copy of an existing remote repository (like from GitHub) to your local machine.
- `git status`: Shows the status of changes – what's staged, what's not, and what’s untracked.
- `git add filename.py`: Add one file  
- `git add .`: Add all changes
- `git commit -m "Fix login bug"`: Records (saves) the staged changes to the project history.
- `git push origin main`: Uploads your local commits to a remote repository like GitHub.
- `git pull origin main`: Fetches the latest changes from the remote repo and merges them into your local branch. Updates your code with any new changes made by others.
- `git fetch origin`: Gets the latest changes from the remote but does not merge them automatically. Think of it as "checking for updates" without applying them yet.
- `git merge feature-branch`: Combines changes from one branch into another.
- `git branch`: List branches  
- `git branch new-branch`: Create a branch
- `git checkout main`: Switches between branches or restores files.
- `git switch dev`: More user-friendly way to switch branches.
- `git log`: Shows a list of all commits made in the current branch.
- `git revert abc1234`: Undoes a specific commit by creating a new commit that reverses it.
- `git reset --hard HEAD~1`: Removes commits or unstages changes (more destructive than revert).
- `git stash`: Temporarily saves your changes without committing.
- `git stash pop`
- `git remote -v`: Shows the remote repositories connected to your local repo.
- `git diff`: Shows differences between files or commits.
- `git tag v1.0.0`: Marks specific commits as versions (e.g., v1.0.0).
- `git config --global user.name "Saurabh"`
- `git config --global user.email "you@example.com"`
