# Division

```python
print(7/4)     # 1.75
print(-7/4)    # -1.75
print(7/-4)    # -1.75
print(-7/-4)   # 1.75
```

# Floor division

It returns the largest integer ≤ (a / b)

```python
print(7 // 4)     # 1              7 / 4 = 1.75 and Floor → 1
print(-7 // 4)    # -2             -7 / 4 = -1.75 and Floor → -2
print(7 // -4)    # -2              7 / -4 = -1.75  and Floor → -2
print(-7 // -4)   #                 -7 / -4 = 1.75 and Floor → 1
```

# Modulus

Python follows this rule: `𝑎%𝑏 = 𝑎−(𝑎//𝑏)×𝑏`

And `//` is floor division, not truncation.

```python
print(7 % 4)     # 3
print(-7 % 4)    # 1
print(7 % -4)    # -1
print(-7 % -4)   # -3
```
