# Creating/deleting databases and collections

- `show dbs` : display all databases
- `use <databse_name>` :  used to switch to a specific database or create one if it doesn't already exist. 
- `show collections` : used to display the names of all collections in the current database.
- `db.createCollection("<collection_name>")` : used to explicitly create a collection in the current database.
- `db.<collection_name>.drop()` : used to delete an entire collection from the current database. Once a collection is dropped, all its data and indexes are permanently removed.
- `db.dropDatabase()` : used to delete the entire current database. This operation removes all collections, documents, and indexes within the database, and the database itself will no longer exist.

# Important consideration:

It is recommeded to use `""` quotes for key even in case of `field`, `operator` as key.

**Example:** 

    db.collection.find({
        "$and": [
            { "age": { "$gt": 18 } },
            { "status": "active" }
        ]
    })

**This works in:**
- MongoDB shell
- Compass
- PyMongo (Python)
- Node.js MongoDB driver
- Any environment using strict JSON

### find()

- The `find()` method is used to query documents from a collection. It returns all documents that match the filter criteria.

```python
# syntax:

db.collection.find(query, projection)

# query (optional): The filter criteria (like a WHERE clause in SQL).
# projection (optional): Specifies which fields to return (like SELECT columns in SQL).
```

```python
db.cars.find() # return all documents of cars
db.cars.find().pretty() # display documents in more organized waya
```

## Update

### updateOne()

- `updateOne()` is used to update a single document that matches a given filter in a collection.

```python
# Syntax:

db.collection.updateOne(
  <filter>,                  # {field: value, ...}
  <update>,                  # {$update_operator: {field: value, ...}}
  { upsert: <boolean> }      # optional
)

# <filter>: A query that matches the document to update.
# <update>: The update operations (e.g., $set, $inc, $unset, etc.).
# upsert (optional): If true, creates a new document if none matches the filter.
```

```python
db.cars.updateOne(
    {"model": "Nexon"},            # filter
    {"$set": {"color": "Red"} }    # update
);

# Updates the color of the first document where model is "Nexon".
```

```python
# Updates the color of the first document where model is "Nexon" and maker is "Tata".

db.cars.updateOne(
    {"model": "Nexon", "name": "Tata"},          # mutiple filter 
    {"$set": {"color": "Red"} }                  # update
    { "upsert": true }                           # insert if not found
);

# Updates if model: "Nexon" exists; otherwise inserts a new document.
```

```python
db.cars.updateOne(
    {"model": "Nexon"},              # filter
    {"$unset": {"color": "Red"} }    # remove
);
```

```python
# update in array field

db.cars.updateOne(
    {"model": "Nexon"},                       # filter
    {"$push": {"features": "Heated Seats"} }    # add "Heated Seats" in features: arraay field
);

# update in only one matching document
```

```python
# update in array field

db.cars.updateOne(
    {"model": "Nexon"},                       # filter
    {"$pull": {"features": "Heated Seats"} }    # remove "Heated Seats" in features: array field
);

# update in only one matching document
```

```python
# update multiple values in array field

db.cars.updateOne(
    {"model": "Nexon"},                       # filter
    {"$push": {"features": {"$each": ["Wireless Charging", "Voice Control"]}} }    # add "Wireless Charging" and "Voice Control" in features field
);
```

```python
# update in nested fields

db.cars.updateOne(
    {"model": "Nexon"},                       # filter
    {"$set": {"engine.torque": "270 Nm"} }    # update nested field value
);
```

### updateMany()

- `updateMany()` is used to update multiple documents that match a given filter in a collection.


```python
# syntax:

db.collection.updateMany(
  <filter>,               # {field: value, ...}
  <update>,               # {$update_operator: {field: value, ...}}
  { upsert: <boolean> }   # optional
)
```

```python
db.cars.updateMany(
    {"fuel_type": "Diesel"},                       # filter
    {"$set": {"alloys": "Yes"} }                    # update 
);

# Updates all documents with fuel: "Diesel" 
```

```python
db.cars.updateMany(
    {"fuel_type": "Diesel"},                       # filter
    {"$set": {"alloys": "Yes"} }                    # update 
    {"upsert": true}
);

# Updates all documents with fuel: "Diesel" or inserts one if none are found.
```

**Note:**  `upsert` with `updateMany()` inserts only one document, even if the filter matches none.

**Common update operators:**

- `$set`: Update or add a field.
- `$inc`: Increment a value.
- `$unset`: Remove a field.
- `$push`: Add an item to an array.
- `$pull`: Remove item(s) from an array.

# Comparison Operators
Used to compare field values in queries (like SQL’s `=`, `>`, etc.)

| Operator | Meaning          | Example                                  |
| -------- | ---------------- | ---------------------------------------- |
| `$eq`    | Equal            | `{ age: { $eq: 25 } }`                   |
| `$ne`    | Not equal        | `{ status: { $ne: "active" } }`          |
| `$gt`    | Greater than     | `{ age: { $gt: 18 } }`                   |
| `$gte`   | Greater or equal | `{ score: { $gte: 90 } }`                |
| `$lt`    | Less than        | `{ age: { $lt: 60 } }`                   |
| `$lte`   | Less or equal    | `{ score: { $lte: 50 } }`                |
| `$in`    | In array         | `{ city: { $in: ["Delhi", "Mumbai"] } }` |
| `$nin`   | Not in array     | `{ age: { $nin: [18, 21] } }`            |


```python
# syntax:

db.collection.find({
    field1: {$comparison_operator: value1, $comparison_operator: value2},
    field2: {$comparison_operator: value3},
    ...
})
```

```python
# Find users whose age is exactly 25:
db.users.find({ age: { $eq: 25 } })

# Find users whose status is not "active":
db.users.find({ status: { $ne: "active" } })

# Find users older than 30:
db.users.find({ age: { $gt: 30 } })

# Find users whose score is 80 or more:
db.users.find({ score: { $gte: 80 } })

# Find active users between ages 20 and 30:
db.users.find({
  active: true,
  age: { $gte: 20, $lte: 30 }
})

# Find users in either Delhi or Mumbai:
db.users.find({ city: { $in: ["Delhi", "Mumbai"] } })

# Find users who are not from Delhi or Mumbai:
db.users.find({ city: { $nin: ["Delhi", "Mumbai"] } })
```

```python
# Employees aged between 25 and 35, and in IT department
db.employees.find({
  age: { $gte: 25, $lte: 35 },
  department: "IT"
})
```

# Logical Operators
Used to combine multiple query conditions.

| Operator | Meaning                        | Example                                                  |
| -------- | ------------------------------ | -------------------------------------------------------- |
| `$and`   | Logical AND                    | `{ $and: [{ age: { $gt: 20 } }, { age: { $lt: 30 } }] }` |
| `$or`    | Logical OR                     | `{ $or: [{ city: "Delhi" }, { city: "Mumbai" }] }`       |
| `$not`   | Negates the condition          | `{ age: { $not: { $gt: 30 } } }`                         |
| `$nor`   | Opposite of OR (none are true) | `{ $nor: [{ status: "active" }, { age: { $lt: 18 } }] }` |


```python
# syntax: (except $not)

db.collection.find({
    "$logical_operator": [
        <condition1>,      # {"field1": value1},
        <condition2>,      # {"field2": {"$comparison_operator": value2}}
        ...
    ]
})
```

## and
- `$and` takes an array of conditions.
- Each element inside the array is a query object.

```python
# syntax:

{ $and: [ { condition1 }, { condition2 }, ... ] }
```

```python
# Find employees in the IT department AND with age greater than 25
db.employees.find({
  $and: [
    { department: "IT" },
    { age: { $gt: 25 } }
  ]
})
```

```python
# # Find employees in the IT department AND with age greater than 18 and less than 25

db.employees.find({
  $and: [
    { department: "IT" },
    { age: { $gt: 18, $lt: 25 } }
  ]
})
```

**Tip: If both conditions are on the same field, you don’t even need `$and`:**

```python
db.employees.find({
  department: "IT",
  age: { $gt: 18, $lt: 25 }
})
```

```python
# Employees in HR with salary >= 50000 AND age <= 40
db.employees.find({
  $and: [
    { department: "HR" },
    { salary: { $gte: 50000 } },
    { age: { $lte: 40 } }
  ]
})
```

```python
# MongoDB automatically ANDs multiple fields if you write them directly:

# Same as $and: [ {department: "IT"}, {age: {$gt: 25}} ]
db.employees.find({
  department: "IT",
  age: { $gt: 25 }
})
```

**Multiple Conditions on Same Field:**

```python
# Employees with age > 25 AND age < 40
db.employees.find({
  $and: [
    { age: { $gt: 25 } },
    { age: { $lt: 40 } }
  ]
})

# Without $and, this wouldn’t work because you’d be overwriting the age key.
```

## or
- `$or` takes an array of query objects.

```python
# syntax:

{ $or: [ { condition1 }, { condition2 }, ... ] }
```

```python
# Find employees in IT department OR with salary > 50000
db.employees.find({
  $or: [
    { department: "IT" },
    { salary: { $gt: 50000 } }
  ]
})
```

**You can combine `$or` with `$and` for complex queries:**

```python
# Employees in IT with salary > 60000 OR in HR with age < 30
db.employees.find({
  $or: [
    { $and: [ { department: "IT" }, { salary: { $gt: 60000 } } ] },
    { $and: [ { department: "HR" }, { age: { $lt: 30 } } ] }
  ]
})
```

```python
# Employees whose age is less than 25 OR greater than 60
db.employees.find({
  $or: [
    { age: { $lt: 25 } },
    { age: { $gt: 60 } }
  ]
})
```

## nor
- Takes an array of conditions.
- Matches documents where none of the conditions are true.

```python
# syntax:
{ $nor: [ { condition1 }, { condition2 }, ... ] }
```

```python
# Find employees who are NOT in IT department AND do NOT have salary > 50000
db.employees.find({
  $nor: [
    { department: "IT" },
    { salary: { $gt: 50000 } }
  ]
})
```

```python
# Employees NOT younger than 25 AND NOT with salary less than 30000
db.employees.find({
  $nor: [
    { age: { $lt: 25 } },
    { salary: { $lt: 30000 } }
  ]
})
```

```python
# Employees whose age is NOT less than 25 and NOT greater than 60
db.employees.find({
  $nor: [
    { age: { $lt: 25 } },
    { age: { $gt: 60 } }
  ]
})
```

## not
- `$not` inverts the result of another condition.
- It’s applied to a single condition (often a comparison or regex).
- It returns documents where the condition is not true.

```python
# syntax:

{ field: { $not: { operator-expression } } }
```

```python
# Employees whose salary is NOT greater than 50000 (i.e., ≤ 50000)
db.employees.find({
  salary: { $not: { $gt: 50000 } }
})
```

```python
# Employees whose department does NOT start with "IT"
db.employees.find({
  department: { $not: /^IT/ }
})
```

```python
# Employees whose department is NOT in HR or Sales
db.employees.find({
  department: { $not: { $in: ["HR", "Sales"] } }
})
```

**Note:** `$not` with `$in` is different from `$nin`.

```python
# Using $nin
{ department: { $nin: ["HR", "Sales"] } } # same outcome, shorter syntax
```

```python
# ❌ Incorrect - this will not work
db.employees.find({
  departent: { $not: ["HR", "Sales"] }
})
```

- `$not` must be used with another operator, like `$gt`, `$in`, `$regex`, etc.
- `$not` does not negate an entire field or array — it negates the result of a condition.

**Confusing example:**

```python
# Confusing Example (Don’t do this):
db.users.find({
  tags: { $not: { $all: ["mongodb", "backend"] } }
})
```

- `$all` expects an array of values that must all be present in the tags array field.
- `$not` in MongoDB does not invert the logic at the top level like SQL’s NOT — instead, it applies to the result of a single operator’s evaluation.

**So this query actually means:** Find documents where the field tags does not match the $all operator's condition.

The Problem

- The `$not` operator only works with regular expressions and other single expressions that can be directly negated.
- Using `$not` with `$all doesn’t produce the logical “NOT contains all” you might think — it can return unexpected results or even no results depending on your dataset.

```python
# Above may not behave as expected. Instead, use $nor:
db.users.find({
  $nor: [
    { tags: { $all: ["mongodb", "backend"] } }
  ]
})
```

**How It Works:**

- `{ tags: { $all: ["mongodb", "backend"] } }` → finds documents where the tags array contains both `"mongodb"` and `"backend"`.
- `$nor: [ ... ]` → returns documents where that condition is false.

In other words: Return users whose `tags` array does not contain both `"mongodb"` and `"backend"`.

```python
{ _id: 1, tags: ["mongodb", "backend", "nodejs"] } // has both
{ _id: 2, tags: ["mongodb"] }                      // missing backend
{ _id: 3, tags: ["backend"] }                      // missing mongodb
{ _id: 4, tags: ["frontend", "nodejs"] }           // missing both
```

**Result:**
- `_id: 1` → ❌ excluded (has both)
- `_id: 2` → ✅ included
- `_id: 3` → ✅ included
- `_id: 4` → ✅ included

This query says: "Find all users where it is not true that tags contains both 'mongodb' and 'backend'."

This behaves as expected. It negates the entire condition using `$nor`.


- `$all` is an operator used with arrays to match documents where the array contains all the specified elements, in any order.

# Element Operators
Used to test for the presence or data type of a field.

| Operator  | Meaning                       | Example                              |
| --------- | ----------------------------- | ------------------------------------ |
| `$exists` | Checks if field exists or not | `{ middleName: { $exists: false } }` |
| `$type`   | Checks field's BSON data type | `{ age: { $type: "int" } }`          |


## exists
- `$exists` checks whether a field is present (or absent) in a document, regardless of its value.
- It does not check if the value is null — it only checks if the field key itself exists.

```python
# syntax:
{ field: { $exists: <boolean> } }
```

```python
# Find users that have an "email" field
db.users.find({
  email: { $exists: true }
})
```

```python
# Find users without an "email" field
db.users.find({
  email: { $exists: false }
})
```

```python
# Users who have "age" field and it's greater than 25
db.users.find({
  age: { $exists: true, $gt: 25 }
})
```

## type
- $type matches documents where the value of the field is of a specific BSON` type.
- Useful for finding fields with unexpected types (e.g., a number stored as a string).

```python
# syntax:
{ field: { $type: <BSON type number or alias> } }

# <BSON type> can be:
    # A number (BSON type code)
    # A string alias (more readable)
```

```python
# Find users where "age" is stored as a Number (int or double)
db.users.find({
  age: { $type: 16 } // 16 = int32
})
```

```python
# Find users where "age" is a string
db.users.find({
  age: { $type: "string" }
})
```

```python
# Find users where "age" is either int or string
db.users.find({
  age: { $type: ["int", "string"] }
})
```

# Evaluation Operators
Used to evaluate values with specific conditions.

| Operator | Meaning                                | Example                                        |
| -------- | -------------------------------------- | ---------------------------------------------- |
| `$regex` | Pattern matching (like LIKE in SQL)    | `{ email: { $regex: /@gmail\.com$/ } }`        |
| `$expr`  | Use aggregation expressions in queries | `{ $expr: { $gt: ["$spent", "$budget"] } }`    |
| `$mod`   | Modulo operation                       | `{ score: { $mod: [5, 0] } }` → divisible by 5 |
| `$text`  | Full-text search (requires text index) | `{ $text: { $search: "mongodb" } }`            |


## regex
- `$regex` lets you search for text using patterns — like "starts with", "contains", or "ends with".

```python
# syntax:
{ field: { $regex: /pattern/, $options: "i" } }

# /pattern/ → your search pattern.
# $options: "i" → makes it case-insensitive (optional).

# OR

{ field: { $regex: "pattern", $options: '<flags>' } }
```

```python
# Find users whose name contains "john"
db.users.find({
  name: { $regex: /john/ }
})
```

```python
# Find names containing "john" (ignore case)
db.users.find({
  name: { $regex: /john/i }
})

# OR using $options
db.users.find({
  name: { $regex: "john", $options: "i" }
})
```

```python
# Find names starting with "J"
db.users.find({
  name: { $regex: /^J/ }
})
```

```python
# Find names ending with "son"
db.users.find({
  name: { $regex: /son$/ }
})
```

```python
# Names starting with 'A' or 'B' and followed by at least 2 letters
db.users.find({
  name: { $regex: /^[AB][a-z]{2,}/i }
})
```

```python
# Users whose name starts with "A" and age > 25
db.users.find({
  name: { $regex: /^A/, $options: "i" },
  age: { $gt: 25 }
})
```

- `$options` operator is used with `$regex` to modify how the regular expression behaves.

**Common `$options` flags:**
    
| Option | Meaning                         |
| ------ | ------------------------------- |
| `i`    | Case-insensitive matching       |
| `m`    | Multiline matching              |
| `s`    | Dot matches newline (`.` = all) |
| `x`    | Ignore whitespace and comments  |


```python
{ email: { $regex: "@gmail\\.com$", $options: "is" } }

# i = case-insensitive
# s = dot . matches newline too
```

```python
db.user.find({ email: { $regex: /@gmail\.com$/ } })
```

- `/.../` → These slashes are not part of the pattern itself, but they indicate the start and end of the regular expression.
- `\.` → Escapes the dot (.), because in regex, a dot means "any character". So \. means a literal dot
- `$` → Anchors the match to the end of the string

**If you want to match `@GMAIL.com`, `@Gmail.com`, etc., add the `"i"` option:**

```python
db.users.find({ email: { $regex: /@gmail\.com$/i } }) #  i = case-insensitive
```

**MongoDB supports two ways to define a regular expression:**

```python
# 1. JavaScript-style with slashes ✅ (Preferred for inline regex):

{ email: { $regex: /@gmail\.com$/ } }
```

```python
# 2. As a string ❗(More portable, useful when regex is dynamic)

{ email: { $regex: "@gmail\\.com$" } }
```

**Note:** In string style, the backslash `\` has to be escaped, so it becomes `\\`.

## expr
- `$expr` in MongoDB lets you use aggregation expressions inside a normal query.

That means you can compare two fields or perform calculations directly inside your `find()` filter — without needing `$project` or `$match` in an aggregation pipeline.

We can compare fields with other fields, or use expressions like `$add`, `$gt`, `$eq`, etc. inside a normal query.

```python
# syntax:
db.collection.find({
  $expr: {
    <aggregation expression>
  }
})
```

Normally, `find()` only allows you to compare a field with a constant value: `db.sales.find({ quantity: { $gt: 10 } })`

But what if you want to compare two fields in the same document? Like: 👉 “Find all sales where `quantity > minQuantity`”

That’s not possible with basic `find()`, so you need `$expr`.

```python
# Compare two fields:
db.sales.find({
  $expr: {
    $gt: ["$quantity", "$minQuantity"] # comparing with two fields quantity > minQuantity
  }
})

# return all documents where quantity is greater than minQuantity
```

**Note:** When referring to a field inside an expression, you need `$` before the field name.

```python
# Calculate and compare inside query:
db.products.find({
  $expr: {
    $eq: [
      { $add: ["$price", "$tax"] },
      "$total"
    ]
  }
})

# This finds products where price + tax == total
```

```python
# products where price is more than twice the cost:

db.products.find({
  $expr: { $gt: ["$price", { $multiply: ["$cost", 2] }] }
})
```

```python
# without $expr 
db.users.find({ age: { $gt: 30 } })


# with $expr
db.users.find({
  $expr: { $gt: ["$age", "$experienceYears"] }
})
```

```python
# Use with $cond (if-else logic):
db.users.find({
  $expr: {
    $cond: {
      if: { $gt: ["$age", 18] },
      then: true,
      else: false
    }
  }
})
```

But note: this one just returns true or false inside `$expr`—it's more useful inside the aggregation pipeline.

## mod

```python
# syntax:
{ field: { $mod: [divisor, remainder] } }

# divisor: The number you divide by.
# remainder: The result you expect after division
```

```python
# Find users whose age is even:
db.users.find({
  age: { $mod: [2, 0] }
})

# Finds documents where the field divided by a number has a specific remainder.
```

**You can’t use it directly to compare two fields — for that, you’d combine it with `$expr` like:**

```python
db.data.find({
  $expr: { $eq: [{ $mod: ["$score", 3] }, 1] }
})
```

## text
- `$text` is a MongoDB query operator that allows you to perform text search on string content in your documents, like searching for keywords — similar to how search engines work.

**📌 But to use `$text`, you must first create a text index on the field(s) you want to search in.**

**It’s useful when you want to:**
- Search for keywords in blog posts, product names, descriptions, etc.
- Perform case-insensitive and language-aware full-text search.



```python
# Suppose you have a posts collection:
{
  title: "MongoDB Tutorial",
  content: "This tutorial explains MongoDB text search."
}


# 1️⃣ Create a text index on the fields you want to search::
db.posts.createIndex({ title: "text", content: "text" })


# 3️⃣ Now perform a search using $text::
db.posts.find({ $text: { $search: "tutorial" } })

# This will return documents where the word "tutorial" appears in either the title or content.
```

**Features of `$text`:**

| Feature                     | Description                                                                        |
| --------------------------- | ---------------------------------------------------------------------------------- |
| Case-insensitive            | Searches are not case-sensitive. `"MongoDB"` and `"mongodb"` are treated the same. |
| Word stemming (by language) | `"run"` and `"running"` may be considered similar if the language is set.          |
| Search multiple fields      | Use a text index across multiple fields.                                           |
| Advanced search options     | You can use operators like `-word` (exclude), `"exact phrase"`, etc.               |


**Note:**
- You can only have one text index per collection, but it can cover multiple fields.

```python
db.posts.createIndex({ title: "text" })        # OK
db.posts.createIndex({ content: "text" })      # ❌ Error: cannot create more than one text index

db.posts.createIndex({
  title: "text",
  content: "text",
  tags: "text"
})                                             # OK

# This creates a single text index that includes title, content, and tags.
```

```python
# Search for "mongodb" but exclude "tutorial":
db.posts.find({ $text: { $search: "mongodb -tutorial" } })
```

- `"mongodb"`: Document must contain the word "mongodb".
- `-tutorial`: Document must NOT contain the word `"tutorial"` (anywhere in text-indexed fields like `title` or  `content`).

Let’s check:
- It contains "mongodb" ✅
- It also contains "tutorial" ❌ (appears in both title and content)

**Final Result:** This document will NOT be returned by the query because it contains the word "tutorial", which is explicitly excluded using `-tutorial`.

**📝 Notes:**
- Only one `$text` operator can be used per query.
- `$text` requires a text index, otherwise the query fails.
- You can weigh fields differently when creating text indexes.

### setting weights

```python
db.posts.createIndex(
  { title: "text", content: "text" },
  { weights: { title: 5, content: 1 } }
)
```

This gives higher importance to keywords in the title.

- `weights`: Specifies the importance of the fields during search ranking.
    - `title: 5`: Title is 5 times more important than content.
    - `content: 1`: Default importance.

```python
# posts collection:
{
  title: "MongoDB tutorial",
  content: "This tutorial teaches MongoDB text indexing"
}


db.posts.find({ $text: { $search: "tutorial" } })
```

MongoDB uses both `title` and ``content`, but matches in `title` will influence ranking more due to the weight 5.

## eleMatch
- `$elemMatch` is used to match documents where at least one element in an array satisfies multiple conditions.

Think of it like: `“Is there any one item in the array that matches all of these conditions?”`

**Without `$elemMatch`, MongoDB checks each condition separately across the array. With `$elemMatch`, you say:` “Apply all conditions to the same single element of the array.”`**

```python
# syntax:
{ array_field: { $elemMatch: { 
    condition1,       # {field1: value}
    condition2,       # {field2: {operator: value}}
    ...
} } }
```

```python
# products collection:
{
  name: "Laptop",
  prices: [
    { country: "US", price: 1000 },
    { country: "India", price: 800 },
    { country: "UK", price: 900 }
  ]
}



# ❌ This query won’t work correctly:
db.products.find({
  "prices.country": "India",
  "prices.price": { $lt: 900 }
})
# This may match across two different elements — country from one, price from another.


# ✅ Correct query using $elemMatch:
db.products.find({
  prices: {
    $elemMatch: {
      country: "India",    # condition1
      price: { $lt: 900 }  # condition2
    }
  }
})
```

```python
# Assume a scores array of objects like:
scores: [
  { subject: "Math", score: 90 },
  { subject: "Science", score: 85 }
]

# Find users who scored more than 80 in Math:
db.users.find({
  scores: {
    $elemMatch: {
      subject: "Math",    # condition1
      score: { $gt: 80 }  # condition2
    }
  }
})
```

## size
- `$size` is used to match arrays that have a specific number of elements.

```python
# syntax:
{ field: { $size: number } }

# field: the name of the array field
# number: the exact number of elements you want the array to have
```

```python
# Find users with exactly 2 tags:
db.users.find({
  tags: { $size: 2 }
})

# find users who have exactly 3 hobbies:
db.users.find({
  hobbies: { $size: 3 }
})
```

**Notes:**
- You can’t use `$size` with other array operators like `$gt` or `$lt` (e.g., you can’t do `{ $size: { $gt: 3 } }`).
- It only works for exact size match.

**How to check array length > 3?**

You’ll need to use `aggregation` for that, like:

```python
db.users.aggregate([
  {
    $project: {
      name: 1,
      hobbiesCount: { $size: "$hobbies" } # $size will count elements in hobbies array
    }
  },
  {
    $match: {
      hobbiesCount: { $gt: 3 }
    }
  }
])
```

**Notes:**
- `$size` only works on arrays. If the field is not an array, it will throw an error.
- If the field might be null, missing, or not always an array, use `$ifNull` or `$isArray` for safety.

```python
# Safely Using $size When Field Might Not Exist:

db.students.aggregate([
  {
    $project: {
      name: 1,
      subjectCount: {
        $cond: {
          if: { $isArray: "$subjects" },
          then: { $size: "$subjects" },
          else: 0
        }
      }
    }
  }
])
```

## addToSet
- `$addToSet` operator is a group accumulator and an update operator that adds unique values to an array. It prevents duplicates, unlike `$push`.


### In Update Operations 

```python
# syntax:
{ $addToSet: { <array_field>: <value> } }

# OR if you want to add multiple values at once:
{ $addToSet: { <arrayField>: { $each: [value1, value2, ...] } } }
```

```python
# users collection:
{
  _id: 1,
  name: "Alice",
  tags: ["database", "nosql"]
}

# add a single value to an array:
db.users.updateOne(
  { _id: 1 },
  { $addToSet: { tags: "mongodb" } }
)


# Output:
{
  _id: 1,
  name: "Alice",
  tags: ["database", "nosql", "mongodb"]  
}
# If "mongodb" was already there, the array would remain unchanged.
```

```python
db.users.updateOne(
  { _id: 1 },
  { $addToSet: { skills: { $each: ["Python", "Node.js", "Docker"] } } }
)
```

### In Aggregation ($group Stage)

```python
# syntax:
{
  $group: {
    _id: "$category",
    uniqueTags: { $addToSet: "$tag" }
  }
}
```

```python
# articles ccollections:
[
  { _id: 1, title: "Intro to MongoDB", author: "Alice", tag: "mongodb" },
  { _id: 2, title: "Data Models", author: "Alice", tag: "database" },
  { _id: 3, title: "Indexing", author: "Bob", tag: "performance" },
  { _id: 4, title: "Sharding", author: "Alice", tag: "scaling" },
  { _id: 5, title: "Aggregation", author: "Bob", tag: "mongodb" }
]

# collect unique values per group:
db.articles.aggregate([
  {
    $group: {
      _id: "$author",
      uniqueTags: { $addToSet: "$tag" }
    }
  }
])
```

- Groups articles by `author`.
- Collects unique `tags` for each author.

```python
# Output:
[
  {
    _id: "Alice",
    uniqueTags: ["mongodb", "database", "scaling"]
  },
  {
    _id: "Bob",
    uniqueTags: ["performance", "mongodb"]
  }
]
```

### With $each
You can use `$addToSet` with `$each` to try to add multiple values at once, all checked for uniqueness.

```python
# users collection:
{
  _id: 1,
  name: "Alice",
  tags: ["database", "nosql"]
}

# add multiple unique values to an array
db.users.updateOne(
  { _id: 1 },
  { $addToSet: { tags: { $each: ["mongodb", "database", "nosql"] } } }
)


# Output:
{
  _id: 1,
  name: "Alice",
  tags: ["database", "nosql", "mongodb"]  # "mongodb" added, others ignored
}
```

| Query                                   | Purpose                    | Result                                    |
| --------------------------------------- | -------------------------- | ----------------------------------------- |
| `$addToSet: { tags: "mongodb" }`        | Add single unique value    | Adds `"mongodb"` if not present           |
| `$addToSet: { tags: { $each: [...] } }` | Add multiple unique values | Adds only values not already in the array |
| `$addToSet` in aggregation              | Group unique values        | Collects distinct tags per group          |

A cursor in MongoDB is an object that allows you to access the results of a query one by one instead of all at once.

When you run a `find()` query in MongoDB, it returns a cursor, not the actual documents immediately.

**Why Cursor?**
- Avoids loading all results into memory at once (important for big datasets).
- Lets you process results in chunks.
- Lets you iterate documents as they come.
- Efficient for large result sets.
- Enables batch processing.
- Allows iteration, pagination, and delayed fetching.

# Aggregation Framework

MongoDB’s Aggregation Framework is a powerful feature used to process data records and return computed results. It works like a data pipeline, where documents pass through multiple stages, transforming them step by step.

In simple terms: `The aggregation framework in MongoDB is used to process data records and transform them into meaningful results — like summary reports, grouping, averages, counts, and more.`

**Why Aggregation?**
- Perform data analysis (like `GROUP BY` in SQL).
- Create custom views of data.
- Summarize, filter, and transform documents.
- Handle complex queries more efficiently than multiple `find()` calls.


1. You send an array of stages.
2. Each stage starts with `$` (e.g., `$match`, `$group`, `$sort`).
3. The output of one stage becomes the input to the next stage.

```python
# basic syntax:

db.collection.aggregate([
  { stage1: { /* options */ } },
  { stage2: { /* options */ } },
  { stage3: { /* options */ } }
])


# Each stage is an object with a pipeline operator, like $match, $group, $sort, etc.
```

```python
# Sample Collection: orders

db.orders.insertMany([
  { item: "pen", qty: 10, price: 5, category: "stationery" },
  { item: "book", qty: 2, price: 100, category: "stationery" },
  { item: "mouse", qty: 1, price: 500, category: "electronics" },
  { item: "keyboard", qty: 1, price: 700, category: "electronics" }
])
```

**Common Aggregation Stages:**

| Stage      | Purpose                                     |
| ---------- | ------------------------------------------- |
| `$match`   | Filter documents (like `find`)              |
| `$group`   | Group documents and calculate aggregates    |
| `$sort`    | Sort documents                              |
| `$project` | Reshape documents (select fields, computed) |
| `$limit`   | Limit number of documents                   |
| `$skip`    | Skip documents                              |
| `$count`   | Count documents                             |
| `$unwind`  | Deconstruct arrays                          |


**When to Use Aggregation Pipeline**

You should use it when you need to:

- Summarize data (e.g., totals, averages, counts).
- Transform data (change structure, rename fields).
- Join collections ($lookup).
- Filter, sort, and limit in complex ways in a single query.
- Work on arrays inside documents.
- Do analytics inside MongoDB without exporting to another tool.

Not ideal for:

- Simple document retrieval (use find()).
- Very large transformations that could be better done in an analytics tool like Spark if your data is massive.

## match

The `$match` stage is used in an aggregation pipeline to filter documents, just like the `.find()` method. It allows you to specify query conditions to pass only matching documents to the next stage in the pipeline.

```python
# syntax:
db.collection.aggregate([
  { $match: { <query> } }
])
```

```python
# Filter users by city:
db.users.aggregate([
  { $match: { city: "Delhi" } }
])


# Filter users with age > 30:
db.users.aggregate([
  { $match: { age: { $gt: 30 } } }
])


# Combine conditions using $and or directly in one object
# filter users aged between 25 and 35 who live in Delhi:
db.users.aggregate([
  { $match: { age: { $gte: 25, $lte: 35 }, city: "Delhi" } }
])

# Filter users who have a specific skill using array operator
# filter users who have "MongoDB" in their skills array:
db.users.aggregate([
  { $match: { skills: "MongoDB" } }
])
```

**You typically use `$match` early in the pipeline to reduce the number of documents passed to subsequent stages like `$group`, `$project`, or `$sort`.**

```python
# Groups users by age in Delhi and counts how many users per age:
db.users.aggregate([
  { $match: { city: "Delhi" } },
  { $group: { _id: "$age", count: { $sum: 1 } } }
])
```

## group

The `$group` stage groups documents by a specified `_id` expression and allows you to perform aggregations like `sum`, `avg`, `max`, `min`, `count`, etc., within each group.

```python
# syntax:

db.collection.aggregate([
      {
      $group: {
        _id: <expression to group by>,                  # Grouping key (can be a field, null, or expression)
        <accumulatorField1>: { <accumulator> : <expression> },
        <accumulatorField2>: { <accumulator> : <expression> },
        ...
      }
    }
])


# <accumulator> → Functions like $sum, $avg, $min, $max, $push, $addToSet, etc.
```

```python
# sample orders collection:

{ _id: 1, customer: "Alice", amount: 250, city: "Delhi" }
{ _id: 2, customer: "Bob", amount: 300, city: "Delhi" }
{ _id: 3, customer: "Alice", amount: 450, city: "Mumbai" }
{ _id: 4, customer: "Charlie", amount: 150, city: "Mumbai" }
```

```python
# Total order amount by customer:
db.orders.aggregate([
  {
    $group: {
      _id: "$customer",                # group by
      totalAmount: { $sum: "$amount" }
    }
  }
])


# Ouput:
{ _id: "Alice", totalAmount: 700 }
{ _id: "Bob", totalAmount: 300 }
{ _id: "Charlie", totalAmount: 150 }
```

```python
# Count of orders per city:
db.orders.aggregate([
  {
    $group: {
      _id: "$city",
      orderCount: { $sum: 1 } # For each group (i.e., each unique city), counts the number of documents.
    }
  }
])


# Output:
{ _id: "Delhi", orderCount: 2 }
{ _id: "Mumbai", orderCount: 2 }
```

`{ $sum: 1 }` tells MongoDB to:
- Add `1` for every document in the group.
- So it works like a counter.



```python
# Average order amount by customer:
db.orders.aggregate([
  {
    $group: {
      _id: "$customer",
      avgAmount: { $avg: "$amount" }
    }
  }
])


# Get max and min amount per city:
db.orders.aggregate([
  {
    $group: {
      _id: "$city",
      maxAmount: { $max: "$amount" },
      minAmount: { $min: "$amount" }
    }
  }
])


# Group all documents together (e.g., total revenue)
db.orders.aggregate([
  {
    $group: {
      _id: null,
      totalRevenue: { $sum: "$amount" }
    }
  }
])
```

**Accumulators You Can Use in `$group`:**

| Operator    | Description                   |
| ----------- | ----------------------------- |
| `$sum`      | Total sum                     |
| `$avg`      | Average                       |
| `$min`      | Minimum                       |
| `$max`      | Maximum                       |
| `$first`    | First document value in group |
| `$last`     | Last document value in group  |
| `$push`     | Create array of values        |
| `$addToSet` | Create array of unique values |


### sum vs add

| Aspect         | `$sum`                                                        | `$add`                                                                    |
| -------------- | ------------------------------------------------------------- | ------------------------------------------------------------------------- |
| **Purpose**    | Used to **calculate total** over multiple documents or fields | Used to **add values** (two or more) within a **single document**         |
| **Used in**    | Often inside `$group` to aggregate across documents           | Used inside `$project` or `$addFields` to add values in a single document |
| **Input type** | Can take an array or field reference                          | Takes two or more expressions/values                                      |
| **Returns**    | The total (sum)                                               | The result of the addition                                                |


### Grouping based on two or more fields

```python
# orders collection:
[
  { "item": "Apple", "category": "Fruit", "city": "Delhi", "quantity": 10, "price": 5 },
  { "item": "Banana", "category": "Fruit", "city": "Delhi", "quantity": 20, "price": 2 },
  { "item": "Carrot", "category": "Vegetable", "city": "Delhi", "quantity": 15, "price": 3 },
  { "item": "Apple", "category": "Fruit", "city": "Mumbai", "quantity": 5, "price": 6 }
]



db.orders.aggregate([
  {
    $group: {
      _id: {
        category: "$category",
        city: "$city"
      },
      totalQuantity: { $sum: "$quantity" },
      totalSales: { $sum: { $multiply: ["$quantity", "$price"] } }
    }
  }
])

```

```python
# output:
[
  { "_id": { "category": "Fruit", "city": "Delhi" }, "totalQuantity": 30, "totalSales": 90 },
  { "_id": { "category": "Vegetable", "city": "Delhi" }, "totalQuantity": 15, "totalSales": 45 },
  { "_id": { "category": "Fruit", "city": "Mumbai" }, "totalQuantity": 5, "totalSales": 30 }
]
```

## sort

The `$sort` stage is used to sort documents in either ascending (`1`) or descending (`-1`) order based on one or more fields.

```python
# syntax:

{ $sort: { <field1>: 1, <field2>: -1, ... } }

# 1 = Ascending order
# -1 = Descending order
```

```python
# Sample orders Collection:
{ _id: 1, customer: "Alice", amount: 250, city: "Delhi" }
{ _id: 2, customer: "Bob", amount: 300, city: "Delhi" }
{ _id: 3, customer: "Alice", amount: 450, city: "Mumbai" }
{ _id: 4, customer: "Charlie", amount: 150, city: "Mumbai" }
```

```python
# Sort by amount in ascending order:
db.orders.aggregate([
  { $sort: { amount: 1 } }
])

# Output:
Charlie: 150  
Alice: 250  
Bob: 300  
Alice: 450  
```

```python
# Sort by amount in descending order:
db.orders.aggregate([
  { $sort: { amount: -1 } }
])


# Output:
Alice: 450  
Bob: 300  
Alice: 250  
Charlie: 150  
```

```python
#  Sort by city ascending, then by amount descending:
db.orders.aggregate([
  { $sort: { city: 1, amount: -1 } }
])


# Output:
Delhi – Bob: 300  
Delhi – Alice: 250  
Mumbai – Alice: 450  
Mumbai – Charlie: 150  
```

**Always use `$sort` after `$group` or `$project` if you're sorting grouped or reshaped data.**

## project

The `$project` stage is used to include, exclude, or transform fields in the output documents of an aggregation pipeline. It lets you reshape each document.

Shapes the output by:
- Including or excluding fields.
- Renaming fields.
- Creating new computed fields.
- Restructuring documents.

Works like a "custom output filter".

```python
# syntax:
{
  $project: {
    field1: 1,          // include
    field2: 0,          // exclude
    newField: <expression>, // create new/computed field
    _id: 0              // hide _id if needed
  }
}

# <expression> → can be a field, calculation, or operator.
```

```python
# sample users collction:
{ _id: 1, name: "Saurabh", age: 25, city: "Delhi" }
{ _id: 2, name: "Ravi", age: 30, city: "Mumbai" }
```

```python
# Include only name and age:
db.users.aggregate([
  { $project: { name: 1, age: 1, _id: 0 } }
])


# Output:
{ name: "Saurabh", age: 25 }
{ name: "Ravi", age: 30 }
```

```python
# Exclude the city field:
db.users.aggregate([
  { $project: { city: 0 } }
])


# Output:
{ _id: 1, name: "Saurabh", age: 25 }
{ _id: 2, name: "Ravi", age: 30 }
```

**Notes:**
- In the `$project` stage, you cannot mix inclusion (`1`) and exclusion (`0`) for fields in the same document, except for the `_id` field.
- `$project` is powerful when transforming or preparing data for reports or APIs.

```python
db.users.aggregate([
  {
    $project: {
      name: 1,     # include
      city: 0,     # exclude ❌
    }
  }
])
```

```python
db.users.aggregate([
  {
    $project: {
      name: 1,
      city: 1,
      _id: 0   # ✅ allowed
    }
  }
])
```

### Create new field

```python
# Create a new computed field isAdult:
db.users.aggregate([
  {
    $project: {
      name: 1,
      isAdult: { $gte: ["$age", 18] }  # returns true if age >= 18
    }
  }
])


# Output:
{ name: "Saurabh", isAdult: true }
{ name: "Ravi", isAdult: true }
```

```python
# Concatenate fields:
db.users.aggregate([
  {
    $project: {
      fullInfo: { $concat: ["Name: ", "$name", ", City: ", "$city"] },
      _id: 0
    }
  }
])


# Output:
{ fullInfo: "Name: Saurabh, City: Delhi" }
{ fullInfo: "Name: Ravi, City: Mumbai" }
```

### Create a Computed Field

```python
db.orders.aggregate([
  {
    $project: {
      _id: 0,
      item: 1,
      total: { $multiply: ["$price", "$quantity"] }
    }
  }
])
```

### Renaming fields or alias

```python
db.users.aggregate([
  {
    $project: {
      _id: 0,              # hide original _id
      "User id": "$_id",   # rename _id to User id
      name: 1,
      age: 1
    }
  }
])
```

### Include Nested Fields

```python
# data:
{ "item": "Apple", "details": { "color": "red", "weight": 150 } }

# You can project:
{ $project: { "details.color": 1, _id: 0 } }


# output:
{ "details": { "color": "red" } }
```

**Note:** `$project` happens after previous stages, so it only works on fields still present in the pipeline.

## addFields
- `$addFields` is a MongoDB aggregation stage that adds new fields to documents or modifies existing fields without removing other fields.

You use it inside an aggregation pipeline to transform your data by adding or updating fields.

```python
# syntax:
{
  $addFields: {
    newField: <expression>,
    existingField: <new value>
  }
}
```

```python
# students collection:
{
  "_id": 1,
  "firstName": "Alice",
  "lastName": "Parker",
  "math": 80,
  "science": 90
}


db.students.aggregate([
  {
    $addFields: {
      fullName: { $concat: ["$firstName", " ", "$lastName"] }, 
      total: { $add: ["$math", "$science"] }
    }
  }
])


# output:
{
  "_id": 1,
  "firstName": "Alice",
  "lastName": "Parker",
  "math": 80,
  "science": 90,
  "fullName": "Alice Parker",
  "total": 170
}
```

We don't want `firstName` and `lastName` also we want to change the order of fields dispalyed. We can do all these using `$project`:

```python
db.students.aggregate([
  {
    $addFields: {
      name: { $concat: ["$firstName", " ", "$lastName"] },
      total: { $add: ["$math", "$science"] }
    }
  },
  {
    $project: {
      _id: 1,
      name: 1,
      math: 1,
      science: 1,
      total: 1,
      firstName: 0,
      lastName: 0
    }
  }
])


# output:
{
  "_id": 1,
  "name": "Alice Parker",
  "math": 80,
  "science": 90,
  "total": 170
}
```

**Why this query will work?**

The “no mixing” rule applies only when you’re talking about existing fields from the input document.

In this case:
- `name` and `total` are not existing fields in the original collection — they were added by `$addFields` in the previous stage.
- `firstName` and `lastName` are existing fields, and you’re excluding them.
- MongoDB allows you to:
    - Include certain fields (original or newly computed)
    - Exclude some other original fields
    …as long as there’s no ambiguity in what the stage should return.

**💡 In other words:**
When some fields are computed in the same pipeline before `$project`, the rule is relaxed because MongoDB knows exactly which fields to include or exclude.

```python
db.students.aggregate([
  {
    $project: {
      firstName: 1,  // Inclusion mode
      lastName: 0    // ❌ Not allowed — mixing with exclusion
    }
  }
])
```

```python
db.students.aggregate([
  {
    $project: {
      firstName: 1,
      _id: 0 // allowed exception
    }
  }
])

# This works because _id is a special case.
```

```python
db.students.aggregate([
  {
    $addFields: {
      fullName: { $concat: ["$firstName", " ", "$lastName"] }
    }
  },
  {
    $project: {
      fullName: 1,
      math: 1,
      science: 1,
      _id: 0,
      firstName: 0, // original field excluded
      lastName: 0   // original field excluded
    }
  }
])
```

We can skip `$addFields` and do both adding new fields and hiding others inside a single `$project`:

```python
db.students.aggregate([
  {
    $project: {
      _id: 1,
      name: { $concat: ["$firstName", " ", "$lastName"] },
      math: 1,
      science: 1,
      total: { $add: ["$math", "$science"] }
    }
  }
])
```

**Another similar example:**

```python
# let's e have:
{ "_id": 1, "a": 10, "b": 20, "c": 30 }

# ❌ Illegal Mixing:
db.test.aggregate([
  { $project: { a: 1, b: 0 } }
])
```

This fails because `a` and `b` are both existing fields, and you’re trying to include one and exclude the other — MongoDB doesn’t allow that.

```python
db.test.aggregate([
  { $addFields: { sum: { $add: ["$a", "$b"] } } },
  { $project: { sum: 1, a: 1, b: 0 } }
])
```

This works because:
- `sum` is a new field (so including it doesn’t conflict with exclusion).
- `a` is included, `b` is excluded — MongoDB sees no ambiguity because the computed field is treated separately.

## limit

The `$limit` stage is used to restrict the number of documents that pass through the aggregation pipeline. It’s similar to the `LIMIT` clause in SQL.

```python
# syntax:
{ $limit: <number> }

# <number> = the maximum number of documents to return.
```

```python
# Sample products Collection:
{ _id: 1, name: "Pen", price: 10 }
{ _id: 2, name: "Notebook", price: 50 }
{ _id: 3, name: "Bag", price: 500 }
{ _id: 4, name: "Pencil", price: 5 }
{ _id: 5, name: "Mouse", price: 700 }
```

```python
# Get products 4 to 6:
db.products.aggregate([
  { $sort: { _id: 1 } },
  { $skip: 3 },
  { $limit: 3 }
])
```

**Notes:**

- It can appear anywhere in the pipeline, but usually comes after `$sort`.


**Use Case:**

- Fetching top-N records (e.g., top customers, top-selling products)

## skip

The `$skip` stage is used to skip over a specified number of documents in the aggregation pipeline. It's commonly used for pagination.

```python
# syntax:
{ $skip: <number> }

# <number> = the number of documents to skip.
```

```python
# Sample users Collection:
{ _id: 1, name: "Saurabh", age: 25 }
{ _id: 2, name: "Amit", age: 30 }
{ _id: 3, name: "Riya", age: 28 }
{ _id: 4, name: "Meena", age: 32 }
{ _id: 5, name: "Kunal", age: 26 }
```

```python
# Skip the first 2 documents:
db.users.aggregate([
  { $skip: 2 }
])

# Output (starts from the 3rd document):
{ _id: 3, name: "Riya", age: 28 }
{ _id: 4, name: "Meena", age: 32 }
{ _id: 5, name: "Kunal", age: 26 }
```

```python
# to get page 2 of results with 2 users per page:
const page = 2;
const limit = 2;
const skip = (page - 1) * limit;

db.users.aggregate([
  { $sort: { name: 1 } },
  { $skip: skip },
  { $limit: limit }
])


# in raw aggregation:
db.users.aggregate([
  { $sort: { name: 1 } },
  { $skip: 2 },
  { $limit: 2 }
])
```

**Use Case:**
- Pagination (skip + limit)
- Ignoring initial N documents
- Efficient for UI-based paged results (tables, lists, etc.)

## unwind

The `$unwind` stage is used to deconstruct an array field from the input documents and output one document for each element of the array.



```python
# syntax:
{ $unwind: "$<arrayField>" }


```

**We use $unwind when we need to:**
- Work with each element of an array individually.
- Perform filtering, grouping, or calculations on array elements.
- Join (`$lookup`) results where the data is nested in arrays.

```python
# Sample products Collection:
{
  _id: 1,
  name: "Laptop",
  tags: ["electronics", "tech", "device"]
}
```

```python
# Simple $unwind:
db.products.aggregate([
  { $unwind: "$tags" }
])


# Output:
{ _id: 1, name: "Laptop", tags: "electronics" }
{ _id: 1, name: "Laptop", tags: "tech" }
{ _id: 1, name: "Laptop", tags: "device" }
```

```python
# Include array index:
db.products.aggregate([
  {
    $unwind: {
      path: "$tags",
      includeArrayIndex: "tagIndex"
    }
  }
])


# Output:
{ _id: 1, name: "Laptop", tags: "electronics", tagIndex: 0 }
{ _id: 1, name: "Laptop", tags: "tech", tagIndex: 1 }
{ _id: 1, name: "Laptop", tags: "device", tagIndex: 2 }
```

```python
# Preserve null and empty arrays:
{
  _id: 2,
  name: "Notebook",
  tags: []
}

{
  _id: 3,
  name: "Pen"
}



db.products.aggregate([
  {
    $unwind: {
      path: "$tags",
      preserveNullAndEmptyArrays: true
    }
  }
])


# Output:
{ _id: 1, name: "Laptop", tags: "electronics" }
{ _id: 1, name: "Laptop", tags: "tech" }
{ _id: 1, name: "Laptop", tags: "device" }
{ _id: 2, name: "Notebook", tags: null }
{ _id: 3, name: "Pen", tags: null }
```

**Notes:**
- `$unwind` is very useful for array processing, filtering, or grouping.
- Combine with `$group`, `$match`, `$project` for powerful transformations.



### Advanced Options

MongoDB `$unwind` has an object form with extra options:

```python
# Optional (from MongoDB 3.2+):
{
  $unwind: {
    path: "$<arrayField>",
    includeArrayIndex: "indexField", // Optional: adds index of array item
    preserveNullAndEmptyArrays: true // Optional: keeps docs with empty/missing arrays
  }
}
```

```python
# students collection:
[
    { "_id": 1, "name": "Alice", "hobbies": ["reading", "swimming", "cycling"] }
    { "_id": 2, "name": "Bob", "hobbies": ["chess", "running"] }

]


db.students.aggregate([
  {
    $unwind: {
      path: "$hobbies",
      includeArrayIndex: "hobbyIndex",
      preserveNullAndEmptyArrays: true
    }
  }
])
```

```python
# output:

{ "_id": 1, "name": "Alice", "hobbies": "reading", "hobbyIndex": 0 }
{ "_id": 1, "name": "Alice", "hobbies": "swimming", "hobbyIndex": 1 }
{ "_id": 1, "name": "Alice", "hobbies": "cycling", "hobbyIndex": 2 }
{ "_id": 2, "name": "Bob", "hobbies": "chess", "hobbyIndex": 0 }
{ "_id": 2, "name": "Bob", "hobbies": "running", "hobbyIndex": 1 }
```

**When to use `$unwind`?**
- When you need to process each item in an array separately.
- When grouping, counting, or filtering array elements.
- Before joining array contents with another collection.

## Combining Multiple Stages

Typical Order of Aggregation Stages:

| Order | Stage                           | Purpose                                                                                      |
| ----- | ------------------------------- | -------------------------------------------------------------------------------------------- |
| 1     | **\$match**                     | Filter documents as early as possible to reduce the amount of data passed down the pipeline. |
| 2     | **\$project** / **\$addFields** | Include only needed fields or create new computed fields before heavy operations.            |
| 3     | **\$unwind**                    | Flatten arrays if needed before grouping or sorting.                                         |
| 4     | **\$group**                     | Aggregate data (sum, avg, count, etc.) based on keys.                                        |
| 5     | **\$project** (again)           | Reshape output after grouping (rename fields, hide group keys, etc.).                        |
| 6     | **\$sort**                      | Sort final or intermediate results.                                                          |
| 7     | **\$skip / \$limit**            | Implement pagination or reduce output size.                                                  |
| 8     | **\$lookup**                    | Join with other collections (do after filtering if possible).                                |
| 9     | **\$facet**                     | Run multiple pipelines in parallel for different types of aggregations.                      |
| 10    | **\$out / \$merge**             | Store the results in a collection.                                                           |


**Why This Order Helps**

1. Filter early (`$match`) → reduces workload for later stages.
2. Project early (`$project`) → removes unnecessary fields to speed up processing.
3. Unwind before grouping → ensures correct aggregation on array elements.
4. Sort and limit late → prevents large intermediate sorts.

```python
db.orders.aggregate([
  { $match: { status: "completed" } },   // Step 1: filter early
  { $project: { product: 1, price: 1, qty: 1 } },  // Step 2: keep only needed fields
  { $unwind: "$product" },               // Step 3: flatten arrays
  { $group: {                             // Step 4: aggregate
      _id: "$product",
      totalSales: { $sum: { $multiply: ["$price", "$qty"] } }
    }
  },
  { $sort: { totalSales: -1 } },          // Step 6: sort results
  { $limit: 5 }                           // Step 7: limit output
])
```

# String Operators

## regex

Used in queries to match string patterns using regular expressions.

```python
# syntax:
{ <field>: { $regex: /pattern/, $options: 'i' } }

# simpler version:
{ <field>: /pattern/i }


# pattern: The regular expression to match.
# options (optional):
    # 'i': case-insensitive
    # 'm': multi-line matching
    # 'x': ignore whitespace in regex
    # 's': allows . to match newline characters

db.collection.find({ field: { $regex: /pattern/, $options: 'i' } })

db.collection.aggregate([
  { $match: { field: { $regex: "pattern", $options: "i" } } }
])
```

```python
# Finds documents where name starts with "Saurabh"
db.users.find({ name: { $regex: /^Saurabh/ } })
```

**Caution:**
- Regex queries cannot use indexes efficiently (especially if not prefix regex like `^pattern`), which may lead to slower performance on large collections.
- You can use the `$options` flag in both formats: inside `$regex` or directly after it as a regex literal like `/pattern/i`.

**Used in the `find()` or `$match` stage to match string fields against a pattern.**

## regexMatch
**Use:** Inside `$project`, `$addFields`, etc. in the aggregation pipeline.
- Used when you want to evaluate if a string matches a regex as part of computed logic.

```python
# syntax:
{ $regexMatch: { input: <string>, regex: <regex pattern>, options: "i" } }
```

```python
db.users.aggregate([
  {
    $project: {
      name: 1,
      startsWithS: {
        $regexMatch: { input: "$name", regex: "^S", options: "i" }
      }
    }
  }
])

# This will add a field startsWithS with true or false depending on whether the name starts with S.
```

| Feature          | `$regex`                | `$regexMatch`                                   |
| ---------------- | ----------------------- | ----------------------------------------------- |

| Purpose          | Filter documents        | Evaluate condition (returns `true/false`)       |
| Returns          | Matched documents       | Boolean value                                   |
| Supports options | Yes (`$options`)        | Yes (`options` key)                             |
| Aggregation use  | Limited to `$match`     | Anywhere in pipeline expressions                |


<br>

**When to use which?**

| Scenario                                                     | Use           |
| ------------------------------------------------------------ | ------------- |
| Filter documents where name starts with `A`                  | `$regex`      |
| Create a field `isEmail` to check if string looks like email | `$regexMatch` |


**Quick Analogy:**
- `$regex` → "Find me the documents where this pattern matches" 📂
- `$regexMatch` → "Tell me whether this value matches the pattern, so I can use it in calculations or other conditions" ✅/❌

## cond

```python
# syntax:
{ $cond: { if: <condition>, then: <true-case>, else: <false-case> } }

# OR

# Short Syntax (with array form)
{
  $cond: [ <condition>, <value if true>, <value if false> ]
}
```

```python
# students collection:
[
  { name: "Alice", marks: 85 },
  { name: "Bob", marks: 40 },
  { name: "Charlie", marks: 60 }
]



db.students.aggregate([
  {
    $project: {
      name: 1,
      marks: 1,
      result: {
        $cond: { if: { $gte: ["$marks", 50] }, then: "Pass", else: "Fail" }
      }
    }
  }
])


# Output:
[
  { name: "Alice", marks: 85, result: "Pass" },
  { name: "Bob", marks: 40, result: "Fail" },
  { name: "Charlie", marks: 60, result: "Pass" }
]
```

```python
db.sales.aggregate([
  {
    $group: {
      _id: null,
      totalPenQty: {
        $sum: {
          $cond: [ { $eq: ["$item", "Pen"] }, "$quantity", 0 ]
        }
      }
    }
  }
])
```

# Variables

In MongoDB's aggregation framework, variables allow you to define and use temporary values within a stage, especially inside expressions like `$let`, `$map`, `$filter`, or `$reduce`.

**Why use variables?**
- Avoid repeating complex expressions.
- Make expressions easier to read and manage.
- Scope values locally for use inside expressions like `$project`, `$addFields`, `$group`, etc.

| Operator | Description                                      |
| -------- | ------------------------------------------------ |
| `$let`   | Defines temporary variables within an expression |
| `$$`     | Used to access variables (e.g., `$$myVar`)       |


## User defined variables
You define your own variables and then use them inside expressions.

```python
# syntax:
{
  $let: {
    vars: {
      <var1>: <expression1>,
      <var2>: <expression2>,
      ...
    },
    in: <expression using variables>
  }
}
```

```python
db.students.aggregate([
  {
    $project: {
      name: 1,
      finalScore: {
        $let: {
          vars: {
            average: { $avg: ["$marks.math", "$marks.science", "$marks.english"] }
          },
          in: { $round: ["$$average", 2] }
        }
      }
    }
  }
])
```

## System Variables

| Variable    | Description                                    |
| ----------- | ---------------------------------------------- |
| `$$ROOT`    | The top-level document                         |
| `$$CURRENT` | The current document at the level of the stage |
| `$$NOW`     | Current date/time                              |
| `$$REMOVE`  | Removes a field from output in `$project`      |


```python
# remove a field:
db.products.aggregate([
  {
    $project: {
      internalNotes: "$$REMOVE",  # this field will be removed
      name: 1,
      price: 1
    }
  }
])
```

**Tips:**
- Always use `$$` to reference variable values.
- You can nest `$let`, but keep it readable.
- `vars` are scoped only inside the expression defined by `in`.

# map
- `$map` is an aggregation operator used to transform each element of an array. It works like a loop that goes through every item in an array, applies a rule (expression), and returns a new array with the transformed values.

In simple words: `"$map lets you take an array, modify each item, and get a new array."`

```python
# syntax:
{
  $map: {
    input: <array>,                  # array where 
    as: <variable name>,
    in: <expression using variable>
  }
}

# input: The array you want to loop over.
# as: A variable name you’ll use to refer to each item (like x or item).
# in: What you want to return for each item in the array (can be a simple value or expression).
```

```python
# students collection:
[
  { name: "Alice", scores: [80, 90, 75] },
  { name: "Bob", scores: [60, 70, 85] }
]

# Add 5 bonus points to each score:
db.students.aggregate([
  {
    $project: {
      name: 1,
      updatedScores: {
        $map: {
          input: "$scores",
          as: "score",
          in: { $add: ["$$score", 5] }
        }
      }
    }
v  }
])


# output:
{ "_id": ..., "name": "Alice", "updatedScores": [85, 95, 80] }
{ "_id": ..., "name": "Bob", "updatedScores": [65, 75, 90] }
```

# Data Modelling

Data Modeling in MongoDB refers to the process of designing the structure of your documents, collections, and relationships between data to best suit your application’s requirements.

Unlike SQL databases, MongoDB uses a schema-less (or flexible schema) model, but good design is still crucial for performance, scalability, and maintainability.

MongoDB is a NoSQL database, it doesn't enforce strict schema relationships like foreign keys in relational databases.

We can still model relationships between documents in MongoDB using a few approaches.

The two main types of relationships are:
1. Embedded Documents (Denormalization)
2. Referenced Documents (Normalization)

| Term            | Description                                                   |
| --------------- | ------------------------------------------------------------- |
| **Document**    | The basic unit of data (like a row in SQL)                    |
| **Collection**  | A group of documents (like a table in SQL)                    |
| **Embedding**   | Nesting documents inside other documents                      |
| **Referencing** | Linking documents by storing their `_id` or another reference |


## Embedding 
- Store related data in the same document

```python
{
  _id: 1,
  name: "John Doe",
  email: "john@example.com",
  orders: [
    { item: "Book", qty: 2 },
    { item: "Pen", qty: 5 }
  ]
}
```

**Use When:**
- Data is frequently accessed together.
- Data has a one-to-few relationship.
- You want atomic updates (single write).

## Referencing 
- Store related data in separate documents

```python
# users collection
{
  _id: 1,
  name: "John Doe",
  email: "john@example.com"
}

# orders collection
{
  _id: 101,
  userId: 1,
  item: "Book",
  qty: 2
}
```

**Use When:**
- Data has a one-to-many or many-to-many relationship.
- Data is large and doesn't need to be loaded all at once.
- You want to avoid document size limits.

## Relationship

### One-to-One

```python
# Embedded:

{
  _id: 1,
  name: "Alice",
  profile: {
    age: 30,
    gender: "female"
  }
}


# Referenced:

# users
{ _id: 1, name: "Alice", profileId: 100 }

# profiles
{ _id: 100, age: 30, gender: "female" }
```

### One-to-Many

```python
# Embedded:

{
  _id: 1,
  title: "MongoDB Basics",
  comments: [
    { user: "A", text: "Great article!" },
    { user: "B", text: "Thanks!" }
  ]
}


# Referenced:

# posts
{ _id: 1, title: "MongoDB Basics" }

# comments
{ _id: 101, postId: 1, user: "A", text: "Great article!" }
```

### Many-to-Many

```python
# Embedded:
{
  _id: 1,
  name: "Saurabh",
  courses: [
    { _id: 101, name: "MongoDB" },
    { _id: 102, name: "FastAPI" }
  ]
}


# Referenced:

# students
{ _id: 1, name: "Saurabh", courseIds: [101, 102] }

# courses
{ _id: 101, name: "MongoDB" }
{ _id: 102, name: "FastAPI" }
```

# Join

In MongoDB, joins are performed using the `$lookup` stage in the Aggregation Framework. MongoDB doesn't support traditional SQL-style joins, but `$lookup` allows you to combine documents from two collections.

By default, `$lookup` performs a left outer join — meaning:
- All documents from the "main" collection are included.
- If there is a match in the "joined" collection, it’s added.
- If no match is found, you still get the original document, but the joined field will be an empty array.

## Basic syntax (localField–foreignField form)

```python
{
  $lookup: {
    from: "<other_collection_name>",
    localField: "<field_in_current_collection>",
    foreignField: "<field_in_other_collection>",
    as: "<name_for_new_array_field>"
  }
}
```

```python
# users collection:
{ _id: 1, name: "Saurabh", email: "saurabh@example.com" }
{ _id: 2, name: "Amit", email: "amit@example.com" }


# orders collection:
{ _id: 101, user_id: 1, item: "Laptop", total: 75000 }
{ _id: 102, user_id: 1, item: "Keyboard", total: 1500 }
{ _id: 103, user_id: 2, item: "Mouse", total: 700 }
```

**🔗 Goal:** Join `users` with their `orders` using `user_id`.

```python
db.users.aggregate([
  {
    $lookup: {
      from: "orders",           # foreign collection (to join)
      localField: "_id",        # field in users
      foreignField: "user_id",  # field in orders
      as: "orders"              # name of the new array field
    }
  }
])


# Output:
{
  _id: 1,
  name: "Saurabh",
  email: "saurabh@example.com",
  orders: [
    { _id: 101, user_id: 1, item: "Laptop", total: 75000 },
    { _id: 102, user_id: 1, item: "Keyboard", total: 1500 }
  ]
}
{
  _id: 2,
  name: "Amit",
  email: "amit@example.com",
  orders: [
    { _id: 103, user_id: 2, item: "Mouse", total: 700 }
  ]
}
```

```python
# Show only users with orders:
db.users.aggregate([
  {
    $lookup: {
      from: "orders",
      localField: "_id",
      foreignField: "user_id",
      as: "orders"
    }
  },
  {
    $match: {
      "orders.0": { $exists: true }
    }
  }
])
```

```python
# Show only selected fields:
db.users.aggregate([
  {
    $lookup: {
      from: "orders",
      localField: "_id",
      foreignField: "user_id",
      as: "orders"
    }
  },
  {
    $project: {
      name: 1,
      "orders.item": 1,
      "orders.total": 1
    }
  }
])
```

## Pipeline syntax (with let and pipeline)

This is the advanced form introduced in `MongoDB 3.6+`:

```python
{
  $lookup: {
    from: "<foreignCollection>",
    let: { <var1>: <expression1>, <var2>: <expression2>, ... }, // optional
    pipeline: [
      { <pipelineStage1> },
      { <pipelineStage2> },
      ...
    ],
    as: "<outputArrayField>"
  }
}
```

- from → Name of the foreign collection to join with.
- let → Defines variables from the input document that can be used inside the `pipeline` of the foreign collection.
- pipeline → An aggregation pipeline applied to the foreign collection, can reference variables via `$$<varName>`.
- as → Name of the new array field to store matching documents.

```python
db.orders.aggregate([
  {
    $lookup: {
      from: "customers",
      let: { customerId: "$cust_id" },
      pipeline: [
        { $match: { $expr: { $eq: ["$_id", "$$customerId"] } } }
      ],
      as: "customer_info"
    }
  }
])
```

# Schema Validation

In MongoDB, schema validation ensures that documents inserted or updated in a collection meet specific rules. While MongoDB is schema-less by default, you can define a schema using JSON Schema validation to enforce structure.

**Why Use Schema Validation?**
- Prevents inconsistent or bad data.
- Enforces data types and required fields.
- Ensures your application logic doesn't break due to unexpected document shapes.

## Basic Syntax
Schema validation is configured when creating or updating a collection using the `validator` option.

```python
db.createCollection("users", {
  validator: {
    $jsonSchema: {
      bsonType: "object",
      required: ["name", "email", "age"],
      properties: {
        name: {
          bsonType: "string",
          description: "must be a string and is required"
        },
        email: {
          bsonType: "string",
          pattern: "^.+@.+$",
          description: "must be a valid email and is required"
        },
        age: {
          bsonType: "int",
          minimum: 18,
          description: "must be an integer >= 18 and is required"
        }
      }
    }
  },
  validationLevel: "strict",     # "strict" or "moderate"
  validationAction: "error"      # "error" or "warn"
})
```

**Validation Levels:**
- `strict`: The document must fully comply with the schema validation rules. If a document does not comply, it will not be inserted or updated in the collection.
- `moderate`: Only new documents and modified fields in existing documents are validated against the schema. This allows for partial validation and can be useful for legacy systems or gradual schema enforcement.

**Validation Actions:**
- `error`: If a document does not meet the schema validation criteria, MongoDB will throw an error and reject the insert or update operation.
- `warn`: MongoDB logs a warning message when a document does not meet the schema validation criteria but still allows the insert or update operation.

| Field                | Purpose                                                                                     |
| -------------------- | ------------------------------------------------------------------------------------------- |
| `bsonType`           | Enforce MongoDB data types like `string`, `int`, `bool`, `array`, etc.                      |
| `required`           | Ensures listed fields must exist.                                                           |
| `pattern`            | Regular expression (e.g., for email).                                                       |
| `minimum`, `maximum` | For numeric ranges.                                                                         |
| `validationLevel`    | `"strict"` validates all inserts/updates, `"moderate"` skips updates with unchanged fields. |
| `validationAction`   | `"error"` rejects invalid docs, `"warn"` logs a warning but allows.                         |

# Indexes

Indexes in MongoDB are special data structures that improve the speed of search queries on collections — similar to indexes in books that help you quickly find content.

Without indexes, MongoDB must scan every document in a collection (a full collection scan), which is inefficient.

**Why Use Indexes?**
- Faster search/query performance
- Efficient sorting and filtering
- Required for unique constraints
- Optimize aggregation and joins

**Types of Indexes in MongoDB:**

| Type             | Description                           |
| ---------------- | ------------------------------------- |
| **Single Field** | Indexes a single field.               |
| **Compound**     | Index on multiple fields.             |
| **Multikey**     | Indexes array fields.                 |
| **Text**         | For full-text search.                 |
| **Hashed**       | For sharding using hashed values.     |
| **Wildcard**     | Dynamic keys in documents.            |
| **Geospatial**   | For location-based queries.           |
| **Unique Index** | Prevents duplicate values in a field. |


## Single Field Index

```python
db.users.createIndex({ name: 1 }) # ascending



# Speeds up queries like:
db.users.find({ name: "Saurabh" })
```

## Compound Index

```python
db.orders.createIndex({ userId: 1, orderDate: -1 })



# Useful for queries involving both fields:
db.orders.find({ userId: 101 }).sort({ orderDate: -1 })
```

## Unique Index

```python
db.users.createIndex({ email: 1 }, { unique: true })



# Prevents duplicate emails:
db.users.insertOne({ email: "saurabh@example.com" }) // ✅
db.users.insertOne({ email: "saurabh@example.com" }) // ❌ Error
```

## Text Index

A Text Index allows you to search for words or phrases in string fields, just like searching in Google.

- Use it when you want to do full-text search in fields like `title`, `description`, or `content`.

```python
# blog collection:
{
  _id: 1,
  title: "Learn MongoDB",
  content: "MongoDB is a NoSQL database used for modern applications."
}

# Step 1: Create a Text Index
db.blogs.createIndex({ content: "text" }) # This tells MongoDB: “Hey, I want to search inside the content field like it's a paragraph of text.”


# Step 2: Search with $text
db.blogs.find({ $text: { $search: "mongodb indexing" } }) # MongoDB will now search for "MongoDB" or "NoSQL" in the content field.
```

**Notes:**
- You can only have one text index per collection (but it can include multiple fields).
- MongoDB will automatically tokenize (split) the content and ignore common words like "is", "the", etc.

```python
# Index on Multiple Fields:
db.blogs.createIndex({
  title: "text",
  content: "text"
})


db.blogs.find({ $text: { $search: "MongoDB" } }) # Now it searches both title and content

db.blogs.find({ $text: { $search: "\"NoSQL database\"" } }) # Use double quotes to match the exact phrase
```

## Multikey Index (on arrays)
A multikey index is used when a field contains an array.

- It allows MongoDB to index each element in the array so that it can search quickly inside it.

```python
# students collection:
{
  _id: 1,
  name: "Saurabh",
  subjects: ["Math", "Science", "English"] # array
}

# Create a Multikey Index
db.students.createIndex({ subjects: 1 })

# This creates a multikey index, because MongoDB sees that subjects is an array and creates an index for 
# each value like:
# "Math"
# "Science"
# "English"



# Querying documents where subjects is an array:
db.students.find({ subjects: "Math" }) # MongoDB will use the index to find documents where "Math" is in the array
```

## Hashed Index (used in sharding)

A hashed index is used when you want to distribute data randomly based on the hash of a field’s value.

- Instead of indexing the actual value, MongoDB converts it into a hash (a fixed-size random-looking number) and indexes that.

**Why use it?**
    
Hashed indexes are mainly used for:
- Sharding (splitting data across servers)
- Evenly distributing documents
- Fast exact-match queries



```python
# users collection:
{ _id: 1, username: "saurabh" }
{ _id: 2, username: "priya" }


# create a hashed index on username like this:
db.users.createIndex({ username: "hashed" })
# MongoDB internally converts "saurabh" and "priya" to hashed values and builds the index on those.

# Good for:
db.users.find({ username: "saurabh" }) # This is an exact-match query, and hashed index helps here.

# Not good for:
db.users.find({ username: { $gt: "a" } }) // range queries
# Hashed indexes cannot be used for range queries or sorting

# Ensures even data distribution in sharded clusters.
```

## CRUD using Python 

```python
from pymongo import MongoClient
from bson.objectid import ObjectId

# 1. Connect to MongoDB
client = MongoClient("mongodb://localhost:27017/")
db = client["user_db"]
collection = db["users"]

# 2. Create
def create_user(name, email, age):
    user = {
        "name": name,
        "email": email,
        "age": age
    }
    result = collection.insert_one(user)
    print(f"✅ User created with _id: {result.inserted_id}")

# 3. Read
def read_users():
    users = collection.find()
    for user in users:
        print(user)

# 4. Update
def update_user(user_id, new_data):
    result = collection.update_one(
        {"_id": ObjectId(user_id)},
        {"$set": new_data}
    )
    if result.modified_count > 0:
        print("✅ User updated.")
    else:
        print("❌ No user updated.")

# 5. Delete
def delete_user(user_id):
    result = collection.delete_one({"_id": ObjectId(user_id)})
    if result.deleted_count > 0:
        print("✅ User deleted.")
    else:
        print("❌ No user deleted.")

# 6. Main Menu
def main():
    while True:
        print("\n=== MongoDB CRUD Menu ===")
        print("1. Create User")
        print("2. Read Users")
        print("3. Update User")
        print("4. Delete User")
        print("5. Exit")

        choice = input("Enter choice: ")

        if choice == "1":
            name = input("Name: ")
            email = input("Email: ")
            age = int(input("Age: "))
            create_user(name, email, age)

        elif choice == "2":
            read_users()

        elif choice == "3":
            user_id = input("Enter user _id to update: ")
            name = input("New Name: ")
            email = input("New Email: ")
            age = int(input("New Age: "))
            update_user(user_id, {"name": name, "email": email, "age": age})

        elif choice == "4":
            user_id = input("Enter user _id to delete: ")
            delete_user(user_id)

        elif choice == "5":
            print("Exiting...")
            break

        else:
            print("❌ Invalid choice.")

if __name__ == "__main__":
    main()
```
