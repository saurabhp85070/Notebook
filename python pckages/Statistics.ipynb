# Hypothesis Testing

Hypothesis testing is a statistical method used to make decisions or inferences about a population based on a sample of data.

Imagine you're a detective trying to solve a case based on evidence (data). Hypothesis testing is the tool you use to decide whether the evidence supports your theory (hypothesis) or not.

## Why We Use Hypothesis Testing?

1. **Uncertainty in Data:** In real-world scenarios, we often don't have access to data from the entire population (e.g., all people, products, etc.). Instead, we collect a sample, which may not perfectly represent the whole population.

    Hypothesis testing allows us to make inferences or draw conclusions about the population based on this sample, while accounting for the possibility of error.

2. **Objective Decision Making:** When we want to test a claim or belief (e.g., whether a new medication is more effective than an old one), hypothesis testing gives us a structured framework to evaluate the evidence, rather than relying on gut feelings or guesses.

3. **Quantifying Uncertainty:** Hypothesis testing uses probabilities to measure uncertainty. By calculating a p-value and comparing it to a predefined threshold (significance level, α), we can objectively determine whether the observed data is consistent with our assumptions or whether we have enough evidence to reject them.


## ✅ Null Hypothesis (H₀)

The null hypothesis (H₀) is the default assumption that `nothing has changed`, or there is `no effect`, `no difference`, or `no relationship`.

Think of it as: “Everything is normal.”

**Examples:**

| Scenario                       | Null Hypothesis (H₀)                               |
| ------------------------------ | -------------------------------------------------- |
| A new medicine test            | The medicine has **no effect**.                    |
| A company's product weight     | The average bottle contains **500ml**, as claimed. |
| Comparing two teaching methods | Both methods are **equally effective**.            |


## ❌ Alternative Hypothesis (H₁ or Ha)

The alternative hypothesis is the claim you want to prove — that `something has changed`, or `there is an effect`, `difference`, or `relationship`.

Think of it as: “Something is happening.”

**Examples:** 

| Scenario                       | Alternative Hypothesis (H₁)              |
| ------------------------------ | ---------------------------------------- |
| A new medicine test            | The medicine **does have an effect**.    |
| A company's product weight     | The average bottle is **not 500ml**.     |
| Comparing two teaching methods | One method is **better** than the other. |


**H₀ vs H₁**

| Feature         | Null Hypothesis (H₀)               | Alternative Hypothesis (H₁)     |
| --------------- | ---------------------------------- | ------------------------------- |
| Meaning         | No change or effect                | There **is** a change or effect |
| Role            | Default assumption                 | The claim you're testing        |
| Action          | You try to reject this             | You try to support this         |
| Evidence Needed | Strong evidence required to reject | Accepted if H₀ is rejected      |

## Basic Workflow of Hypothesis Testing

1. **State the Hypotheses:** Define the null hypothesis (H₀) and the alternative hypothesis (H₁). The null hypothesis represents the default assumption (e.g., no effect or no difference), while the alternative hypothesis represents what you're testing for (e.g., an effect or difference).

2. **Choose the Significance Level (α):** This is the probability threshold for rejecting the null hypothesis. Common choices are `α = 0.05` or `α = 0.01`. A significance level of 0.05 means that you're willing to accept a 5% chance of rejecting the null hypothesis when it is actually true (Type I error).

3. **Collect and Analyze Data:** Gather your sample data and perform the necessary statistical test (e.g., z-test, t-test). This will yield a test statistic, such as a z-score or t-score.

4. **Calculate the P-Value:** The p-value represents the probability of observing your data (or something more extreme) if the null hypothesis is true. A small p-value indicates that your sample data is unlikely under the null hypothesis.

    Smaller p-value = stronger evidence against the null hypothesis.

5. **Make a Decision:** Compare the p-value to α:
    - If the p-value is less than α, reject the null hypothesis. This means the data provides enough evidence to support the alternative hypothesis.
    - If the p-value is greater than α, fail to reject the null hypothesis. This means there is not enough evidence to support the alternative hypothesis.


#### P-value

The p-value represents the probability of observing your data (or something more extreme) if the null hypothesis is true. 

The p-value tells you how likely it is to see your data (or something even more extreme) if nothing special is going on (if the null hypothesis is true).

In other words, it asks: `"If the null hypothesis is true (nothing’s happening), how unusual is this data?"`

**Let's understand this with example:**

Imagine you flip a coin 10 times and it lands on heads 9 times. You might be wondering: "Is this just normal random luck, or is the coin biased?"

The null hypothesis (H₀) would be: "The coin is fair. Heads and tails are equally likely."

Now, the p-value tells you: "If the coin really is fair, how likely is it that I’d get 9 heads in 10 flips?"
- If the p-value is small, like 0.01, this means it’s very unlikely to get 9 heads by chance if the coin were fair.
- If the p-value is big, like 0.50, this means it’s pretty likely to get 9 heads by chance even if the coin is fair.

So, let's say we get p-value as 0.02 tells you: `"If the coin were really fair, there’s only a 2% chance that you’d get 9 heads (or something more extreme, like 10 heads) in 10 flips."`

#### Significance Level (α)
This is the probability threshold for rejecting the null hypothesis. The significance level is a rule you set before doing the test. It’s written as α (alpha).
It tells you how much risk you’re willing to take when deciding that your result is not due to chance.

Let's say we set `α = 0.05`, You are willing to accept a 5% chance of being wrong when you say “The coin is not fair.”

In other words: `“If my p-value is less than 0.05, I’ll reject the idea that the coin is fair — even though I know there’s a 5% chance I might be making a mistake.”`

Significance level (0.05) is the cutoff you choose. It means you’re okay with making a wrong decision 5% of the time when rejecting the null hypothesis. If your p-value is smaller than 0.05, you say: “This result is rare under the assumption of fairness — I reject the null hypothesis.”

**🧠 So what does α = 0.05 mean here?** It means: “If the coin is really fair, I will only say it’s biased if there’s less than a 5% chance of getting this result by luck.”

In other words: “I’m okay with being wrong 5 times out of 100 if I say the coin is biased when it’s actually fair.” This type of mistake is called a Type I Error — falsely accusing the coin of being unfair.

Since 0.02 < 0.05 (p-value < α):You say: “This result is too rare to happen just by chance — I think the coin is not fair.” You are accepting the risk that you might be wrong 5% of the time when saying this.               
If your p-value had been 0.10, then: 0.10 > 0.05 → Not significant, then you say: “This result could easily happen by chance — I won’t say the coin is biased.”



## Errors in Hypothesis Testing

| Type    | Description                                       |
| ------- | ------------------------------------------------- |
| Type I  | Rejecting H₀ when it's true (False Positive)      |
| Type II | Not rejecting H₀ when it's false (False Negative) |


### Type I Error (False Positive)

- Rejecting the null hypothesis when it is actually true (e.g., concluding a drug is effective when it is not).
- The probability of making a Type I error is equal to the significance level (α).


### Type II Error (False Negative) 
- Failing to reject the null hypothesis when the alternative hypothesis is true (e.g., concluding a drug is not effective when it is).
- The probability of making a Type II error is denoted by β, and 1 - β is called the power of the test (the probability of correctly rejecting the null hypothesis when the alternative is true).


## How to Formulate the Hypotheses?

The process of formulating null and alternative hypotheses depends on the objective of your test.
    
Step-by-Step Process:
    
1. **Start with a Research Question:** What do you want to investigate? For example, do you want to check if a new medication works better than the current one, or if there's a difference between two groups?

2. **Identify the Variables:** Determine the population parameter (`mean`, `proportion`, etc.) that you are interested in and the conditions you want to test.

3. **Formulate the Null Hypothesis (H₀):**
    - Assume that there is no effect or no difference.
    - The null hypothesis `should be an equality statement` (e.g., μ=0, p=0.5 etc.).
    
4. **Formulate the Alternative Hypothesis (H₁):**
    - This represents what you want to test for.
    - It’s `typically an inequality statement` (e.g., μ≠0, p≠0.5, μ>0, etc.).


### Example: Medical Testing
**Scenario:** A researcher wants to test if a new drug is more effective than the current drug in lowering blood pressure. The current drug lowers blood pressure by an average of 10 mmHg.
    
**Null Hypothesis (H₀):** The new drug does not lower blood pressure more than the current drug.
			`H₀ : μ=10 mmHg`

    Here, the null assumes that the mean reduction in blood pressure for the new drug is the same as the old one.

**Alternative Hypothesis (H₁):** The new drug lowers blood pressure more than the current drug.
			`H₁ : μ>10 mmHg`

    This is a one-tailed test because you are specifically testing if the new drug lowers blood pressure more (not less) than the old one.


### Example: Quality Control

**Scenario:** A factory claims that their machines produce light bulbs with a lifespan of 1,000 hours on average. You suspect that the actual lifespan might be different (either higher or lower).

**Null Hypothesis (H₀):** The mean lifespan of the light bulbs is 1,000 hours.
			`H₀ : μ=1000 hours`

    The null hypothesis assumes no difference from the claimed average lifespan.

**Alternative Hypothesis (H₁):** The mean lifespan of the light bulbs is not 1,000 hours.
			`H₁ : μ≠1000 hours`

    This is a two-tailed test because you are testing whether the actual lifespan is either higher or lower than 1000 hours.


### Example: : Product Testing

**Scenario:** A company manufactures batteries and claims that their batteries last at least 300 hours. A customer wants to test if this claim is true.
    
**Null Hypothesis (H₀):** The batteries last at least 300 hours.
			`H₀ : μ≥300 hours`

    The null hypothesis assumes that the mean battery life is at least 300 hours.
        
**Alternative Hypothesis (H₁):** The batteries last less than 300 hours.
			`H₁ : μ<300 hours`

    This is a one-tailed test because you are specifically testing if the battery life is less than the claimed 300 hours.

**The alternative hypothesis can be either two-tailed or one-tailed:**
- `Two-tailed:` Tests for any significant difference, either higher or lower than a specified value.
- `One-tailed:` Tests for a difference in one specific direction (e.g., greater than or less than a value).


| Type of Test   | Null Hypothesis (H₀) | Alternative Hypothesis (H₁) | Direction               | When to Use                         |
| -------------- | -------------------- | --------------------------- | ----------------------- | ----------------------------------- |
| **Two-Tailed** | μ = value            | μ ≠ value                   | Any difference (↑ or ↓) | To detect any significant change    |
| **One-Tailed** | μ = value / ≥ / ≤    | μ > value **or** μ < value  | Specific direction      | To detect increase or decrease only |


## Guidelines for Formulating Hypotheses

1. **What is your claim or assumption?**

    - The null hypothesis (H₀) reflects the existing belief, status quo, or no effect.
    - It's the hypothesis you try to test against.

    **🔹 Example:**
        A company claims the average battery life is 300 hours.
        → `H₀: μ = 300` (You assume the claim is true unless proven otherwise.)

2. **What do you want to prove?**

- The alternative hypothesis (H₁) represents what you want to demonstrate with evidence—a change, improvement, or difference.
- It's what you hope to support with your test results.

    **🔹 Example:**
        You believe the battery actually lasts less than 300 hours.
        → `H₁: μ < 300`

3. **Use equality in the null hypothesis**
- The null hypothesis should use an equality sign (e.g., =, ≥, ≤).
- This is because hypothesis testing works by assuming H₀ is true, and then using evidence to reject it if necessary.

    🔹 Correct Format Examples: `H₀: μ = 100`, `H₀: p ≥ 0.5`, `H₀: μ ≤ 5`

    **📌 Note:** You never write the alternative hypothesis with an equality (you want to detect a change, not assume it).

4. **Choose the correct alternative hypothesis**

    Choose based on your research question:

    - **🔸 Two-Tailed Test**
        `Use H₁: μ ≠ value`
        
        When you care if the result is either higher or lower than expected.
        
        🔹 Example:
        Testing if the average score is different from 70.
        → `H₁: μ ≠ 70`

    
    - **🔸 One-Tailed Test**
        `Use H₁: μ > value` or `H₁: μ < value`
        
        When you only care if the result is greater than or less than expected.
        
        🔹 Example:
        You believe a new fertilizer increases yield.
        → `H₁: μ > 50` (Right-tailed)
    
        You suspect a machine is producing less weight than it should.
        → `H₁: μ < 500g` (Left-tailed)

<br>

| Step | Guideline                    | Example                                           |
| ---- | ---------------------------- | ------------------------------------------------- |
| 1    | State assumption (null)      | H₀: μ = 300                                       |
| 2    | State what you want to prove | H₁: μ < 300                                       |
| 3    | Use equality in H₀           | H₀: μ ≥ 0.5                                       |
| 4    | Choose correct H₁ direction  | H₁: μ ≠ 100 (2-tailed), or H₁: μ > 100 (1-tailed) |


### One-Tailed Test Example (Right-Tailed):

**Research Question:** Are the batteries lasting more than 100 hours on average?

**Hypotheses:**
- Null hypothesis (H₀): The mean battery life is 100 hours or less (μ≤100).
- Alternative hypothesis (H₁): The mean battery life is greater than 100 hours (μ>100).

**Interpretation:**
- If the test statistic (e.g., z-score or t-score) falls into the right tail of the distribution (i.e., it’s much greater than expected under H₀), you would reject the null hypothesis and conclude that the batteries last more than 100 hours.

### Two-Tailed Test Example:
**Research Question:** Are the batteries lasting exactly 100 hours on average, or is the lifespan different (could be either more or less)?

**Hypotheses:**
- Null hypothesis (H₀): The mean battery life is exactly 100 hours (μ=100).
- Alternative hypothesis (H₁): The mean battery life is not equal to 100 hours (μ≠100).

**Interpretation:**
- Here, you are testing for a difference in either direction (more than 100 hours or less than 100 hours).
- If the test statistic falls into either tail of the distribution (i.e., it’s much greater or much smaller than expected under H₀), you would reject the null hypothesis and conclude that the battery life is different from 100 hours.

## Python Implementation

### Two-Tailed Test

**Scenario:**
A bottling company claims that the average volume in its soda bottles is 500 ml. A quality inspector suspects it might be different (could be more or less) and collects a sample.

```python
import numpy as np
from scipy import stats

# Sample data collected by the quality inspector (example values in ml)
sample_volumes = [498, 502, 497, 503, 499, 500, 496, 504, 501, 497]

# Hypothesized population mean (claimed by company)
mu = 500

# Calculate sample statistics
sample_mean = np.mean(sample_volumes)
sample_std = np.std(sample_volumes, ddof=1)  # sample standard deviation
n = len(sample_volumes)

# Perform one-sample t-test
t_statistic, p_value = stats.ttest_1samp(sample_volumes, mu)

# Set significance level (alpha)
alpha = 0.05

# Display results
print(f"Sample Mean = {sample_mean:.2f} ml")
print(f"Sample Standard Deviation = {sample_std:.2f} ml")
print(f"T-Statistic = {t_statistic:.4f}")
print(f"P-Value = {p_value:.4f}")

# Conclusion
if p_value < alpha:
    print("Reject the null hypothesis (H₀). The average volume is significantly different from 500 ml.")
else:
    print("Fail to reject the null hypothesis (H₀). There's no significant evidence that the average volume is different from 500 ml.")
```

**Result:**
- Sample Mean (x̄) = 499.70 ml
- Hypothesized Mean (μ) = 500 ml
- t-statistic = -0.3449
- p-value = 0.7381
- Significance Level (α) = 0.05 (typically used unless stated otherwise)

**🔍 Interpretation:**
- The p-value (0.7381) is much greater than α = 0.05, which means: You fail to reject the null hypothesis.

**🧾 Conclusion:**
- There is no statistically significant evidence to suggest that the average soda bottle volume is different from 500 ml.
- The small difference in the sample mean (499.8 ml) is likely due to random variation, not a true difference in the population mean.

**✅ What It Means Practically:**
- The company's claim (μ = 500 ml) holds up based on this sample. No action is needed based on this data.

### One-Tailed Test
**Scenario:**
A gym trainer claims his new diet plan helps clients gain more than 2 kg of muscle in a month. A client measures gains of 10 people and wants to test if it's significantly more than 2 kg.

```python
import numpy as np
from scipy import stats

# Sample muscle gains of 10 clients (example values in kg)
muscle_gains = [2.3, 2.1, 2.5, 2.6, 2.8, 2.4, 1.9, 2.7, 2.9, 2.2]

# Claimed average muscle gain (null hypothesis mean)
mu = 2

# Sample statistics
sample_mean = np.mean(muscle_gains)
sample_std = np.std(muscle_gains, ddof=1)
n = len(muscle_gains)

# One-sample t-test (two-tailed initially)
t_statistic, p_value_two_tailed = stats.ttest_1samp(muscle_gains, mu)

# Convert to one-tailed p-value for right-tailed test
if t_statistic > 0:
    p_value = p_value_two_tailed / 2
else:
    p_value = 1 - (p_value_two_tailed / 2)

# Set significance level
alpha = 0.05

# Display results
print(f"Sample Mean = {sample_mean:.2f} kg")
print(f"Sample Standard Deviation = {sample_std:.2f} kg")
print(f"T-Statistic = {t_statistic:.4f}")
print(f"One-tailed P-Value = {p_value:.4f}")

# Conclusion
if p_value < alpha:
    print("Reject the null hypothesis (H₀). The diet plan significantly helps in gaining more than 2 kg of muscle.")
else:
    print("Fail to reject the null hypothesis (H₀). There is not enough evidence to support the trainer's claim.")
```

**Result:**
- Sample Mean (x̄) = 2.44
- Hypothesized Mean (μ) = let's assume μ = 2.0 (based on context)
- t-statistic = 4.3425
- p-value (one-tailed) = 0.0009
- Significance Level (α) = 0.05

**🔍 Interpretation:**
- The p-value (0.0009) is less than the significance level (α = 0.05).
- You reject the null hypothesis.

**🧾 Conclusion:**
- There is strong statistical evidence to suggest that the true mean is significantly greater than 2.0.
The sample mean of 2.44 supports the claim that the actual average is higher than 2.0.

**✅ Practical Meaning:**
    
Let’s say this data represents the average customer rating of a new product, and your null hypothesis was:

    - H₀: μ ≤ 2.0 (the product performs poorly)
    - H₁: μ > 2.0 (the product performs better than average)

Since we rejected H₀, we conclude: The product is rated significantly higher than average, and the higher mean rating is unlikely due to chance.

**More detailed**

```python
import numpy as np
import scipy.stats as stats
import matplotlib.pyplot as plt
import seaborn as sns

# Sample data - muscle gains in kg
# In a real scenario, this would come from actual measurements
muscle_gains = np.array([2.3, 2.1, 2.5, 2.6, 2.8, 2.4, 1.9, 2.7, 2.9, 2.2])

# The null hypothesis value (claimed threshold)
mu = 2.0

# 1. Basic statistics
print("Basic Statistics:")
print(f"Sample size: {len(muscle_gains)}")
print(f"Mean muscle gain: {np.mean(muscle_gains):.2f} kg")
print(f"Standard deviation: {np.std(muscle_gains, ddof=1):.2f} kg")
print(f"Minimum: {np.min(muscle_gains):.2f} kg")
print(f"Maximum: {np.max(muscle_gains):.2f} kg")
print()

# 2. One-sample t-test
# H0: μ <= 2 kg (the diet doesn't lead to gains greater than 2 kg)
# Ha: μ > 2 kg (the diet leads to gains greater than 2 kg)
t_stat, p_value = stats.ttest_1samp(muscle_gains, mu)

# Since scipy.stats.ttest_1samp gives two-sided p-value, we need to adjust for one-sided test
# If the mean of the sample is greater than mu, we divide the p-value by 2
# If the mean of the sample is less than mu, the one-sided p-value is 1 - (p_value/2)
if np.mean(muscle_gains) > mu:
    p_value_one_sided = p_value / 2
else:
    p_value_one_sided = 1 - (p_value / 2)

print("Hypothesis Test Results:")
print(f"t-statistic: {t_stat:.4f}")
print(f"Two-sided p-value: {p_value:.4f}")
print(f"One-sided p-value (Ha: μ > {mu}): {p_value_one_sided:.4f}")
print()

# 3. Decision
alpha = 0.05
if p_value_one_sided < alpha:
    print(f"Decision: Since p-value ({p_value_one_sided:.4f}) < alpha ({alpha}), we reject the null hypothesis.")
    print("Conclusion: There is significant evidence that the diet plan helps clients gain more than 2 kg of muscle in a month.")
else:
    print(f"Decision: Since p-value ({p_value_one_sided:.4f}) >= alpha ({alpha}), we fail to reject the null hypothesis.")
    print("Conclusion: There is not enough evidence to conclude that the diet plan helps clients gain more than 2 kg of muscle in a month.")
print()

# 4. Confidence Interval
n = len(muscle_gains)
mean = np.mean(muscle_gains)
std_err = stats.sem(muscle_gains)  # Standard error using built-in function
ci = stats.t.interval(0.95, n-1, loc=mean, scale=std_err)  # 95% confidence interval
print(f"95% Confidence Interval: ({ci[0]:.2f}, {ci[1]:.2f}) kg")
print()

# 5. Effect size - Cohen's d
cohens_d = (mean - mu) / np.std(muscle_gains, ddof=1)
print(f"Effect size (Cohen's d): {cohens_d:.2f}")

# Interpret effect size
if abs(cohens_d) < 0.2:
    effect = "negligible"
elif abs(cohens_d) < 0.5:
    effect = "small"
elif abs(cohens_d) < 0.8:
    effect = "medium"
else:
    effect = "large"
print(f"This indicates a {effect} effect size.")
print()

# 6. Visualizations
plt.figure(figsize=(12, 10))

# Histogram with density
plt.subplot(2, 2, 1)
sns.histplot(muscle_gains, kde=True)
plt.axvline(mu, color='red', linestyle='--', label=f'H0: μ = {mu} kg')
plt.axvline(mean, color='green', label=f'Sample mean: {mean:.2f} kg')
plt.title('Distribution of Muscle Gains')
plt.xlabel('Muscle Gain (kg)')
plt.ylabel('Frequency')
plt.legend()

# Box plot
plt.subplot(2, 2, 2)
sns.boxplot(y=muscle_gains)
plt.axhline(mu, color='red', linestyle='--', label=f'H0: μ = {mu} kg')
plt.title('Box Plot of Muscle Gains')
plt.ylabel('Muscle Gain (kg)')

# Individual data points
plt.subplot(2, 2, 3)
plt.scatter(range(1, n+1), muscle_gains)
plt.axhline(mu, color='red', linestyle='--', label=f'H0: μ = {mu} kg')
plt.axhline(mean, color='green', label=f'Sample mean: {mean:.2f} kg')
plt.fill_between(range(1, n+1), ci[0], ci[1], color='green', alpha=0.2, label=f'95% CI: ({ci[0]:.2f}, {ci[1]:.2f})')
plt.title('Individual Muscle Gains')
plt.xlabel('Participant')
plt.ylabel('Muscle Gain (kg)')
plt.legend()

# QQ plot to check normality
plt.subplot(2, 2, 4)
stats.probplot(muscle_gains, plot=plt)
plt.title('QQ Plot (Check for Normality)')

plt.tight_layout()
plt.show()

# 7. Power analysis
from statsmodels.stats.power import TTestPower
power_analysis = TTestPower()

# Calculate the power of the current test
power = power_analysis.power(effect_size=cohens_d, nobs=n, alpha=alpha, alternative='larger')
print(f"Statistical power of the test: {power:.2f} (or {power*100:.0f}%)")

# Calculate required sample size for 80% power
required_n = power_analysis.solve_power(effect_size=cohens_d, alpha=alpha, power=0.8, alternative='larger')
print(f"Sample size needed for 80% power: {int(np.ceil(required_n))}")
```

### Python Implementation

**📘 Scenario:**
Let’s say we want to test whether the average weight of a population is 70 kg.
We take a sample of 50 people and find their average weight is 73 kg, with a known population standard deviation of 10 kg.

We’ll test this at significance level α = 0.05 (one-tailed test: is the sample mean greater than the population mean?)

```python
import math
from scipy.stats import norm

# Known values
population_mean = 70
population_std_dev = 10
sample_mean = 73
sample_size = 50
alpha = 0.05  # Significance level

# Step 1: Calculate Z-score
standard_error = population_std_dev / math.sqrt(sample_size)
z_score = (sample_mean - population_mean) / standard_error

# Step 2: Calculate p-value (one-tailed test)
p_value = 1 - norm.cdf(z_score)  # Right-tailed test (mean > population mean)

# Step 3: Compare with alpha
print(f"Z-score: {z_score:.2f}")
print(f"p-value: {p_value:.4f}")

if p_value < alpha:
    print("✅ Reject the null hypothesis: Sample mean is significantly greater than population mean.")
else:
    print("❌ Fail to reject the null hypothesis: No significant difference.")
```

**Key Correct Elements**

- The subtraction order (p1 - p2) doesn't matter for the conclusion in a two-tailed test, though the negative z-score correctly indicates that p2 > p1
- Two-tailed p-value calculation is appropriate when testing for any difference between proportions

```python
import math
from scipy.stats import norm

# Data for Version A
n1 = 200       # sample size A
x1 = 50        # successes A

# Data for Version B
n2 = 180       # sample size B
x2 = 72        # successes B

# Sample proportions
p1 = x1 / n1
p2 = x2 / n2

# Combined (pooled) proportion
p_pool = (x1 + x2) / (n1 + n2)

# Standard error
se = math.sqrt(p_pool * (1 - p_pool) * (1/n1 + 1/n2))

# Z-score
z = (p1 - p2) / se

# Two-tailed p-value
p_value = 2 * (1 - norm.cdf(abs(z)))

# Significance level
alpha = 0.05

# Output
print(f"Proportion A (p1): {p1:.3f} ({x1} out of {n1})")
print(f"Proportion B (p2): {p2:.3f} ({x2} out of {n2})")
print(f"Pooled proportion: {p_pool:.3f}")
print(f"Standard error: {se:.4f}")
print(f"Z-score: {z:.3f}")
print(f"P-value: {p_value:.4f}")

# Decision
if p_value < alpha:
    print(f"Reject the null hypothesis: There is a significant difference between the proportions (p-value {p_value:.4f} < {alpha}).")
    print(f"Version {'A' if p1 > p2 else 'B'} has a {'higher' if p1 > p2 else 'lower'} signup rate.")
else:
    print(f"Fail to reject the null hypothesis: No significant difference between the proportions (p-value {p_value:.4f} ≥ {alpha}).")
```

## One-sample T-test

A One-sample T-test is a statistical test that compares the mean of a single group to a known value (such as a population mean, a standard, or a target value). It helps to check if the average of your sample is significantly different from the known value or if the difference could have happened by chance.

**How Does It Work?**
    
1. **Null Hypothesis (H₀):**
    
    The mean of the sample is equal to the known value.

## Paired T-test

A Paired T-test is used when you have two related groups (or measurements from the same group) and you want to compare the means before and after a treatment or intervention. The groups are paired because each observation in the first group has a corresponding observation in the second group (i.e., they are not independent).

In simple terms, the Paired T-test checks if the average difference between the two measurements (before and after) is significantly different from zero.

**How Does It Work?**

1. **Null Hypothesis (H₀):**

    The mean difference between the paired values is zero (no effect).
    Example: The average score before and after a training session is the same (no improvement).

2. **Alternative Hypothesis (H₁):**

    The mean difference between the paired values is not zero (there is a significant effect).
    Example: The average score before and after the training session is different (there was an improvement).

3. **Test Statistic:**
    
    You calculate the t-statistic (the difference between the paired values) and compare it to the t-distribution.

4. **Decision Rule:**

    - If the p-value (probability) is smaller than the significance level (usually 0.05), you reject the null hypothesis (there is a significant difference).
    - If the p-value is larger, you fail to reject the null hypothesis (no significant difference).

    - Alternative Hypothesis (H₁): The mean difference is not 0 (there is an improvement).

    The formula for the t-statistic in a Paired T-test is:



6. **Determine the p-value:**

    Using a t-distribution table or statistical tool, you calculate the p-value corresponding to the t-statistic. This p-value tells you the likelihood of observing the results by random chance.

7. **Interpret the Result:**
    
    - If the p-value is less than 0.05, you reject the null hypothesis, meaning the training had a significant effect on the test scores (there was improvement).

    - If the p-value is greater than 0.05, you fail to reject the null hypothesis, meaning the training did not have a significant effect.

Let’s assume you are comparing the before and after scores of students who took a training program. You want to check if there is a significant difference in scores after the training.

# ANOVA (Analysis of Variance)

ANOVA (Analysis of Variance) is a statistical test used to compare the means (averages) of three or more groups to see if at least one group is significantly different from the others.

💡 It answers the question: “Do all these groups have the same average value, or is at least one different?”

**🧠 Why not just use multiple t-tests?**
A t-test is used to compare the means of two groups. If you have more than two groups and you perform many t-tests, you increase the risk of getting wrong results (Type I error).

ANOVA handles this better by testing all groups at once with a single test.


**Scenario:** A teacher wants to know if students in three different teaching methods perform differently in exams.

- Group A: Traditional classroom
- Group B: Online learning
- Group C: Hybrid (online + classroom)

The teacher collects the exam scores of students from each group.

🎯 Question: Is there a significant difference in the average exam scores among the three groups?

This is where ANOVA is used.

**🔢 How it works (In simple steps):**

1. Calculate the average score for each group.
2. Compare:
    - Variance between groups (how far apart group means are)
    - Variance within groups (how much students in the same group differ from each other)
3. If the between-group variance is large compared to the within-group variance, ANOVA concludes that at least one group is significantly different.


**🧮 Hypotheses:**
- Null Hypothesis (H₀): All group means are equal (no difference).
- Alternative Hypothesis (H₁): At least one group mean is different.

**✅ ANOVA Output:**
    
You get a p-value:
- If p-value < 0.05, reject the null hypothesis (means at least one group is different) or there is a significant difference in means.
- If p-value ≥ 0.05, you fail to reject the null hypothesis (means no significant difference) or the differences are likely due to random chance..


**If ANOVA finds a significant difference, you can do post-hoc tests (like Tukey's test) to find which specific groups differ.**

<br>

| Type                        | When to Use                                                     |
| --------------------------- | --------------------------------------------------------------- |
| **One-Way ANOVA**           | One independent variable with 3+ groups (e.g., teaching method) |
| **Two-Way ANOVA**           | Two independent variables (e.g., teaching method and gender)    |
| **Repeated Measures ANOVA** | Same subjects tested under different conditions/times (e.g., before/after treatment)           |


## One-Way ANOVA
One-Way ANOVA is a statistical test used to compare the means of 3 or more groups based on one factor (independent variable).

**💡 It helps answer:** "Is there a significant difference between the group means?"

“One-Way” means we are testing the effect of one independent variable (like "Teaching Method", "Diet Type", "Marketing Strategy", etc.) on a numerical outcome (like "Scores", "Weight", "Sales", etc.).

**🧪 When to Use One-Way ANOVA:**
                                                                                                                                                                
Use it when:
- You have one categorical independent variable (e.g., diet plan)
- You have one numeric dependent variable (e.g., weight loss)
- You want to compare 3 or more groups

**❗ Assumptions of One-Way ANOVA:**
1. Independence of observations
2. Normal distribution within groups
3. Equal variance (homogeneity of variance) across groups*  
4. Groups are independent


### Example

Scenario: A fitness coach wants to test if different diet plans lead to different weight loss results. Groups:
- Group A: Keto diet
- Group B: Low-carb diet
- Group C: Vegetarian diet

He records the weight loss (in kg) of people from each group after 1 month.

| Diet Plan      | Weight Loss (kg)   |
| -------------- | ------------------ |
| Keto (A)       | 4.5, 5.0, 4.2, 4.9 |
| Low-carb (B)   | 3.8, 4.0, 3.5, 3.9 |
| Vegetarian (C) | 2.5, 2.8, 2.6, 2.9 |

Now the coach wants to know: Is the average weight loss different across these 3 diet plans?

Use One-Way ANOVA!

**🧠 Hypotheses:**
- Null Hypothesis (H₀): All group means are equal (no difference).
- Alternative Hypothesis (H₁): At least one group mean is different.

**🧮 What does One-Way ANOVA do?**
    
1. Calculates mean weight loss in each group.
2. Compares:
    - Variance between groups (how much the group averages differ)
    - Variance within groups (how much individuals in the same group differ)
3. Computes a test statistic (F-value) and a p-value.

**🧪 Interpreting the Result:**
- If p-value < 0.05, we reject the null hypothesis → at least one group mean is significantly different.
- If p-value ≥ 0.05, we fail to reject the null hypothesis → no significant difference.

**✅ Example Result:**
    
    After running One-Way ANOVA, suppose the p-value = 0.004
    
        - 0.004 < 0.05 → Reject H₀
        - ✅ There is a significant difference in average weight loss between the diet plans.
    
**To know which diet plans differ, you can do a post-hoc test like Tukey's HSD.**


<br>


**⚙️ How One-Way ANOVA Works (Simply):**
1. Group Means: It calculates the average score for each group.
2. Overall Mean: It calculates the overall average score of all groups combined.
3. Between-group variation: Measures how different the group means are from the overall mean.
4. Within-group variation: Measures how spread out the scores are inside each group.
5. F-statistic: Compares between-group to within-group variation.
6. p-value: Tells if the observed differences in means are statistically significant.

### Post-hoc Test
A post-hoc test is done after ANOVA when the result is significant. It compares each pair of groups to find out which ones are different.

Think of ANOVA as telling you “Something is different.”
A post-hoc test tells you “Exactly where the difference is.”

**🧪 Most Common Post-hoc Test: Tukey’s HSD**

HSD = Honestly Significant Difference

Tukey’s HSD compares all possible pairs of group means and tells you:
- Whether the difference is statistically significant
- How big that difference is

**📊 Let's Continue with Your Example:**

Groups:
- Group A: Keto diet → 4.5, 5.0, 4.2, 4.9
- Group B: Low-carb diet → 3.8, 4.0, 3.5, 3.9
- Group C: Vegetarian diet → 2.5, 2.8, 2.6, 2.9

**ANOVA Result:**
p-value = 0.004 → At least one group is different

**✅ Use Tukey's HSD:**

    Tukey’s test will:
    
        - Compare A vs B
        - Compare A vs C
        - Compare B vs C

    And for each pair, it will say:
    
        - Is the difference significant?
        - What is the p-value?
        - What is the mean difference?



```python
import pandas as pd
from statsmodels.stats.multicomp import pairwise_tukeyhsd

# Data
data = {
    'WeightLoss': [4.5, 5.0, 4.2, 4.9, 3.8, 4.0, 3.5, 3.9, 2.5, 2.8, 2.6, 2.9],
    'Diet': ['Keto']*4 + ['Low-carb']*4 + ['Vegetarian']*4
}
df = pd.DataFrame(data)

# Tukey HSD
tukey = pairwise_tukeyhsd(endog=df['WeightLoss'], groups=df['Diet'], alpha=0.05)
print(tukey)
```

**✅ When to Use Tukey’s HSD (or other post-hoc tests)**

| Situation                                                             | Use Post-hoc Test?                                                                   |
| --------------------------------------------------------------------- | ------------------------------------------------------------------------------------ |
| One-Way ANOVA result is significant                                   | ✅ Yes, to find which groups differ                                                   |
| Two-Way ANOVA main effect (e.g., 'Diet' or 'Exercise') is significant | ✅ Yes, to compare specific levels within that factor                                 |
| Interaction effect is significant                                     | ✅ Yes, but you may need a more advanced post-hoc test (like simple effects or plots) |
| No significant result in ANOVA                                        | ❌ No need — post-hoc not valid or meaningful                                         |


**💡 Think of it like this:**
- ANOVA = "Is there a difference somewhere?"
- Post-hoc (Tukey) = "Okay, now tell me exactly where the differences are."

**If at least one of the main effects is significant in an ANOVA test (either One-Way or Two-Way), then you can and should use a post-hoc test like Tukey's HSD — but only for that significant effect.**

**🔁 Let's break it down simply:**

📌 Suppose you run a Two-Way ANOVA with:
- Factor A = Diet (e.g., Keto, Vegetarian, etc.)
- Factor B = Exercise (e.g., Cardio, Weight Training)
- Interaction = Diet × Exercise

And you get this result:

| Effect        | p-value | Significant? | Use Tukey HSD?  |
| ------------- | ------- | ------------ | --------------- |
| Diet          | 0.002   | ✅ Yes        | ✅ Yes (on Diet) |
| Exercise      | 0.08    | ❌ No         | ❌ No            |
| Diet×Exercise | 0.12    | ❌ No         | ❌ Not directly  |

So:
- You only apply Tukey HSD on the Diet factor, because it's the only one with p < 0.05.
- You do not apply Tukey HSD to Exercise or the interaction unless they are also significant.

**🧠 Why this rule?**
    
Tukey’s HSD (and other post-hoc tests) perform multiple pairwise comparisons — which increase the risk of false positives (Type I error).

So, we only do them if the ANOVA result says there’s something worth exploring.

**✅ Final Rule (General):**
✔️ Use Tukey HSD for pairwise comparisons only when the overall ANOVA result for that factor is significant (p < 0.05).

**In simple words:** Only use Tukey’s HSD to compare individual group pairs (like Keto vs Vegetarian) if the ANOVA tells you that the overall group differences are significant.

**🔍 Think of it like this:**
                                                                           
Imagine you're a teacher checking if students from 3 different classes (A, B, and C) performed differently in an exam.

- 🔸 Step 1: Run an ANOVA
                                                                           
    You ask:
        - “Is there any difference in average scores between classes A, B, and C?”
                                                                               
    This gives you one p-value.
    - If p < 0.05 ➝ It means “Yes, there is at least one class that’s different.”
    - If p ≥ 0.05 ➝ It means “No strong evidence that classes are different.”

**🟢 IF the ANOVA p-value is < 0.05**
                                                                           
✅ Now you’re allowed to look deeper and ask: “Which specific classes are different from each other?”

So you run Tukey’s HSD to do:
- A vs B
- A vs C
- B vs C

This is called pairwise comparison.

**🔴 IF the ANOVA p-value is ≥ 0.05**
                                                                           
❌ Then it means:
- “There's no strong evidence that group means differ overall.”

So it doesn’t make sense to compare individual pairs — because there's no signal to begin with.

✅ Final Rule Simplified:

| ANOVA Result (p-value) | Can You Use Tukey HSD? | Why?                          |
| ---------------------- | ---------------------- | ----------------------------- |
| p < 0.05               | ✅ Yes                  | Significant difference exists |
| p ≥ 0.05               | ❌ No                   | No evidence of any difference |


**🧪 Example:**
You have 3 diets: Keto, Vegetarian, Paleo
And their average weight losses are different.

You run a One-Way ANOVA:
- Result: p = 0.004 ✅ (Significant)

So now, you’re allowed to use Tukey HSD to compare:
- Keto vs Vegetarian
- Keto vs Paleo
- Vegetarian vs Paleo

To find which pairs differ.

### Python Implementation: One-way ANOVA + Tukey

We’ll use a diet plan example, where we compare the weight loss of people following three different diets.

```python
import pandas as pd
from scipy import stats
from statsmodels.stats.multicomp import pairwise_tukeyhsd

# Sample weight loss data for 3 diet plans
data = {
    'Diet': ['Keto'] * 5 + ['Vegetarian'] * 5 + ['Paleo'] * 5,
    'Weight_Loss': [5.1, 4.8, 5.5, 5.0, 4.9,   # Keto
                    3.2, 3.8, 3.5, 3.6, 3.7,   # Vegetarian
                    4.0, 4.3, 4.1, 4.2, 4.5]   # Paleo
}

df = pd.DataFrame(data)

# 🎯 One-Way ANOVA test
grouped = df.groupby('Diet')['Weight_Loss'].apply(list)
f_stat, p_val = stats.f_oneway(*grouped)
print("🔍 One-Way ANOVA Test Result:")
print(f"F-statistic: {f_stat:.3f}, p-value: {p_val:.4f}")

# ➕ Post-hoc test (Tukey's HSD) — only if ANOVA is significant
if p_val < 0.05:
    print("\n✅ ANOVA is significant (p < 0.05). Running Tukey's HSD test...")
    tukey = pairwise_tukeyhsd(endog=df['Weight_Loss'], groups=df['Diet'], alpha=0.05)
    print(tukey)
else:
    print("\n❌ ANOVA is not significant. No need for Tukey's test.")
```

**📌 Interpretation**

- The p-value is 0.0000, which is less than 0.05.
- ✅ This means: There is a statistically significant difference in average weight loss between at least two diet groups.

This tells us which specific pairs of diet plans differ. 🧾 Output Table Summary:

✅ Interpretation of each row:

    - Mean difference = -0.84: Keto leads to 0.84 kg more weight loss than Paleo (on average).
    - p-value = 0.0003 → Significant
    - ✅ So, Keto and Paleo are statistically different.

2. Keto vs Vegetarian
    - Mean difference = -1.50: Keto leads to 1.5 kg more weight loss than Vegetarian.
    - p-value = 0.0000 → Highly significant

3. Paleo vs Vegetarian

## Two-Way ANOVA

Two-Way ANOVA is used to compare the effects of TWO different factors (independent variables) on ONE outcome (dependent variable).

It also checks whether the two factors interact with each other to affect the outcome.

**🔍 When to Use Two-Way ANOVA?**

Use it when:

- You have two categorical independent variables (e.g., diet type and exercise type)
- You have one numerical dependent variable (e.g., weight loss, exam score, growth rate)

### Example

Scenario:
A fitness coach wants to test how diet and exercise affect weight loss.

**Two factors:**
- Diet Type: Keto, Vegetarian
- Exercise Type: Cardio, Weight Training

**Outcome:**
- Weight loss (in kg) after 1 month

**📊 Sample Data:**

| Diet       | Exercise        | Weight Loss (kg) |
| ---------- | --------------- | ---------------- |
| Keto       | Cardio          | 5.1              |
| Keto       | Weight Training | 4.9              |
| Vegetarian | Cardio          | 3.0              |
| Vegetarian | Weight Training | 3.8              |
| ...        | ...             | ...              |

(Assume we collect several weight loss records for each diet & exercise combo.)

**🧠 What Does Two-Way ANOVA Test?**
It answers 3 questions:

1. Main Effect of Diet
    - → Does the type of diet alone affect weight loss?

2. Main Effect of Exercise
    - → Does the type of exercise alone affect weight loss?

3. Interaction Effect
    - → Do diet and exercise together have a combined effect?

Example: Maybe Keto + Cardio works much better together than you'd expect from either one alone.

**📋 Hypotheses**
    
For Diet:
- H₀: No difference in weight loss between different diets
- H₁: At least one diet leads to different weight loss

For Exercise:
- H₀: No difference in weight loss between exercise types
- H₁: At least one exercise type leads to different weight loss

For Interaction:
- H₀: Diet and Exercise do not interact
- H₁: There is an interaction between Diet and Exercise



```python
import pandas as pd
import statsmodels.api as sm
from statsmodels.formula.api import ols

# Sample data
data = pd.DataFrame({
    'WeightLoss': [5.1, 4.9, 5.3, 3.0, 3.2, 3.5, 3.8, 3.9],
    'Diet': ['Keto', 'Keto', 'Keto', 'Vegetarian', 'Vegetarian', 'Vegetarian', 'Vegetarian', 'Vegetarian'],
    'Exercise': ['Cardio', 'Weight', 'Cardio', 'Cardio', 'Cardio', 'Weight', 'Weight', 'Weight']
})

# Two-Way ANOVA
model = ols('WeightLoss ~ C(Diet) + C(Exercise) + C(Diet):C(Exercise)', data=data).fit()
anova_table = sm.stats.anova_lm(model, typ=2)
print(anova_table)
```

**🧾 Output (Interpretation):**

| Source         | p-value | Interpretation                      |
| -------------- | ------- | ----------------------------------- |
| C(Diet)        | 0.003   | ✅ Diet has a significant effect     |
| C(Exercise)    | 0.041   | ✅ Exercise has a significant effect |
| Diet\:Exercise | 0.120   | ❌ No significant interaction        |

**Meaning:**
- Both diet and exercise individually affect weight loss.
- No interaction effect → The combination doesn't change results beyond their individual effects.

**✅ In Simple Words:**
    
Two-Way ANOVA helps you answer:
- "Do both diet and exercise affect weight loss?
- And do some combinations (like Keto + Cardio) work better together than others?"



**✅ When to Use Tukey’s HSD (or other post-hoc tests)**

| Situation                                                             | Use Post-hoc Test?                                                                   |
| --------------------------------------------------------------------- | ------------------------------------------------------------------------------------ |
| One-Way ANOVA result is significant                                   | ✅ Yes, to find which groups differ                                                   |
| Two-Way ANOVA main effect (e.g., 'Diet' or 'Exercise') is significant | ✅ Yes, to compare specific levels within that factor                                 |
| Interaction effect is significant                                     | ✅ Yes, but you may need a more advanced post-hoc test (like simple effects or plots) |
| No significant result in ANOVA                                        | ❌ No need — post-hoc not valid or meaningful                                         |


### Tukey’s HSD after Two-Way ANOVA

Yes — we need Tukey's HSD after ANOVA but only if at least one of the main effects is significant, and you want to know which specific groups are different.

**🔍 Explanation:**
Two-Way ANOVA tells you whether there is a significant effect, but it doesn’t tell you where the difference lies.

For example:
    
From the ANOVA output:
- Diet effect: p = 0.003 → Significant ✅
- Exercise effect: p = 0.041 → Significant ✅
- Interaction effect: p = 0.120 → Not significant ❌

So now we know:
- Diet types are different, but which ones?
- Exercise types are different, but which ones?

This is where Tukey's HSD comes in — it performs pairwise comparisons for:
- Different diet groups (e.g., Keto vs Vegetarian)
- Different exercise groups (e.g., Cardio vs Weight Training)

**📊 So, What to Do Next?**
    
Use Tukey’s HSD separately on each significant factor:
- If Diet is significant → Apply Tukey HSD to Diet groups.
- If Exercise is significant → Apply Tukey HSD to Exercise groups.

You do not apply Tukey HSD to the interaction term (like “Keto + Cardio vs Vegetarian + Weight”), unless you restructure your data for that purpose.


```python
from statsmodels.stats.multicomp import pairwise_tukeyhsd

# Tukey test on Diet
tukey_diet = pairwise_tukeyhsd(endog=data['WeightLoss'], groups=data['Diet'], alpha=0.05)
print(tukey_diet)
```

### Python Implementation: 

We want to check how Diet Type and Exercise Level affect Weight Loss.

- Factor A = Diet (Keto, Vegetarian, Paleo)
- Factor B = Exercise (Low, High)
- Dependent Variable = Weight Loss

```python
import pandas as pd
import statsmodels.api as sm
from statsmodels.formula.api import ols
from statsmodels.stats.multicomp import pairwise_tukeyhsd

# Sample data
data = {
    'Diet':       ['Keto', 'Keto', 'Keto', 'Keto', 'Keto', 'Keto',
                   'Vegetarian', 'Vegetarian', 'Vegetarian', 'Vegetarian', 'Vegetarian', 'Vegetarian',
                   'Paleo', 'Paleo', 'Paleo', 'Paleo', 'Paleo', 'Paleo'],
    
    'Exercise':   ['Low', 'Low', 'High', 'High', 'Low', 'High',
                   'Low', 'Low', 'High', 'High', 'Low', 'High',
                   'Low', 'Low', 'High', 'High', 'Low', 'High'],
    
    'Weight_Loss': [4.8, 5.0, 5.5, 5.2, 4.9, 5.3,
                    3.3, 3.4, 3.9, 4.1, 3.5, 4.0,
                    4.1, 4.0, 4.3, 4.2, 4.2, 4.4]
}

df = pd.DataFrame(data)

# Perform Two-Way ANOVA
model = ols('Weight_Loss ~ C(Diet) + C(Exercise) + C(Diet):C(Exercise)', data=df).fit()
anova_table = sm.stats.anova_lm(model, typ=2)
print("🔍 Two-Way ANOVA Result:\n")
print(anova_table)


# If Diet effect is significant (p < 0.05), run Tukey HSD for Diet
if anova_table.loc["C(Diet)", "PR(>F)"] < 0.05:
    print("\n✅ Diet has significant effect. Running Tukey HSD for Diet...\n")
    tukey_diet = pairwise_tukeyhsd(endog=df['Weight_Loss'], groups=df['Diet'], alpha=0.05)
    print(tukey_diet)
    
# If Exercise effect is significant (p < 0.05), run Tukey HSD for Exercise
if anova_table.loc["C(Exercise)", "PR(>F)"] < 0.05:
    print("\n✅ Exercise has significant effect. Running Tukey HSD for Exercise...\n")
    tukey_exercise = pairwise_tukeyhsd(endog=df['Weight_Loss'], groups=df['Exercise'], alpha=0.05)
    print(tukey_exercise)

# If interaction effect is significant, more advanced post-hoc is needed
if anova_table.loc["C(Diet):C(Exercise)", "PR(>F)"] < 0.05:
    print("\n⚠️ Interaction effect is significant. You may need separate post-hoc comparisons for each combination of Diet & Exercise.")
```

📌 You cannot fully interpret the main effects in isolation anymore.
                                                             
You now need to look at the combinations of Diet and Exercise.

**🧪 What to do next?**
Since interaction is significant:

- ✅ You should run Tukey HSD on combined groups (e.g., 'Keto-High', 'Keto-Low', 'Paleo-High', etc.) to find which combinations differ.

<br>

 Python code to perform Tukey's HSD test on the combinations of Diet and Exercise — since the interaction effect is significant, this is the correct next step.

```python
import pandas as pd
import statsmodels.api as sm
from statsmodels.formula.api import ols
from statsmodels.stats.multicomp import pairwise_tukeyhsd

# Sample DataFrame
data = pd.DataFrame({
    'Diet': ['Keto', 'Keto', 'Keto', 'Keto', 'Keto', 'Keto',
             'Paleo', 'Paleo', 'Paleo', 'Paleo', 'Paleo', 'Paleo',
             'Vegetarian', 'Vegetarian', 'Vegetarian', 'Vegetarian', 'Vegetarian', 'Vegetarian'],
    'Exercise': ['High', 'High', 'Low', 'Low', 'High', 'Low',
                 'High', 'High', 'Low', 'Low', 'High', 'Low',
                 'High', 'High', 'Low', 'Low', 'High', 'Low'],
    'WeightLoss': [4.5, 4.7, 4.0, 3.9, 4.6, 4.1,
                   3.8, 3.9, 3.3, 3.4, 3.9, 3.5,
                   3.2, 3.1, 2.6, 2.7, 3.0, 2.5]
})

# Create a combined group column
data['Group'] = data['Diet'] + '-' + data['Exercise']

# Run Tukey's HSD on the combined groups
tukey = pairwise_tukeyhsd(endog=data['WeightLoss'],
                          groups=data['Group'],
                          alpha=0.05)

# Print result
print("\n📊 Tukey HSD Test on Diet-Exercise Combinations:\n")
print(tukey.summary())
```

# A/B Testing

A/B testing is like a comparison experiment where you test two versions of something to see which one performs better.

- A = Version A (the original, also called control)
- B = Version B (the new version or variant)

You split your users into two groups randomly:
- One group sees Version A
- The other group sees Version B

Then you measure a specific outcome (called a metric, like clicks, purchases, or signups) to see which version is better.

**🧠 In Simple Terms:**
A/B testing is like having two recipes for a cake. You bake both, give each to 100 people, and see which one more people like. Then, you start using the winning recipe!




**💡 Why Use A/B Testing?**

To make data-driven decisions rather than guessing.
Instead of asking "Which one feels better?", you ask, "Which one actually performs better in the real world?"

**How A/B Testing Works:**
1. Formulate a Hypothesis: You need to decide what you want to test. For example, "Will changing the color of the 'Buy Now' button from green to red increase the purchase rate?"
2. Create Variants:
    - A: The original version (control group).
    - B: The modified version (variant group).
3. Random Assignment: Randomly assign users to each group so that any difference in results can be attributed to the change, not to the selection of participants.
4. Collect Data: Track the key performance indicators (KPIs) for both groups. For example, conversion rate, click-through rate, or sales.
5. Analyze the Results: Use statistical tests like the `T-test` or `Z-test` to determine whether the difference between the two groups is statistically significant.


## Example

### Website Button Color

Suppose you want more people to click the “Buy Now” button on your website.

- Version A (Control): Red button
- Version B (Variant): Green button

After running the test on 1,000 users:- 
- Red button gets 50 clicks (5% conversion)
- Green button gets 80 clicks (8% conversion)

➡️ You choose green, because it performs better.



### Email Subject Lines

You want more people to open your marketing emails.

- Version A: Subject line - “Limited Time Offer!”
- Version B: Subject line - “Get 20% Off Today Only!”

You send:
- A to 5,000 people → 500 open (10% open rate)
- B to 5,000 people → 700 open (14% open rate)

➡️ Version B wins.

### App Design Changes

You redesign a mobile app homepage to make it simpler.

- Group A: sees the current version
- Group B: sees the new design

You track how many users complete a signup:
- A: 1,000 users → 300 signups (30%)
- B: 1,000 users → 380 signups (38%)

➡️ New design improves signups, so you launch it for everyone.

## Python Implementation

```python
from statsmodels.stats.proportion import proportions_ztest, proportion_effectsize
from statsmodels.stats.power import zt_ind_solve_power
import matplotlib.pyplot as plt
import math

# === 1. Define test parameters ===
# Baseline and expected conversion rates
baseline_rate = 0.20  # current open/conversion rate (e.g., 20%)
target_rate = 0.22    # expected rate with new variant

# Statistical settings
alpha = 0.05          # significance level (5%)
power = 0.80          # power (80%)
alternative = 'two-sided'

# === 2. Calculate required sample size ===
effect_size = proportion_effectsize(baseline_rate, target_rate)
sample_size = math.ceil(zt_ind_solve_power(effect_size=effect_size, alpha=alpha, power=power, alternative=alternative))

print(f"📏 Required sample size per group: {sample_size}")
print(f"👥 Total users needed: {sample_size * 2}\n")

# === 3. Collect observed data (Example) ===
# Simulated or real observed data after running experiment
success_a = 800   # opens/conversions in control group
size_a = 4000     # total users in control group

success_b = 880   # opens/conversions in variant group
size_b = 4000     # total users in variant group

# === 4. Perform z-test for proportions ===
successes = [success_a, success_b]
samples = [size_a, size_b]

z_stat, p_value = proportions_ztest(count=successes, nobs=samples, alternative=alternative)

print(f"🧪 Z-statistic: {z_stat:.4f}")
print(f"📊 P-value: {p_value:.4f}")

if p_value < alpha:
    print("✅ Statistically significant difference (Reject null hypothesis)")
else:
    print("❌ No statistically significant difference (Fail to reject null hypothesis)")

# === 5. Visualize conversion rates ===
conversion_rates = [success_a / size_a, success_b / size_b]
labels = ['Control (A)', 'Variant (B)']
colors = ['blue', 'green']

plt.bar(labels, conversion_rates, color=colors)
plt.ylabel('Conversion Rate')
plt.title('A/B Test Result: Conversion Rates')
for i, rate in enumerate(conversion_rates):
    plt.text(i, rate + 0.005, f"{rate:.2%}", ha='center', fontsize=12)
plt.ylim(0, max(conversion_rates) + 0.05)
plt.show()
```

**When comparing revenue per user between two groups in an A/B test, you're dealing with numeric data (continuous), so you'll use a t-test instead of a z-test.**

Let’s say you tested two product pages:
- Group A (Control): 1000 users, avg revenue per user = ₹48.3
- Group B (Variant): 1000 users, avg revenue per user = ₹52.1

You want to know:
Is the difference in average revenue statistically significant?


```python
from scipy.stats import ttest_ind
import numpy as np
import matplotlib.pyplot as plt

# === 1. Sample Data (Simulated or Real) ===
# Revenue per user (these can be real data or simulated)

# Simulated revenue data for 1000 users in each group
np.random.seed(42)
group_a_revenue = np.random.normal(loc=48.3, scale=10.0, size=1000)
group_b_revenue = np.random.normal(loc=52.1, scale=10.0, size=1000)

# === 2. Run t-test ===
t_stat, p_value = ttest_ind(group_a_revenue, group_b_revenue, equal_var=False)  # Welch’s t-test

# === 3. Output results ===
print(f"📊 Mean Revenue - Group A: ₹{group_a_revenue.mean():.2f}")
print(f"📊 Mean Revenue - Group B: ₹{group_b_revenue.mean():.2f}")
print(f"🧪 T-statistic: {t_stat:.4f}")
print(f"📉 P-value: {p_value:.4f}")

alpha = 0.05
if p_value < alpha:
    print("✅ Statistically significant difference in revenue (Reject null hypothesis)")
else:
    print("❌ No statistically significant difference (Fail to reject null hypothesis)")

# === 4. Visualize ===
means = [group_a_revenue.mean(), group_b_revenue.mean()]
labels = ['Group A (Control)', 'Group B (Variant)']
colors = ['blue', 'green']

plt.bar(labels, means, color=colors)
plt.ylabel('Average Revenue per User (₹)')
plt.title('A/B Test - Revenue Comparison')
for i, m in enumerate(means):
    plt.text(i, m + 0.5, f"₹{m:.2f}", ha='center', fontsize=12)
plt.show()
```

**When to Use Which Test?**

| Comparison Type       | Test to Use | Function            |
| --------------------- | ----------- | ------------------- |
| Conversion rate (%)   | **Z-test**  | `proportions_ztest` |
| Average revenue, time | **T-test**  | `ttest_ind`         |
