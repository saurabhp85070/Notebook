# CSS

## selectors

```python
p {
  text-align: center;
  color: red;
}

# <p> elements on the page will be center-aligned, with a red text color
```

```python
#para1 {
  text-align: center;
  color: red;
}

# this css rule below will be applied to the HTML element with id="para1"
```

```python
.center {
  text-align: center;
  color: red;
}

# this css rule below will be applied to the HTML element with class="center"
```

```python
p.center {
  text-align: center;
  color: red;
}

# only <p> elements with class="center" will be red and center-aligned
```

```python
* {
  text-align: center;
  color: blue;
}

# this css rule below will affect every HTML element on the page
```

```python
h1, h2, p {
  text-align: center;
  color: red;
}

# this css rule below will be applied to all HTML element with <h1>, <h2>, <p> tags
```

```python
nav ul {
  list-style: none;
}

# This rule removes the bullet points from unordered lists (<ul>) that are inside a <nav> element (navigation section).

<nav>
  <ul>
    <li>Home</li>
    <li>About</li>
  </ul>
</nav>
```

```python
div > p {
  color: blue;
}

# This CSS rule selects all <p> (paragraph) elements that are direct children of a <div> — and it sets their text color to blue.

<div>
  <p>This will be blue</p>         <!-- ✅ Direct child -->
  <section>
    <p>This will NOT be blue</p>   <!-- ❌ Not a direct child -->
  </section>
</div>
```

```python
h1 + p {
  font-weight: bold;
}

# This selects only the <p> tag that comes immediately after an <h1>, and makes its text bold.

<h1>Welcome</h1>
<p>This will be bold.</p>
<p>This will not be bold.</p>
```

```python
h1 ~ p {
  color: gray;
}

# This selects all <p> elements that come after an <h1> (not just the first one), and changes their text color

<h1>Welcome</h1>
<p>This will be gray.</p>
<p>This will also be gray.</p>
```

## Specificity & Cascade
- Inline styles > IDs > Classes/Attributes/Pseudo-classes > Elements
- More specific selectors override less specific ones.
- Later styles override earlier ones if specificity is equal.

```python
/* Element */
p {
  color: black;
}

/* Class - overrides the above */
p.highlight {
  color: green;
}

/* ID - overrides both */
#intro {
  color: red;
}
```

## box model

The CSS box model is essentially a box that wraps around every HTML element. It consists of: content, padding, borders and margins. The image below illustrates the box model:



- `Content` - The content of the box, where text and images appear
- `Padding` - Clears an area around the content. The padding is transparent
- `Border` - A border that goes around the padding and content
- `Margin` - Clears an area outside the border. The margin is transparent

```python
div {
  width: 320px;
  height: 50px;
  padding: 10px;
  border: 5px solid gray;
  margin: 0;
}

# Total element width = width + left padding + right padding + left border + right border

  320px (width of content area)
+ 20px (left padding + right padding)
+ 10px (left border + right border)
= 350px (total width)

# Total element height = height + top padding + bottom padding + top border + bottom border

  50px (height of content area)
+ 20px (top padding + bottom padding)
+ 10px (top border + bottom border)
= 80px (total height)
```

## block and inline elements

**Block elements:**

- A block-level element always starts on a new line, and the browsers automatically add some space (a margin) before and after the element.
- A block-level element always takes up the full width available (stretches out to the left and right as far as it can).

```python
# Here are the block-level elements in HTML:

<address> <article> <aside> <blockquote> <canvas> <dd> <div> <dl> <dt> <fieldset>
<figcaption> <figure> <footer> <form> <h1>-<h6> <header> <hr> <li> <main> <nav>
<noscript> <ol> <p> <pre> <section> <table> <tfoot> <ul> <video>
```

**Inline elements:**

- An inline element does not start on a new line.
- An inline element only takes up as much width as necessary.

```python
# Here are the inline elements in HTML:

<a> <abbr> <acronym> <b> <bdo> <big> <br> <button> <cite> <code> <dfn> <em> <i> <img>
<input> <kbd> <label> <map> <object> <output> <q> <samp> <script> <select> <small>
<span> <strong> <sub> <sup> <textarea> <time> <tt> <var>

# Note: An inline element cannot contain a block-level element!
```
